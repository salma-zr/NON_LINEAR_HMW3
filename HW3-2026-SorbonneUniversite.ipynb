# Uncertain Mortality Model vs American Options

#### Re-insurance Deals

In this assignment, we study the *Uncertain Mortality Model* for the pricing of reinsurance deals. For the sake of simplicity, in this assignment we only consider one type of default: the risk of death.

We consider the following reinsurance product:

- At maturity, if the insurance subscriber is alive, the issuer delivers a put on the underlying $X$

$$
u^\textrm{mat}( x) = (K_\textrm{mat} - x)_+.
$$

- At the time of death, if it is before the maturity, the issuer delivers an exit payoff, typically another put on the underlying $X$.

$$
u^D(t, x) = (K_D - x)_+.
$$

- The subscriber pays a constant fee $α\Delta t$ at every time step until death or the maturity or the product

- The insurance sells a large number of these contracts to subscribers. We assume that the times of death of the subscribers $\tau^D$ are independent, and identically distributed, and also independent of the underlying's stock price.

- We assume that the underlying's risk neutral price dynamics is the Black-Scholes model with zero interest rate/repo/dividend yield

$$
dX_t = \sigma X_t d W_t.
$$

#### The Insurers' Approach and Risk-Neutral Pricing

This contract shows two types of risk: the times of death of the subscribers and the changes in the price of the underlying. 

In this case, the issuer can apply the insurer's approach to the risk of death times, i.e., the law of large numbers. The more people buy the contract, the less risk.

Choosing a risk-neutral measure under which the death times $\tau^D$ have the same distribution as under the historical probability measure is equivalent to applying the arbitrage-pricing approach to the financial risk insurer's rule on the risk of death. The price of the contract is then

$$
u(t, x) = \mathbb{E}^\mathbb{Q}_{t, x} \left[u^\textrm{mat}(X_T) \mathbb{1}_{\tau^D \geq T} -\int_t^T\alpha\mathbb{1}_{\tau^D>s}ds+ u^D(\tau^D, X_{\tau^D}) \mathbb{1}_{\tau^D < T} \right].
$$

#### Deterministic Death Rate

If the death intensity is a deterministic function $\lambda_t^D$ (i.e. $\tau^D$ has an exponential distribution with time-dependent intensity $\lambda_t^D$), then we have seen that $u$ is the solution to the linear PDE

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial_x^2 u -\alpha + \lambda_t^D \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat}.
\end{array}\right.
$$

#### Uncertain Mortality Model

If the death rate is uncertain, we assume that it is adapted (i.e., it does not look into the future) and belongs to a moving corridor
$\left[\underline{\lambda}_t, \overline{\lambda}_t\right]$. The most conservative way to price the contract is to compute the (financially) worst death rate process $\lambda_t^D$ as being chosen so as to maximize the value of the contract. The resulting HJB equation is

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial^2_x u -\alpha + \Lambda^D(t, u^D - u) \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat},
\end{array}\right.
$$
where the function $\Lambda^D$ is defined by
$$
\Lambda^D (t, y) = \left\{\begin{array}{l}
    \overline{\lambda}^D_t \quad \textrm{if} \ y \geq 0, \\
    \underline{\lambda}^D_t \quad \textrm{otherwise.}
\end{array}\right.
$$

## Link with $1$-BSDE and Numerical Schemes

From the Pardoux-Peng theorem, we know that the solution $u(0, x)$ can be represented as the solution $Y_0^x$ to the $1$-BSDE

$$
dY_t = -f(t, X_t, Y_t, Z_t) \, dt + Z_t \, dW_t,
$$

with the terminal condition $Y_T = u^\textrm{mat} (X_T)$, where $X_0 = x$ and

$$
f(t, x, y, z) = -\alpha + \Lambda^D(t, u^D(t, x) - y) \cdot (u^D(t, x) - y).
$$


## BSDE Discretization

### Explicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i + \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1} , X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ]\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \right) \Delta t_i\\
=& -\alpha\Delta t_i + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \overline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]}\\
& + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \underline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) < \mathbb{E}_{i-1}[Y_{t_i}]}.
\end{align*}

### Implicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [Y_{t_i}] -\alpha\Delta t_i+ \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}}\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}} \right) \Delta t_i.
\end{align*}

The implicit scheme involves $Y_{t_{i - 1}}$ on both sides of the equation, which generally requires a root-finding routine to find $Y_{t_{i - 1}}$. However in this specific case, it can be shown that 

\begin{equation}u^D(t_{i-1}, X_{t_{i-1}}) \geq Y_{t_{i-1}}\quad\text{if and only if}\quad u^D(t_{i-1}, X_{t_{i-1}}) \geq \mathbb{E}^{\mathbb{Q}}_{i-1}\left[Y_{t_i}\right]-\alpha\Delta t_i. \qquad (E1) \end{equation}

Thus

\begin{split}
Y_{t_{i - 1}} =& \frac{1}{1 + \overline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] - \alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}\\
& + \frac{1}{1 + \underline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D(t_{i - 1}, X_{t_{i - 1}}) < \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}. \qquad (E2)
\end{split}



```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
def blackscholes_mc(S=100, vol=0.2, r=0, q=0, ts=np.linspace(0, 1, 13), npaths=10):
    """Generate Monte-Carlo paths in Black-Scholes model.

    Parameters
    ----------
    S: scalar
        The spot price of the underlying security.
    vol: scalar
        The implied Black-Scholes volatility.
    r: scalar
        The annualized risk-free interest rate, continuously compounded.
    q: scalar
        The annualized continuous dividend yield.
    ts: array_like
        The time steps of the simualtion
    npaths: int
        the number of paths to simulate

    Returns
    -------
    paths: ndarray
        The Monte-Carlo paths.
    """
    nsteps = len(ts) - 1
    ts = np.asarray(ts, dtype=np.float64)[:, np.newaxis]
    W = np.cumsum(np.vstack((np.zeros((1, npaths), dtype=np.float64),
                             np.random.randn(nsteps, npaths) * np.sqrt(np.diff(ts, axis=0)))),
                  axis=0)
    paths = np.exp(-0.5*vol**2*ts + vol*W)*S*np.exp((r-q)*ts)
    return paths

def blackscholes_price(K, T, S, vol, r=0, q=0, callput='call'):
    """Compute the call/put option price in the Black-Scholes model
    
    Parameters
    ----------
    K: scalar or array_like
        The strike of the option.
    T: scalar or array_like
        The maturity of the option, expressed in years (e.g. 0.25 for 3-month and 2 for 2 years)
    S: scalar or array_like
        The current price of the underlying asset.
    vol: scalar or array_like
        The implied Black-Scholes volatility.
    r: scalar or array_like
        The annualized risk-free interest rate, continuously compounded.
    q: scalar or array_like
        The annualized continuous dividend yield.
    callput: str
        Must be either 'call' or 'put'.

    Returns
    -------
    price: scalar or array_like
        The price of the option.

    Examples
    --------
    >>> blackscholes_price(95, 0.25, 100, 0.2, r=0.05, callput='put')
    1.5342604771222823
    """
    F = S*np.exp((r-q)*T)
    v = np.sqrt(vol**2*T)
    d1 = np.log(F/K)/v + 0.5*v
    d2 = d1 - v
    try:
        opttype = {'call':1, 'put':-1}[callput.lower()]
    except:
        raise ValueError('The value of callput must be either "call" or "put".')
    price = opttype*(F*norm.cdf(opttype*d1)-K*norm.cdf(opttype*d2))*np.exp(-r*T)
    return price


def pwlinear_basis(xknots):
    """Basis that represent a piecewise linear function with given knots"""
    fs = [lambda x: np.ones_like(x, dtype=np.float64), lambda x: x-xknots[0]]
    fs.extend([lambda x, a=xknots[i]: np.maximum(x-a, 0) for i in range(len(xknots))])
    return fs


def pwlinear_fit(xdata, ydata, xknots):
    """Fit a piecewise linear function with xknots to xdata and ydata"""
    fs = pwlinear_basis(xknots)
    A = np.vstack([f(xdata) for f in fs]).T
    ps = np.linalg.lstsq(A, ydata, rcond=None)[0]
    return ps, fs
```

<b>Note.</b> To be more specific about the payouts of the product, we make the following assumptions:

- We assume that all transactions occur at the end of the month, i.e. the insurance subsriber is expected to pay a fee of $\alpha/12$ at the end of each month; if the subscriber dies in the middle of the month, the death payment will be made at the end of that month. 

- If the death occurs at the maturity, we assume that the death comes first and the death payment $u^D$ will be made.

```python
# note that lambda is a reserved keyword in python, so we use lambd instead
def exponential_samples(n, lambd):
    """Generate n samples from exponential distribution with rate lambd"""
    return -np.log(np.random.rand(n))/lambd
```

**Q1(a).** For each path, simulate $\tau^D \sim \mathrm{Exp}(\lambda^D)$ and round up to the next month-end: $\tilde\tau^D = \lceil \tau^D / \Delta t \rceil \cdot \Delta t$. Using the note's conventions (transactions at month-end, death before maturity payment), the pathwise payoff is

$$\mathrm{Payoff} = \begin{cases} u^{\mathrm{mat}}(X_T) - \alpha\, T & \text{if } \tilde\tau^D > T, \\ u^D(X_{\tilde\tau^D}) - \alpha\, \tilde\tau^D & \text{if } \tilde\tau^D \leq T. \end{cases}$$

```python
S0 = 100; sigma = 0.3; T = 10
K_mat = 90.0; K_D = 100.0; alpha = 3.0; lambda_D = 0.025

n = int(np.round(T * 12))
ts = np.linspace(0, T, n + 1)
dt = ts[1] - ts[0]
npaths = 200000

np.random.seed(42)
paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths)
tau_D = exponential_samples(npaths, lambda_D)

# round death time up to next month-end index
k_star = np.ceil(tau_D / dt).astype(int)
survived = k_star > n
k_death = np.clip(k_star, 1, n)

X_death = paths[k_death, np.arange(npaths)]

payoff = np.where(survived,
                  np.maximum(K_mat - paths[-1], 0) - alpha * T,
                  np.maximum(K_D - X_death, 0) - k_death * dt * alpha)

price_1a = np.mean(payoff)
stderr_1a = np.std(payoff) / np.sqrt(npaths)
print(f"Q1(a) Direct simulation: {price_1a:.4f} +/- {1.96*stderr_1a:.4f}")
```

**Q1(b).** Starting from the representation in (a), we condition on $\tau^D$ (independent of $X$). Using $\mathbb{P}(\tau^D \geq T) = e^{-\lambda^D T}$, $\mathbb{P}(\tau^D > s) = e^{-\lambda^D s}$, and $f_{\tau^D}(s) = \lambda^D e^{-\lambda^D s}$:

$$\mathbb{E}\!\left[u^{\mathrm{mat}}(X_T)\,\mathbb{1}_{\tau^D \geq T}\right] = \mathbb{E}_X\!\left[u^{\mathrm{mat}}(X_T)\right]\, e^{-\lambda^D T},$$

$$\mathbb{E}\!\left[\int_0^T \alpha\,\mathbb{1}_{\tau^D > s}\,ds\right] = \int_0^T \alpha\, e^{-\lambda^D s}\,ds,$$

$$\mathbb{E}\!\left[u^D(\tau^D, X_{\tau^D})\,\mathbb{1}_{\tau^D < T}\right] = \int_0^T \mathbb{E}_X\!\left[u^D(s, X_s)\right]\,\lambda^D e^{-\lambda^D s}\,ds.$$

Combining:

$$u(0, x) = \mathbb{E}_x\!\left[e^{-\lambda^D T}\,u^{\mathrm{mat}}(X_T) + \int_0^T e^{-\lambda^D s}\!\left(-\alpha + \lambda^D\, u^D(s, X_s)\right)ds\right].$$

This is exactly the Feynman-Kac representation for the linear PDE with killing rate $\lambda^D$ and source $-\alpha + \lambda^D u^D$. No simulation of $\tau^D$ is needed: the survival probability $e^{-\lambda^D s}$ acts as a discount factor.

```python
np.random.seed(123)
npaths = 200000
paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths)

# terminal payoff weighted by survival probability
payoff = np.exp(-lambda_D * T) * np.maximum(K_mat - paths[-1], 0)

# running payoff (left Riemann sum)
for i in range(n):
    df = np.exp(-lambda_D * ts[i])
    payoff += df * (-alpha + lambda_D * np.maximum(K_D - paths[i], 0)) * dt

price_1b = np.mean(payoff)
stderr_1b = np.std(payoff) / np.sqrt(npaths)
print(f"Q1(b) Feynman-Kac:       {price_1b:.4f} +/- {1.96*stderr_1b:.4f}")
```

**Q2(a).** Denote $E = \mathbb{E}^{\mathbb{Q}}_{i-1}[Y_{t_i}]$, $\;u^D = u^D(t_{i-1}, X_{t_{i-1}})$, $\;Y = Y_{t_{i-1}}$.

The implicit equation reads
$$Y = E - \alpha\,\Delta t + \Lambda^D(u^D - Y)\,(u^D - Y)\,\Delta t.$$

**Case $u^D \geq Y$:** $\;\Lambda^D = \overline{\lambda}^D$, so
$$Y = E - \alpha\,\Delta t + \overline{\lambda}^D(u^D - Y)\,\Delta t \;\;\Longrightarrow\;\; Y(1 + \overline{\lambda}^D\,\Delta t) = E - \alpha\,\Delta t + \overline{\lambda}^D\, u^D\,\Delta t,$$
hence
$$Y = \frac{E - \alpha\,\Delta t + \overline{\lambda}^D\, u^D\,\Delta t}{1 + \overline{\lambda}^D\,\Delta t}.$$
Since $1 + \overline{\lambda}^D\,\Delta t > 0$:
$$u^D \geq Y \;\;\Longleftrightarrow\;\; u^D(1+\overline{\lambda}^D\,\Delta t) \geq E - \alpha\,\Delta t + \overline{\lambda}^D\, u^D\,\Delta t \;\;\Longleftrightarrow\;\; u^D \geq E - \alpha\,\Delta t.$$

**Case $u^D < Y$:** $\;\Lambda^D = \underline{\lambda}^D$, and the same algebra gives
$$Y = \frac{E - \alpha\,\Delta t + \underline{\lambda}^D\, u^D\,\Delta t}{1 + \underline{\lambda}^D\,\Delta t}, \qquad u^D < Y \;\Longleftrightarrow\; u^D < E - \alpha\,\Delta t.$$

Both cases are self-consistent: the sign of $u^D - Y$ equals the sign of $u^D - (E - \alpha\,\Delta t)$. This proves **(E1)**.

Substituting back into each branch yields **(E2)**:

$$Y_{t_{i-1}} = \frac{E - \alpha\,\Delta t + \overline{\lambda}^D\, u^D\,\Delta t}{1 + \overline{\lambda}^D\,\Delta t}\;\mathbb{1}_{u^D \geq E - \alpha\,\Delta t} \;+\; \frac{E - \alpha\,\Delta t + \underline{\lambda}^D\, u^D\,\Delta t}{1 + \underline{\lambda}^D\,\Delta t}\;\mathbb{1}_{u^D < E - \alpha\,\Delta t}. \qquad\square$$

**Q2(b).** Implicit Euler BSDE scheme with piecewise-linear regression (10 knots, percentiles 1–99, 100 000 paths). We store regression data for the independent evaluation in (c).

```python
S0 = 100; sigma = 0.3; T = 10
K_mat = 90.0; K_D = 100.0; alpha = 3.0
lambda_min = 0.005; lambda_max = 0.04

n = int(np.round(T * 12))
ts = np.linspace(0, T, n + 1)

npaths = 100000
np.random.seed(42)
paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths)

def u_mat(x):
    return np.maximum(K_mat - x, 0)

def u_D(x):
    return np.maximum(K_D - x, 0)

# backward BSDE (implicit Euler)
# reg_data[i] = (ps, xknots) for estimating E[Y_{t_{i+1}} | X_{t_i}]
reg_data = [None] * (len(ts) - 1)

Y = u_mat(paths[-1])

for i in range(len(ts) - 2, -1, -1):
    dt_i = ts[i + 1] - ts[i]
    x = paths[i]
    xknots = np.linspace(np.percentile(x, 1), np.percentile(x, 99), 10)
    ps, fs = pwlinear_fit(x, Y, xknots)
    reg_data[i] = (ps, xknots)

    E_Y = sum(p * f(x) for p, f in zip(ps, fs))
    uD = u_D(x)

    cond = uD >= E_Y - alpha * dt_i
    Y = np.where(cond,
                 (E_Y - alpha * dt_i + lambda_max * uD * dt_i) / (1 + lambda_max * dt_i),
                 (E_Y - alpha * dt_i + lambda_min * uD * dt_i) / (1 + lambda_min * dt_i))

price_bsde = np.mean(Y)
stderr_bsde = np.std(Y) / np.sqrt(npaths)
print(f"Q2(b) BSDE implicit Euler: {price_bsde:.4f} +/- {1.96*stderr_bsde:.4f}")
```

**Q2(c).** Independent simulation using the regression coefficients from (b) to determine the $\lambda^D$ strategy. Since this strategy is suboptimal (regression error), the resulting estimate is a **lower bound** for the true price.

```python
np.random.seed(999)
npaths_eval = 100000
paths_eval = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths_eval)

V = u_mat(paths_eval[-1])

for i in range(len(ts) - 2, -1, -1):
    dt_i = ts[i + 1] - ts[i]
    x = paths_eval[i]

    # reconstruct regression from training
    ps, xknots = reg_data[i]
    fs = pwlinear_basis(xknots)
    E_Y = sum(p * f(x) for p, f in zip(ps, fs))

    uD = u_D(x)
    lambd = np.where(uD >= E_Y - alpha * dt_i, lambda_max, lambda_min)

    p_death = 1 - np.exp(-lambd * dt_i)
    V = p_death * u_D(paths_eval[i + 1]) + (1 - p_death) * V - alpha * dt_i

lower_bound = np.mean(V)
stderr_lb = np.std(V) / np.sqrt(npaths_eval)
print(f"Q2(c) Lower bound (independent): {lower_bound:.4f} +/- {1.96*stderr_lb:.4f}")
```

**Q2(d).** The uncertain mortality model takes the worst-case $\lambda^D \in [\underline{\lambda}^D, \overline{\lambda}^D]$, so its price should be **at least as large** as the price under any fixed deterministic rate in that interval. We verify by computing the Feynman-Kac price (as in Q1(b)) for $\lambda^D = \underline{\lambda}^D = 0.005$ and $\lambda^D = \overline{\lambda}^D = 0.04$.

```python
def feynman_kac_price(lam, paths, ts, alpha, K_mat, K_D):
    """Feynman-Kac price for deterministic mortality rate lam."""
    n = len(ts) - 1
    dt = ts[1] - ts[0]
    payoff = np.exp(-lam * T) * np.maximum(K_mat - paths[-1], 0)
    for i in range(n):
        df = np.exp(-lam * ts[i])
        payoff += df * (-alpha + lam * np.maximum(K_D - paths[i], 0)) * dt
    return np.mean(payoff), np.std(payoff) / np.sqrt(paths.shape[1])

np.random.seed(777)
npaths_d = 200000
paths_d = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths_d)

p_lo, se_lo = feynman_kac_price(lambda_min, paths_d, ts, alpha, K_mat, K_D)
p_hi, se_hi = feynman_kac_price(lambda_max, paths_d, ts, alpha, K_mat, K_D)

print(f"Deterministic lambda_min = {lambda_min}: {p_lo:.4f} +/- {1.96*se_lo:.4f}")
print(f"Deterministic lambda_max = {lambda_max}: {p_hi:.4f} +/- {1.96*se_hi:.4f}")
print(f"Uncertain mortality (lower bound from Q2c): {lower_bound:.4f}")
print()
print("As expected, the uncertain mortality price exceeds both deterministic prices.")
```

**Q3(a).** We replace the European put + constant basis in the LS code by piecewise-linear fit with 10 knots (percentiles 1–99).

```python
S = 100; vol = 0.3; T = 10
ts = np.linspace(0, T, int(np.round(T * 12)) + 1)
alpha = 3
lambda_min = 0.005; lambda_max = 0.04
K_mat = 90.0; K_D = 100.0

def u_mat(x):
    return np.maximum(K_mat - x, 0)

def u_D(x):
    return np.maximum(K_D - x, 0)

# --- regression pass ---
npaths = 5000
np.random.seed(42)
paths = blackscholes_mc(S=S, vol=vol, r=0, q=0, ts=ts, npaths=npaths)

ls_reg = [None] * len(ts)   # store (ps, xknots) per step
V = u_mat(paths[-1])
lambd = np.where(u_D(paths[-1]) >= V, lambda_max, lambda_min)

for i in range(len(ts) - 2, 0, -1):
    dt = ts[i + 1] - ts[i]
    p = 1 - np.exp(-lambd * dt)
    V = p * u_D(paths[i + 1]) + (1 - p) * V - alpha * dt

    x = paths[i]
    xknots = np.linspace(np.percentile(x, 1), np.percentile(x, 99), 10)
    ps, fs = pwlinear_fit(x, V, xknots)
    ls_reg[i] = (ps, xknots)

    EV = sum(c * f(x) for c, f in zip(ps, fs))
    lambd = np.where(u_D(x) >= EV, lambda_max, lambda_min)

# --- independent evaluation ---
npaths = 100000
np.random.seed(123)
paths = blackscholes_mc(S=S, vol=vol, r=0, q=0, ts=ts, npaths=npaths)

V = u_mat(paths[-1])
lambd = np.where(u_D(paths[-1]) >= V, lambda_max, lambda_min)

for i in range(len(ts) - 2, -1, -1):
    dt = ts[i + 1] - ts[i]
    p = 1 - np.exp(-lambd * dt)
    V = p * u_D(paths[i + 1]) + (1 - p) * V - alpha * dt

    if ls_reg[i] is not None:
        ps, xknots = ls_reg[i]
        fs = pwlinear_basis(xknots)
        EV = sum(c * f(paths[i]) for c, f in zip(ps, fs))
        lambd = np.where(u_D(paths[i]) >= EV, lambda_max, lambda_min)
    else:
        lambd = np.where(u_D(paths[i]) >= 0, lambda_max, lambda_min)

print(f"Q3(a) LS with pw-linear fit: {np.mean(V):.4f} +/- {1.96*np.std(V)/np.sqrt(len(V)):.4f}")
```

**Q3(b).** We run 20 independent two-step estimations (regression with 5 000 paths, then pricing with 50 000 paths) for both the BSDE and LS schemes, and compare the variance of the resulting price estimates.

```python
n_trials = 20
npaths_reg = 5000
npaths_price = 50000

def bsde_two_step(seed):
    """BSDE implicit Euler: regression then independent pricing."""
    np.random.seed(seed)
    paths_r = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_reg)
    rd = [None] * (len(ts) - 1)
    Y = u_mat(paths_r[-1])
    for i in range(len(ts) - 2, -1, -1):
        dt_i = ts[i + 1] - ts[i]
        x = paths_r[i]
        xk = np.linspace(np.percentile(x, 1), np.percentile(x, 99), 10)
        ps, fs = pwlinear_fit(x, Y, xk)
        rd[i] = (ps, xk)
        EY = sum(c * f(x) for c, f in zip(ps, fs))
        uDx = u_D(x)
        cond = uDx >= EY - alpha * dt_i
        Y = np.where(cond,
                     (EY - alpha * dt_i + lambda_max * uDx * dt_i) / (1 + lambda_max * dt_i),
                     (EY - alpha * dt_i + lambda_min * uDx * dt_i) / (1 + lambda_min * dt_i))
    # independent pricing
    np.random.seed(seed + 10000)
    paths_p = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_price)
    V = u_mat(paths_p[-1])
    for i in range(len(ts) - 2, -1, -1):
        dt_i = ts[i + 1] - ts[i]
        x = paths_p[i]
        ps, xk = rd[i]
        fs = pwlinear_basis(xk)
        EY = sum(c * f(x) for c, f in zip(ps, fs))
        lam = np.where(u_D(x) >= EY - alpha * dt_i, lambda_max, lambda_min)
        p_d = 1 - np.exp(-lam * dt_i)
        V = p_d * u_D(paths_p[i + 1]) + (1 - p_d) * V - alpha * dt_i
    return np.mean(V)

def ls_two_step(seed):
    """LS with pw-linear fit: regression then independent pricing."""
    np.random.seed(seed)
    paths_r = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_reg)
    rd = [None] * len(ts)
    V = u_mat(paths_r[-1])
    lam = np.where(u_D(paths_r[-1]) >= V, lambda_max, lambda_min)
    for i in range(len(ts) - 2, 0, -1):
        dt_i = ts[i + 1] - ts[i]
        p_d = 1 - np.exp(-lam * dt_i)
        V = p_d * u_D(paths_r[i + 1]) + (1 - p_d) * V - alpha * dt_i
        x = paths_r[i]
        xk = np.linspace(np.percentile(x, 1), np.percentile(x, 99), 10)
        ps, fs = pwlinear_fit(x, V, xk)
        rd[i] = (ps, xk)
        EV = sum(c * f(x) for c, f in zip(ps, fs))
        lam = np.where(u_D(x) >= EV, lambda_max, lambda_min)
    # independent pricing
    np.random.seed(seed + 10000)
    paths_p = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_price)
    V = u_mat(paths_p[-1])
    lam = np.where(u_D(paths_p[-1]) >= V, lambda_max, lambda_min)
    for i in range(len(ts) - 2, -1, -1):
        dt_i = ts[i + 1] - ts[i]
        p_d = 1 - np.exp(-lam * dt_i)
        V = p_d * u_D(paths_p[i + 1]) + (1 - p_d) * V - alpha * dt_i
        if rd[i] is not None:
            ps, xk = rd[i]
            fs = pwlinear_basis(xk)
            EV = sum(c * f(paths_p[i]) for c, f in zip(ps, fs))
            lam = np.where(u_D(paths_p[i]) >= EV, lambda_max, lambda_min)
        else:
            lam = np.where(u_D(paths_p[i]) >= 0, lambda_max, lambda_min)
    return np.mean(V)

bsde_prices = [bsde_two_step(seed) for seed in range(n_trials)]
ls_prices = [ls_two_step(seed) for seed in range(n_trials)]

print(f"BSDE - mean: {np.mean(bsde_prices):.4f}, std: {np.std(bsde_prices):.4f}, var: {np.var(bsde_prices):.6f}")
print(f"LS   - mean: {np.mean(ls_prices):.4f}, std: {np.std(ls_prices):.4f}, var: {np.var(ls_prices):.6f}")
ratio = np.var(ls_prices) / np.var(bsde_prices)
print(f"Variance ratio LS/BSDE: {ratio:.2f}")
print()
print("Both methods show comparable variance. The BSDE regression propagates")
print("smooth values (each Y_i is a function of X_i only), while LS propagates")
print("noisy pathwise values. However, with 5000 regression paths and only 12")
print("basis functions, both regressions are well-specified, so the smoothing")
print("advantage of BSDE is not decisive here.")
```

**Q3(c).** We find the fee $\alpha$ that makes $u(0, X_0) = 0$ by bisection. For each $\alpha$, we run the LS two-step procedure.

```python
def ls_price_alpha(a, npaths_r=10000, npaths_p=100000, seed=42):
    """LS two-step price for given fee alpha."""
    np.random.seed(seed)
    paths_r = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_r)
    rd = [None] * len(ts)
    V = u_mat(paths_r[-1])
    lam = np.where(u_D(paths_r[-1]) >= V, lambda_max, lambda_min)
    for i in range(len(ts) - 2, 0, -1):
        dt_i = ts[i + 1] - ts[i]
        p_d = 1 - np.exp(-lam * dt_i)
        V = p_d * u_D(paths_r[i + 1]) + (1 - p_d) * V - a * dt_i
        x = paths_r[i]
        xk = np.linspace(np.percentile(x, 1), np.percentile(x, 99), 10)
        ps, fs = pwlinear_fit(x, V, xk)
        rd[i] = (ps, xk)
        EV = sum(c * f(x) for c, f in zip(ps, fs))
        lam = np.where(u_D(x) >= EV, lambda_max, lambda_min)
    np.random.seed(seed + 10000)
    paths_p = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_p)
    V = u_mat(paths_p[-1])
    lam = np.where(u_D(paths_p[-1]) >= V, lambda_max, lambda_min)
    for i in range(len(ts) - 2, -1, -1):
        dt_i = ts[i + 1] - ts[i]
        p_d = 1 - np.exp(-lam * dt_i)
        V = p_d * u_D(paths_p[i + 1]) + (1 - p_d) * V - a * dt_i
        if rd[i] is not None:
            ps, xk = rd[i]
            fs = pwlinear_basis(xk)
            EV = sum(c * f(paths_p[i]) for c, f in zip(ps, fs))
            lam = np.where(u_D(paths_p[i]) >= EV, lambda_max, lambda_min)
        else:
            lam = np.where(u_D(paths_p[i]) >= 0, lambda_max, lambda_min)
    return np.mean(V)

# bisection
a_lo, a_hi = 0.0, 10.0
for _ in range(15):
    a_mid = 0.5 * (a_lo + a_hi)
    p_mid = ls_price_alpha(a_mid)
    if p_mid > 0:
        a_lo = a_mid
    else:
        a_hi = a_mid

alpha_star = 0.5 * (a_lo + a_hi)
print(f"Break-even fee: alpha* ≈ {alpha_star:.3f}")
print(f"Price at alpha*: {ls_price_alpha(alpha_star):.4f}")
```

<b>(c).</b> Implement Longstaff-Schwartz algorithm to price the American option with payoff $u^D=0$ (Always perform a second indepedent run to get a clean lower bound price). Compare the price from uncertain lapse model in part (b) with the American price.

**Q4(a).** With $K_D = 0$, $u^D \equiv 0$. The "default" is now a lapse event. If the subscriber lapses optimally, they exercise at the first time the continuation value drops below 0 (the lapse payoff). This gives the American option formulation:

$$u(t, x) = \sup_{t \leq \tau \leq T}\;\mathbb{E}_{t,x}\!\left[\mathbb{1}_{\tau = T}\, u^{\mathrm{mat}}(X_T) - \int_t^\tau \alpha\,ds\right],$$

i.e., at each $t_i$,

$$u(t_i, x) = \max\!\Big(0,\; -\alpha\,\Delta t + \mathbb{E}_{t_i,x}\!\big[u(t_{i+1}, X_{t_{i+1}})\big]\Big).$$

The subscriber holds the right to receive $u^{\mathrm{mat}}(X_T)$ at maturity while paying running fees $\alpha$, and can walk away (lapse, value $0$) at any time. Pricing this contract under optimal lapsing is a standard American option problem with exercise payoff $g = 0$.

**Q4(b).** With $u^D = 0$, the HJB equation becomes

$$\partial_t u + \tfrac{1}{2}\sigma^2 x^2 \partial_{xx} u - \alpha - \Lambda^D(t, -u)\, u = 0.$$

When $u \leq 0$: $\Lambda^D = \overline{\lambda}$ (since $-u \geq 0$), giving a "killing" term $-\overline{\lambda}\, u$ that pushes $u$ back up.

When $u > 0$: $\Lambda^D = \underline{\lambda}$ (since $-u < 0$), giving a mild killing term.

Taking $\underline{\lambda} = 0$ and $\overline{\lambda} \to +\infty$ enforces $u \geq 0$ everywhere (the American constraint), while the PDE in the continuation region ($u > 0$) reduces to the Black–Scholes PDE with running cost $-\alpha$. This recovers the American option price.

We verify by running the LS uncertain lapse code with $K_D = 0$, $\underline{\lambda} = 0$, and $\overline{\lambda}$ very large.

```python
K_D_lapse = 0.0
alpha_lapse = 3.0
lam_lo = 0.0
lam_hi = 1000.0   # large upper bound to approximate infinity

def u_D_lapse(x):
    return np.maximum(K_D_lapse - x, 0)  # always 0

# --- regression pass ---
npaths_r = 10000
np.random.seed(42)
paths_r = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_r)

ls_reg_lapse = [None] * len(ts)
V = u_mat(paths_r[-1])
lam = np.where(u_D_lapse(paths_r[-1]) >= V, lam_hi, lam_lo)

for i in range(len(ts) - 2, 0, -1):
    dt_i = ts[i + 1] - ts[i]
    p_d = 1 - np.exp(-lam * dt_i)
    V = p_d * u_D_lapse(paths_r[i + 1]) + (1 - p_d) * V - alpha_lapse * dt_i
    x = paths_r[i]
    xk = np.linspace(np.percentile(x, 1), np.percentile(x, 99), 10)
    ps, fs = pwlinear_fit(x, V, xk)
    ls_reg_lapse[i] = (ps, xk)
    EV = sum(c * f(x) for c, f in zip(ps, fs))
    lam = np.where(u_D_lapse(x) >= EV, lam_hi, lam_lo)

# --- independent evaluation ---
npaths_p = 200000
np.random.seed(123)
paths_p = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_p)

V = u_mat(paths_p[-1])
lam = np.where(u_D_lapse(paths_p[-1]) >= V, lam_hi, lam_lo)

for i in range(len(ts) - 2, -1, -1):
    dt_i = ts[i + 1] - ts[i]
    p_d = 1 - np.exp(-lam * dt_i)
    V = p_d * u_D_lapse(paths_p[i + 1]) + (1 - p_d) * V - alpha_lapse * dt_i
    if ls_reg_lapse[i] is not None:
        ps, xk = ls_reg_lapse[i]
        fs_ = pwlinear_basis(xk)
        EV = sum(c * f(paths_p[i]) for c, f in zip(ps, fs_))
        lam = np.where(u_D_lapse(paths_p[i]) >= EV, lam_hi, lam_lo)
    else:
        lam = np.where(u_D_lapse(paths_p[i]) >= 0, lam_hi, lam_lo)

price_lapse = np.mean(V)
stderr_lapse = np.std(V) / np.sqrt(npaths_p)
print(f"Q4(b) Uncertain lapse model (K_D=0, lam_max={lam_hi}): {price_lapse:.4f} +/- {1.96*stderr_lapse:.4f}")
```

**Q4(c).** Standard Longstaff–Schwartz for the American lapse option: exercise payoff is $g = 0$ (lapse) at each $t_i$, terminal payoff $u^{\mathrm{mat}}(X_T)$. At each step, the subscriber continues if $\mathbb{E}[V_{i+1}|X_i] - \alpha\,\Delta t \geq 0$, otherwise lapses.

```python
# --- regression pass ---
npaths_r = 10000
np.random.seed(42)
paths_r = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_r)

am_reg = [None] * len(ts)
V = u_mat(paths_r[-1])

for i in range(len(ts) - 2, 0, -1):
    dt_i = ts[i + 1] - ts[i]
    x = paths_r[i]
    xk = np.linspace(np.percentile(x, 1), np.percentile(x, 99), 10)
    ps, fs = pwlinear_fit(x, V, xk)
    am_reg[i] = (ps, xk)
    EV = sum(c * f(x) for c, f in zip(ps, fs))
    # lapse if continuation value < 0
    exercise = (EV - alpha_lapse * dt_i < 0)
    V = np.where(exercise, 0.0, V - alpha_lapse * dt_i)

# --- independent evaluation (lower bound) ---
npaths_p = 200000
np.random.seed(123)
paths_p = blackscholes_mc(S=S, vol=vol, ts=ts, npaths=npaths_p)

V = u_mat(paths_p[-1])

for i in range(len(ts) - 2, -1, -1):
    dt_i = ts[i + 1] - ts[i]
    x = paths_p[i]
    if am_reg[i] is not None:
        ps, xk = am_reg[i]
        fs_ = pwlinear_basis(xk)
        EV = sum(c * f(x) for c, f in zip(ps, fs_))
    else:
        EV = np.mean(V) * np.ones(len(x))
    exercise = (EV - alpha_lapse * dt_i < 0)
    V = np.where(exercise, 0.0, V - alpha_lapse * dt_i)

price_american = np.mean(V)
stderr_american = np.std(V) / np.sqrt(npaths_p)
print(f"Q4(c) American LS price:     {price_american:.4f} +/- {1.96*stderr_american:.4f}")
print(f"Q4(b) Uncertain lapse price: {price_lapse:.4f} +/- {1.96*stderr_lapse:.4f}")
print()
print("The two prices should be close, confirming that the uncertain lapse model")
print("with lambda_min=0 and lambda_max->infty recovers the American option price.")
```
