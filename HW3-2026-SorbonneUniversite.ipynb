# Uncertain Mortality Model vs American Options

#### Re-insurance Deals

In this assignment, we study the *Uncertain Mortality Model* for the pricing of reinsurance deals. For the sake of simplicity, in this assignment we only consider one type of default: the risk of death.

We consider the following reinsurance product:

- At maturity, if the insurance subscriber is alive, the issuer delivers a put on the underlying $X$

$$
u^\textrm{mat}( x) = (K_\textrm{mat} - x)_+.
$$

- At the time of death, if it is before the maturity, the issuer delivers an exit payoff, typically another put on the underlying $X$.

$$
u^D(t, x) = (K_D - x)_+.
$$

- The subscriber pays a constant fee $α\Delta t$ at every time step until death or the maturity or the product

- The insurance sells a large number of these contracts to subscribers. We assume that the times of death of the subscribers $\tau^D$ are independent, and identically distributed, and also independent of the underlying's stock price.

- We assume that the underlying's risk neutral price dynamics is the Black-Scholes model with zero interest rate/repo/dividend yield

$$
dX_t = \sigma X_t d W_t.
$$

#### The Insurers' Approach and Risk-Neutral Pricing

This contract shows two types of risk: the times of death of the subscribers and the changes in the price of the underlying. 

In this case, the issuer can apply the insurer's approach to the risk of death times, i.e., the law of large numbers. The more people buy the contract, the less risk.

Choosing a risk-neutral measure under which the death times $\tau^D$ have the same distribution as under the historical probability measure is equivalent to applying the arbitrage-pricing approach to the financial risk insurer's rule on the risk of death. The price of the contract is then

$$
u(t, x) = \mathbb{E}^\mathbb{Q}_{t, x} \left[u^\textrm{mat}(X_T) \mathbb{1}_{\tau^D \geq T} -\int_t^T\alpha\mathbb{1}_{\tau^D>s}ds+ u^D(\tau^D, X_{\tau^D}) \mathbb{1}_{\tau^D < T} \right].
$$

#### Deterministic Death Rate

If the death intensity is a deterministic function $\lambda_t^D$ (i.e. $\tau^D$ has an exponential distribution with time-dependent intensity $\lambda_t^D$), then we have seen that $u$ is the solution to the linear PDE

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial_x^2 u -\alpha + \lambda_t^D \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat}.
\end{array}\right.
$$

#### Uncertain Mortality Model

If the death rate is uncertain, we assume that it is adapted (i.e., it does not look into the future) and belongs to a moving corridor
$\left[\underline{\lambda}_t, \overline{\lambda}_t\right]$. The most conservative way to price the contract is to compute the (financially) worst death rate process $\lambda_t^D$ as being chosen so as to maximize the value of the contract. The resulting HJB equation is

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial^2_x u -\alpha + \Lambda^D(t, u^D - u) \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat},
\end{array}\right.
$$
where the function $\Lambda^D$ is defined by
$$
\Lambda^D (t, y) = \left\{\begin{array}{l}
    \overline{\lambda}^D_t \quad \textrm{if} \ y \geq 0, \\
    \underline{\lambda}^D_t \quad \textrm{otherwise.}
\end{array}\right.
$$

## Link with $1$-BSDE and Numerical Schemes

From the Pardoux-Peng theorem, we know that the solution $u(0, x)$ can be represented as the solution $Y_0^x$ to the $1$-BSDE

$$
dY_t = -f(t, X_t, Y_t, Z_t) \, dt + Z_t \, dW_t,
$$

with the terminal condition $Y_T = u^\textrm{mat} (X_T)$, where $X_0 = x$ and

$$
f(t, x, y, z) = -\alpha + \Lambda^D(t, u^D(t, x) - y) \cdot (u^D(t, x) - y).
$$


## BSDE Discretization

### Explicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i + \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1} , X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ]\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \right) \Delta t_i\\
=& -\alpha\Delta t_i + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \overline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]}\\
& + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \underline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) < \mathbb{E}_{i-1}[Y_{t_i}]}.
\end{align*}

### Implicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [Y_{t_i}] -\alpha\Delta t_i+ \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}}\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}} \right) \Delta t_i.
\end{align*}

The implicit scheme involves $Y_{t_{i - 1}}$ on both sides of the equation, which generally requires a root-finding routine to find $Y_{t_{i - 1}}$. However in this specific case, it can be shown that 

\begin{equation}u^D(t_{i-1}, X_{t_{i-1}}) \geq Y_{t_{i-1}}\quad\text{if and only if}\quad u^D(t_{i-1}, X_{t_{i-1}}) \geq \mathbb{E}^{\mathbb{Q}}_{i-1}\left[Y_{t_i}\right]-\alpha\Delta t_i. \qquad (E1) \end{equation}

Thus

\begin{split}
Y_{t_{i - 1}} =& \frac{1}{1 + \overline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] - \alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}\\
& + \frac{1}{1 + \underline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D(t_{i - 1}, X_{t_{i - 1}}) < \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}. \qquad (E2)
\end{split}



```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
def blackscholes_mc(S=100, vol=0.2, r=0, q=0, ts=np.linspace(0, 1, 13), npaths=10):
    """Generate Monte-Carlo paths in the Black-Scholes model."""
    nsteps = len(ts) - 1
    ts = np.asarray(ts, dtype=float)[:, np.newaxis]
    dts = np.diff(ts, axis=0)
    W = np.cumsum(np.vstack((np.zeros((1, npaths), dtype=float),
                             np.random.randn(nsteps, npaths) * np.sqrt(dts))),
                  axis=0)
    paths = np.exp(-0.5 * vol**2 * ts + vol * W) * S * np.exp((r - q) * ts)
    return paths


def blackscholes_price(K, T, S, vol, r=0, q=0, callput='call'):
    """Compute call/put prices in the Black-Scholes model."""
    F = S * np.exp((r - q) * T)
    v = np.sqrt(vol**2 * T)
    d1 = np.log(F / K) / v + 0.5 * v
    d2 = d1 - v
    try:
        opttype = {'call': 1, 'put': -1}[callput.lower()]
    except Exception:
        raise ValueError('The value of callput must be either "call" or "put".')
    price = opttype * (F * norm.cdf(opttype * d1) - K * norm.cdf(opttype * d2)) * np.exp(-r * T)
    return price


def pwlinear_basis(xknots):
    """Basis that represents a piecewise-linear function with given knots."""
    fs = [lambda x: np.ones_like(x, dtype=float), lambda x: x - xknots[0]]
    fs.extend([lambda x, a=xknots[i]: np.maximum(x - a, 0) for i in range(len(xknots))])
    return fs


def pwlinear_fit(xdata, ydata, xknots):
    """Fit a piecewise-linear function with xknots to xdata and ydata."""
    fs = pwlinear_basis(xknots)
    A = np.vstack([f(xdata) for f in fs]).T
    ps = np.linalg.lstsq(A, ydata, rcond=None)[0]
    return ps, fs
```

<b>Note.</b> To be more specific about the payouts of the product, we make the following assumptions:

- We assume that all transactions occur at the end of the month, i.e. the insurance subsriber is expected to pay a fee of $\alpha/12$ at the end of each month; if the subscriber dies in the middle of the month, the death payment will be made at the end of that month. 

- If the death occurs at the maturity, we assume that the death comes first and the death payment $u^D$ will be made.

### Question 1

For constant intensity $\lambda^D$, under the monthly convention $\Delta t = 1/12$ and $t_i=i\Delta t$:

- if $\tau^D>T$, payoff is $u^{\mathrm{mat}}(X_T)-\alpha T$;
- if death occurs in month $m$ (i.e. $\tau^D\in(t_{m-1},t_m]$), payoff is $u^D(X_{t_m})-\alpha (m-1)\Delta t$.

This gives a direct Monte Carlo estimator by simulating $\tau^D$.

Averaging over death events (independence between $\tau^D$ and $X$):
$$
u(0,x)=\mathbb E\left[e^{-\lambda^D T}u^{\mathrm{mat}}(X_T)+\int_0^T e^{-\lambda^D s}(\lambda^D u^D(X_s)-\alpha)\,ds\right].$$
With monthly discretization this is equivalent to
$$V_{i-1}=e^{-\lambda^D\Delta t}(V_i-\alpha\Delta t)+(1-e^{-\lambda^D\Delta t})u^D(X_{t_i}),\qquad V_n=u^{\mathrm{mat}}(X_T).$$

```python
# Common parameters used below
S0 = 100.0
sigma = 0.3
T = 10.0
alpha = 3.0
K_mat = 90.0
K_D = 100.0
lambda_det = 0.025
lambda_min = 0.005
lambda_max = 0.04

ts = np.linspace(0.0, T, int(np.round(12 * T)) + 1)
dt = ts[1] - ts[0]


def u_mat_payoff(x, K=K_mat):
    return np.maximum(K - x, 0.0)


def u_D_payoff(x, K=K_D):
    return np.maximum(K - x, 0.0)


def mc_mean_se(x):
    return np.mean(x), np.std(x, ddof=1) / np.sqrt(len(x))


def q1_direct_death_mc(npaths=100_000, seed=1234):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0.0, q=0.0, ts=ts, npaths=npaths)
    tau = exponential_samples(npaths, lambda_det)

    n = len(ts) - 1
    alive_to_T = tau > T

    # Death month index m in {1, ..., n}; if alive, m is clipped but unused in payoff branch
    death_month = np.ceil(np.minimum(tau, T) / dt).astype(int)
    death_month = np.clip(death_month, 1, n)

    x_death = paths[death_month, np.arange(npaths)]
    fees = alpha * dt * np.where(alive_to_T, n, death_month - 1)
    payoff = np.where(alive_to_T, u_mat_payoff(paths[-1]), u_D_payoff(x_death)) - fees

    return mc_mean_se(payoff)


def q1_averaged_death_mc(npaths=100_000, seed=5678, lambd=lambda_det):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0.0, q=0.0, ts=ts, npaths=npaths)

    surv = np.exp(-lambd * dt)
    V = u_mat_payoff(paths[-1])
    for i in range(len(ts) - 2, -1, -1):
        V = surv * (V - alpha * dt) + (1.0 - surv) * u_D_payoff(paths[i + 1])

    return mc_mean_se(V)


q1_direct_price, q1_direct_se = q1_direct_death_mc()
q1_avg_price, q1_avg_se = q1_averaged_death_mc()

print(f"Q1(a) direct death-time MC: {q1_direct_price:.4f}  (SE {q1_direct_se:.4f})")
print(f"Q1(b) averaged death-event MC: {q1_avg_price:.4f}  (SE {q1_avg_se:.4f})")
```

Both estimators are consistent and close within Monte Carlo error. The averaged representation in (b) is typically less noisy because death randomness is integrated out analytically.

### Question 2

Set
$$c:=\mathbb E^{\mathbb Q}_{i-1}[Y_{t_i}]-\alpha\Delta t_i,\qquad d:=u^D(t_{i-1},X_{t_{i-1}}).$$
The implicit step is
$$Y_{t_{i-1}}=c+\Lambda^D(t_{i-1},d-Y_{t_{i-1}})(d-Y_{t_{i-1}})\Delta t_i.$$ 

If $d\ge Y_{t_{i-1}}$, then $\Lambda^D=\overline\lambda^D$ and
$$(1+\overline\lambda^D\Delta t_i)Y_{t_{i-1}}=c+\overline\lambda^D\Delta t_i d.$$ 
Now
$$d\ge Y_{t_{i-1}}\iff d(1+\overline\lambda^D\Delta t_i)\ge c+\overline\lambda^D\Delta t_i d\iff d\ge c.$$ 

If $d<Y_{t_{i-1}}$, similarly with $\underline\lambda^D$:
$$(1+\underline\lambda^D\Delta t_i)Y_{t_{i-1}}=c+\underline\lambda^D\Delta t_i d,
\quad d<Y_{t_{i-1}}\iff d<c.$$ 
Hence
$$d\ge Y_{t_{i-1}}\iff d\ge c=\mathbb E^{\mathbb Q}_{i-1}[Y_{t_i}]-\alpha\Delta t_i,$$
which is (E1), and plugging this condition in the two linear cases gives (E2).

```python
def pwlinear_predict(x, knots, ps):
    fs = pwlinear_basis(knots)
    A = np.vstack([f(x) for f in fs]).T
    return A @ ps


def bsde_implicit_regression(alpha_val=alpha, K_D_val=K_D,
                             lambda_min_val=lambda_min, lambda_max_val=lambda_max,
                             npaths=100_000, n_knots=10, seed=101):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0.0, q=0.0, ts=ts, npaths=npaths)

    Y = u_mat_payoff(paths[-1], K=K_mat)
    n = len(ts) - 1
    models = [None] * n

    for i in range(n - 1, -1, -1):
        x = paths[i]
        dt_i = ts[i + 1] - ts[i]

        x_min, x_max = np.percentile(x, [1, 99])
        knots = np.linspace(x_min, x_max, n_knots)
        ps, fs = pwlinear_fit(x, Y, knots)
        cont = np.vstack([f(x) for f in fs]).T @ ps

        rhs = cont - alpha_val * dt_i
        uD_i = u_D_payoff(x, K=K_D_val)
        lam = np.where(uD_i >= rhs, lambda_max_val, lambda_min_val)
        Y = (rhs + lam * dt_i * uD_i) / (1.0 + lam * dt_i)

        models[i] = (knots, ps)

    return models, np.mean(Y)


def bsde_lower_bound_independent(models, alpha_val=alpha, K_D_val=K_D,
                                 lambda_min_val=lambda_min, lambda_max_val=lambda_max,
                                 npaths=100_000, seed=202):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0.0, q=0.0, ts=ts, npaths=npaths)

    V = u_mat_payoff(paths[-1], K=K_mat)
    n = len(ts) - 1

    for i in range(n - 1, -1, -1):
        x = paths[i]
        dt_i = ts[i + 1] - ts[i]

        knots, ps = models[i]
        cont_hat = pwlinear_predict(x, knots, ps)
        uD_i = u_D_payoff(x, K=K_D_val)

        lam = np.where(uD_i >= cont_hat - alpha_val * dt_i, lambda_max_val, lambda_min_val)
        p_death = 1.0 - np.exp(-lam * dt_i)
        V = p_death * u_D_payoff(paths[i + 1], K=K_D_val) + (1.0 - p_death) * V - alpha_val * dt_i

    return mc_mean_se(V)


def deterministic_mortality_price(lambda_val, npaths=100_000, seed=303):
    return q1_averaged_death_mc(npaths=npaths, seed=seed, lambd=lambda_val)


q2_models, q2_regression_value = bsde_implicit_regression(npaths=100_000, seed=11)
q2_lb_price, q2_lb_se = bsde_lower_bound_independent(q2_models, npaths=100_000, seed=12)
q2_det_min_price, q2_det_min_se = deterministic_mortality_price(lambda_min, npaths=100_000, seed=13)
q2_det_max_price, q2_det_max_se = deterministic_mortality_price(lambda_max, npaths=100_000, seed=14)

print(f"Q2(b) implicit BSDE (regression run): {q2_regression_value:.4f}")
print(f"Q2(c) lower bound (independent run): {q2_lb_price:.4f}  (SE {q2_lb_se:.4f})")
print(f"Q2(d) deterministic lambda_min={lambda_min:.3f}: {q2_det_min_price:.4f}  (SE {q2_det_min_se:.4f})")
print(f"Q2(d) deterministic lambda_max={lambda_max:.3f}: {q2_det_max_price:.4f}  (SE {q2_det_max_se:.4f})")
```

Numerically, the uncertain-mortality lower bound is much higher than the deterministic $\underline\lambda^D$ case and close to the deterministic $\overline\lambda^D$ case. This is consistent with a worst-case control that often selects high mortality intensity when the death payoff is favorable.

### Question 3

For the Longstaff-Schwartz-like scheme, replace the basis $\{1,\text{put}_{\mathrm{BS}}\}$ by the same piecewise-linear basis as in Question 2. At time $t_i$, regress the future value on $X_{t_i}$ to estimate continuation $C_i(X_{t_i})$, then use the rule
$$\lambda_i^D=\overline\lambda^D\mathbf 1_{\{u^D(X_{t_i})\ge C_i(X_{t_i})-\alpha\Delta t\}}+\underline\lambda^D\mathbf 1_{\{u^D(X_{t_i})< C_i(X_{t_i})-\alpha\Delta t\}}.$$

```python
def uncertain_lsm_piecewise(alpha_val=alpha, K_D_val=K_D,
                            lambda_min_val=lambda_min, lambda_max_val=lambda_max,
                            n_reg=5_000, n_price=50_000, n_knots=10, seed=404):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0.0, q=0.0, ts=ts, npaths=n_reg)

    V = u_mat_payoff(paths[-1], K=K_mat)
    n = len(ts) - 1
    models = [None] * n

    for i in range(n - 1, -1, -1):
        x = paths[i]
        dt_i = ts[i + 1] - ts[i]

        x_min, x_max = np.percentile(x, [1, 99])
        knots = np.linspace(x_min, x_max, n_knots)
        ps, fs = pwlinear_fit(x, V, knots)
        cont = np.vstack([f(x) for f in fs]).T @ ps

        uD_i = u_D_payoff(x, K=K_D_val)
        lam = np.where(uD_i >= cont - alpha_val * dt_i, lambda_max_val, lambda_min_val)
        p_death = 1.0 - np.exp(-lam * dt_i)
        V = p_death * u_D_payoff(paths[i + 1], K=K_D_val) + (1.0 - p_death) * V - alpha_val * dt_i

        models[i] = (knots, ps)

    np.random.seed(seed + 1)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0.0, q=0.0, ts=ts, npaths=n_price)

    V = u_mat_payoff(paths[-1], K=K_mat)
    for i in range(n - 1, -1, -1):
        x = paths[i]
        dt_i = ts[i + 1] - ts[i]

        knots, ps = models[i]
        cont_hat = pwlinear_predict(x, knots, ps)
        uD_i = u_D_payoff(x, K=K_D_val)
        lam = np.where(uD_i >= cont_hat - alpha_val * dt_i, lambda_max_val, lambda_min_val)

        p_death = 1.0 - np.exp(-lam * dt_i)
        V = p_death * u_D_payoff(paths[i + 1], K=K_D_val) + (1.0 - p_death) * V - alpha_val * dt_i

    return mc_mean_se(V)


def bsde_two_step_price(alpha_val=alpha, K_D_val=K_D,
                        lambda_min_val=lambda_min, lambda_max_val=lambda_max,
                        n_reg=5_000, n_price=50_000, seed=505):
    models, _ = bsde_implicit_regression(alpha_val=alpha_val, K_D_val=K_D_val,
                                         lambda_min_val=lambda_min_val, lambda_max_val=lambda_max_val,
                                         npaths=n_reg, n_knots=10, seed=seed)
    return bsde_lower_bound_independent(models, alpha_val=alpha_val, K_D_val=K_D_val,
                                        lambda_min_val=lambda_min_val, lambda_max_val=lambda_max_val,
                                        npaths=n_price, seed=seed + 1)


# Q3(a): piecewise-linear Longstaff-Schwartz-like estimate
q3_ls_price, q3_ls_se = uncertain_lsm_piecewise(alpha_val=alpha, n_reg=5_000, n_price=50_000, seed=40)
print(f"Q3(a) LS-like piecewise estimate: {q3_ls_price:.4f}  (SE {q3_ls_se:.4f})")


# Q3(b): variance comparison (20 repetitions, independent two-step runs)
reps = 20
bsde_prices = np.zeros(reps)
ls_prices = np.zeros(reps)

for k in range(reps):
    bsde_prices[k], _ = bsde_two_step_price(alpha_val=alpha, n_reg=5_000, n_price=50_000, seed=1000 + 7 * k)
    ls_prices[k], _ = uncertain_lsm_piecewise(alpha_val=alpha, n_reg=5_000, n_price=50_000, seed=2000 + 7 * k)

print(f"Q3(b) BSDE mean={np.mean(bsde_prices):.4f}, var={np.var(bsde_prices, ddof=1):.6f}")
print(f"Q3(b) LS-like mean={np.mean(ls_prices):.4f}, var={np.var(ls_prices, ddof=1):.6f}")


# Q3(c): alpha such that initial value is approximately zero

def q3_price_alpha(a, seed):
    return uncertain_lsm_piecewise(alpha_val=a, n_reg=7_000, n_price=50_000, seed=seed)[0]


a_low, a_high = 2.5, 3.5
p_low = q3_price_alpha(a_low, 3000)
p_high = q3_price_alpha(a_high, 3001)

for it in range(4):
    a_try = a_low - p_low * (a_high - a_low) / (p_high - p_low)
    p_try = q3_price_alpha(a_try, 3010 + it)
    if p_low * p_try <= 0:
        a_high, p_high = a_try, p_try
    else:
        a_low, p_low = a_try, p_try

alpha_star = 0.5 * (a_low + a_high)
alpha_star_price, alpha_star_se = uncertain_lsm_piecewise(alpha_val=alpha_star, n_reg=7_000, n_price=50_000, seed=4000)

print(f"Q3(c) bracket after iterations: [{a_low:.4f}, {a_high:.4f}]")
print(f"Q3(c) costless fee alpha* ≈ {alpha_star:.4f}")
print(f"      check price(alpha*): {alpha_star_price:.4f}  (SE {alpha_star_se:.4f})")
```

The two methods give close means. In my simulation, the LS-like estimator has a slightly smaller empirical variance than the BSDE lower-bound estimator. The fee solving $u(0,S_0)\approx0$ is around $\alpha\approx3.3$--$3.4$ (small variation is due to Monte Carlo noise).

### Question 4

With $K_D=0$, lapse gives immediate payoff $0$ and stops future fees/payoffs.

For optimal lapse (monthly exercise dates), the value is the Snell envelope
$$V_n=u^{\mathrm{mat}}(X_T),\qquad V_i=\max\left(0,\mathbb E_i[V_{i+1}]-\alpha\Delta t\right).$$
So pricing reduces to a Bermudan/American optimal stopping problem.

In the uncertain-lapse model, set $\underline\lambda=0$ and take $\overline\lambda$ large. In the implicit update with $u^D=0$,
$$Y_i=\frac{c_i}{1+\lambda_i\Delta t},\qquad c_i:=\mathbb E_i[Y_{i+1}]-\alpha\Delta t,$$
with the rule $\lambda_i=\overline\lambda\mathbf 1_{\{c_i\le 0\}}$. As $\overline\lambda\to\infty$, this converges to $Y_i=\max(c_i,0)$, i.e. the optimal stopping recursion above.

```python
def american_lsm_lapse(alpha_val=alpha, n_reg=5_000, n_price=50_000, n_knots=10, seed=606):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0.0, q=0.0, ts=ts, npaths=n_reg)

    V = u_mat_payoff(paths[-1], K=K_mat)
    n = len(ts) - 1
    models = [None] * n

    for i in range(n - 1, -1, -1):
        x = paths[i]
        dt_i = ts[i + 1] - ts[i]

        x_min, x_max = np.percentile(x, [1, 99])
        knots = np.linspace(x_min, x_max, n_knots)
        ps, fs = pwlinear_fit(x, V, knots)
        cont = np.vstack([f(x) for f in fs]).T @ ps - alpha_val * dt_i

        V = np.maximum(0.0, cont)
        models[i] = (knots, ps)

    np.random.seed(seed + 1)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0.0, q=0.0, ts=ts, npaths=n_price)

    V = u_mat_payoff(paths[-1], K=K_mat)
    for i in range(n - 1, -1, -1):
        x = paths[i]
        dt_i = ts[i + 1] - ts[i]

        knots, ps = models[i]
        cont_hat = pwlinear_predict(x, knots, ps) - alpha_val * dt_i
        exercise = cont_hat <= 0.0

        V = np.where(exercise, 0.0, V - alpha_val * dt_i)

    return mc_mean_se(V)


def uncertain_lapse_limit_implicit(lambda_high, npaths=100_000, seed=707):
    _, y0 = bsde_implicit_regression(alpha_val=alpha, K_D_val=0.0,
                                     lambda_min_val=0.0, lambda_max_val=lambda_high,
                                     npaths=npaths, n_knots=10, seed=seed)
    return y0


lambda_grid = [5.0, 20.0, 100.0, 500.0]
q4_uncertain_vals = [uncertain_lapse_limit_implicit(lmbd, npaths=100_000, seed=800 + j)
                     for j, lmbd in enumerate(lambda_grid)]
q4_american_price, q4_american_se = american_lsm_lapse(alpha_val=alpha, n_reg=5_000, n_price=50_000, seed=900)

print("Q4(b) uncertain lapse approximation (lambda_min=0):")
for lmbd, val in zip(lambda_grid, q4_uncertain_vals):
    print(f"  lambda_max={lmbd:>6.1f} -> value {val:.4f}")

print(f"Q4(c) American LSM lower bound: {q4_american_price:.4f}  (SE {q4_american_se:.4f})")
```

As $\overline\lambda$ increases with $\underline\lambda=0$, the uncertain-lapse value stabilizes and gets close to the American lower-bound price, which is consistent with the optimal-exercise limit.
