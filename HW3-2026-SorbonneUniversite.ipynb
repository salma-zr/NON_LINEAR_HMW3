# Uncertain Mortality Model vs American Options

#### Re-insurance Deals

In this assignment, we study the *Uncertain Mortality Model* for the pricing of reinsurance deals. For the sake of simplicity, in this assignment we only consider one type of default: the risk of death.

We consider the following reinsurance product:

- At maturity, if the insurance subscriber is alive, the issuer delivers a put on the underlying $X$

$$
u^\textrm{mat}( x) = (K_\textrm{mat} - x)_+.
$$

- At the time of death, if it is before the maturity, the issuer delivers an exit payoff, typically another put on the underlying $X$.

$$
u^D(t, x) = (K_D - x)_+.
$$

- The subscriber pays a constant fee $α\Delta t$ at every time step until death or the maturity or the product

- The insurance sells a large number of these contracts to subscribers. We assume that the times of death of the subscribers $\tau^D$ are independent, and identically distributed, and also independent of the underlying's stock price.

- We assume that the underlying's risk neutral price dynamics is the Black-Scholes model with zero interest rate/repo/dividend yield

$$
dX_t = \sigma X_t d W_t.
$$

#### The Insurers' Approach and Risk-Neutral Pricing

This contract shows two types of risk: the times of death of the subscribers and the changes in the price of the underlying. 

In this case, the issuer can apply the insurer's approach to the risk of death times, i.e., the law of large numbers. The more people buy the contract, the less risk.

Choosing a risk-neutral measure under which the death times $\tau^D$ have the same distribution as under the historical probability measure is equivalent to applying the arbitrage-pricing approach to the financial risk insurer's rule on the risk of death. The price of the contract is then

$$
u(t, x) = \mathbb{E}^\mathbb{Q}_{t, x} \left[u^\textrm{mat}(X_T) \mathbb{1}_{\tau^D \geq T} -\int_t^T\alpha\mathbb{1}_{\tau^D>s}ds+ u^D(\tau^D, X_{\tau^D}) \mathbb{1}_{\tau^D < T} \right].
$$

#### Deterministic Death Rate

If the death intensity is a deterministic function $\lambda_t^D$ (i.e. $\tau^D$ has an exponential distribution with time-dependent intensity $\lambda_t^D$), then we have seen that $u$ is the solution to the linear PDE

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial_x^2 u -\alpha + \lambda_t^D \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat}.
\end{array}\right.
$$

#### Uncertain Mortality Model

If the death rate is uncertain, we assume that it is adapted (i.e., it does not look into the future) and belongs to a moving corridor
$\left[\underline{\lambda}_t, \overline{\lambda}_t\right]$. The most conservative way to price the contract is to compute the (financially) worst death rate process $\lambda_t^D$ as being chosen so as to maximize the value of the contract. The resulting HJB equation is

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial^2_x u -\alpha + \Lambda^D(t, u^D - u) \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat},
\end{array}\right.
$$
where the function $\Lambda^D$ is defined by
$$
\Lambda^D (t, y) = \left\{\begin{array}{l}
    \overline{\lambda}^D_t \quad \textrm{if} \ y \geq 0, \\
    \underline{\lambda}^D_t \quad \textrm{otherwise.}
\end{array}\right.
$$

## Link with $1$-BSDE and Numerical Schemes

From the Pardoux-Peng theorem, we know that the solution $u(0, x)$ can be represented as the solution $Y_0^x$ to the $1$-BSDE

$$
dY_t = -f(t, X_t, Y_t, Z_t) \, dt + Z_t \, dW_t,
$$

with the terminal condition $Y_T = u^\textrm{mat} (X_T)$, where $X_0 = x$ and

$$
f(t, x, y, z) = -\alpha + \Lambda^D(t, u^D(t, x) - y) \cdot (u^D(t, x) - y).
$$


## BSDE Discretization

### Explicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i + \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1} , X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ]\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \right) \Delta t_i\\
=& -\alpha\Delta t_i + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \overline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]}\\
& + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \underline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) < \mathbb{E}_{i-1}[Y_{t_i}]}.
\end{align*}

### Implicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [Y_{t_i}] -\alpha\Delta t_i+ \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}}\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}} \right) \Delta t_i.
\end{align*}

The implicit scheme involves $Y_{t_{i - 1}}$ on both sides of the equation, which generally requires a root-finding routine to find $Y_{t_{i - 1}}$. However in this specific case, it can be shown that 

\begin{equation}u^D(t_{i-1}, X_{t_{i-1}}) \geq Y_{t_{i-1}}\quad\text{if and only if}\quad u^D(t_{i-1}, X_{t_{i-1}}) \geq \mathbb{E}^{\mathbb{Q}}_{i-1}\left[Y_{t_i}\right]-\alpha\Delta t_i. \qquad (E1) \end{equation}

Thus

\begin{split}
Y_{t_{i - 1}} =& \frac{1}{1 + \overline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] - \alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}\\
& + \frac{1}{1 + \underline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D(t_{i - 1}, X_{t_{i - 1}}) < \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}. \qquad (E2)
\end{split}



```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
def blackscholes_mc(S=100, vol=0.2, r=0, q=0, ts=np.linspace(0, 1, 13), npaths=10):
    """Generate Monte-Carlo paths in Black-Scholes model.

    Parameters
    ----------
    S: scalar
        The spot price of the underlying security.
    vol: scalar
        The implied Black-Scholes volatility.
    r: scalar
        The annualized risk-free interest rate, continuously compounded.
    q: scalar
        The annualized continuous dividend yield.
    ts: array_like
        The time steps of the simualtion
    npaths: int
        the number of paths to simulate

    Returns
    -------
    paths: ndarray
        The Monte-Carlo paths.
    """
    nsteps = len(ts) - 1
    ts = np.asarray(ts, dtype=float)[:, np.newaxis]
    W = np.cumsum(np.vstack((np.zeros((1, npaths), dtype=float),
                             np.random.randn(nsteps, npaths) * np.sqrt(np.diff(ts, axis=0)))),
                  axis=0)
    paths = np.exp(-0.5*vol**2*ts + vol*W)*S*np.exp((r-q)*ts)
    return paths

def blackscholes_price(K, T, S, vol, r=0, q=0, callput='call'):
    """Compute the call/put option price in the Black-Scholes model
    
    Parameters
    ----------
    K: scalar or array_like
        The strike of the option.
    T: scalar or array_like
        The maturity of the option, expressed in years (e.g. 0.25 for 3-month and 2 for 2 years)
    S: scalar or array_like
        The current price of the underlying asset.
    vol: scalar or array_like
        The implied Black-Scholes volatility.
    r: scalar or array_like
        The annualized risk-free interest rate, continuously compounded.
    q: scalar or array_like
        The annualized continuous dividend yield.
    callput: str
        Must be either 'call' or 'put'.

    Returns
    -------
    price: scalar or array_like
        The price of the option.

    Examples
    --------
    >>> blackscholes_price(95, 0.25, 100, 0.2, r=0.05, callput='put')
    1.5342604771222823
    """
    F = S*np.exp((r-q)*T)
    v = np.sqrt(vol**2*T)
    d1 = np.log(F/K)/v + 0.5*v
    d2 = d1 - v
    try:
        opttype = {'call':1, 'put':-1}[callput.lower()]
    except:
        raise ValueError('The value of callput must be either "call" or "put".')
    price = opttype*(F*norm.cdf(opttype*d1)-K*norm.cdf(opttype*d2))*np.exp(-r*T)
    return price


def pwlinear_basis(xknots):
    """Basis that represent a piecewise linear function with given knots"""
    fs = [lambda x: np.ones_like(x, dtype=float), lambda x: x-xknots[0]]
    fs.extend([lambda x, a=xknots[i]: np.maximum(x-a, 0) for i in range(len(xknots))])
    return fs


def pwlinear_fit(xdata, ydata, xknots):
    """Fit a piecewise linear function with xknots to xdata and ydata"""
    fs = pwlinear_basis(xknots)
    A = np.vstack([f(xdata) for f in fs]).T
    ps = np.linalg.lstsq(A, ydata, rcond=None)[0]
    return ps, fs
```

<b>Note.</b> To be more specific about the payouts of the product, we make the following assumptions:

- We assume that all transactions occur at the end of the month, i.e. the insurance subsriber is expected to pay a fee of $\alpha/12$ at the end of each month; if the subscriber dies in the middle of the month, the death payment will be made at the end of that month. 

- If the death occurs at the maturity, we assume that the death comes first and the death payment $u^D$ will be made.

```python
# note that lambda is a reserved keyword in python, so we use lambd instead
def exponential_samples(n, lambd):
    """Generate n samples from exponential distribution with rate lambd"""
    return -np.log(np.random.rand(n))/lambd
```

### Question 1(a) — Direct simulation of death times

We simulate $\tau^D \sim \mathrm{Exp}(\lambda^D)$ independently for each path. Since transactions occur at month-ends, the death is snapped to the end of the month in which it falls. For each path:

- If $\tau^D > T$: subscriber is alive at maturity, payoff $= u^{\mathrm{mat}}(X_T) - \alpha T$
- If $\tau^D \leq T$: death at discretized time $t_k = \lceil \tau^D \cdot 12 \rceil / 12$, payoff $= u^D(X_{t_k}) - \alpha \cdot k \cdot \Delta t$

where $\Delta t = 1/12$ and the fee is paid at every month-end (including the month of death).

```python
np.random.seed(42)

S0 = 100
vol = 0.3
T = 10
alpha = 3.0
K_mat = 90.0
K_D = 100.0
lambd = 0.025
nsteps = int(np.round(T * 12))
ts = np.linspace(0, T, nsteps + 1)
dt = ts[1] - ts[0]

npaths = 200000
paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths)

tau_D = exponential_samples(npaths, lambd)

# snap death to end of month; if tau_D > T, subscriber is alive
death_step = np.ceil(tau_D / dt).astype(int)
alive = death_step > nsteps
death_step_clipped = np.clip(death_step, 1, nsteps)

# X at death time (for alive paths, value won't be used)
death_X = paths[death_step_clipped, np.arange(npaths)]

payoff = np.where(alive,
                  np.maximum(K_mat - paths[-1], 0) - alpha * T,
                  np.maximum(K_D - death_X, 0) - alpha * death_step_clipped * dt)

price_1a = np.mean(payoff)
se_1a = np.std(payoff) / np.sqrt(npaths)
print(f"Q1(a) price (direct simulation): {price_1a:.4f} ± {1.96*se_1a:.4f}")
```

### Question 1(b) — Feynman-Kac representation (averaging over death events)

The PDE $\partial_t u + \tfrac{1}{2}\sigma^2 x^2 \partial_x^2 u - \alpha + \lambda^D(u^D - u) = 0$ can be rewritten as

$$\partial_t u + \tfrac{1}{2}\sigma^2 x^2 \partial_x^2 u - \lambda^D u = \alpha - \lambda^D u^D.$$

By Feynman-Kac, since the "killing rate" is $\lambda^D$ and the source term is $\lambda^D u^D(t, X_t) - \alpha$:

$$u(0, X_0) = \mathbb{E}\left[ u^{\mathrm{mat}}(X_T)\, e^{-\lambda^D T} + \int_0^T \big(\lambda^D\, u^D(s, X_s) - \alpha\big)\, e^{-\lambda^D s}\, ds \right].$$

**Interpretation from (a):** since $\tau^D$ and $X$ are independent, we can condition on $\tau^D$ in the formula from (a). Using $\mathbb{P}(\tau^D \geq T) = e^{-\lambda^D T}$, $\mathbb{P}(\tau^D > s) = e^{-\lambda^D s}$, and the density $\lambda^D e^{-\lambda^D s}$, we get exactly the Feynman-Kac representation above. The key point is that the independence of $\tau^D$ and $X$ allows us to analytically integrate out the death time, removing the need to simulate $\tau^D$.

```python
np.random.seed(123)

npaths = 200000
paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths)

# Feynman-Kac: no need to simulate tau^D
discount = np.exp(-lambd * ts[:-1])[:, np.newaxis]  # e^{-lambda*t_i}, shape (nsteps, 1)
integrand = (lambd * np.maximum(K_D - paths[:-1], 0) - alpha) * discount * dt
integral = np.sum(integrand, axis=0)

terminal = np.maximum(K_mat - paths[-1], 0) * np.exp(-lambd * T)

payoff_1b = terminal + integral
price_1b = np.mean(payoff_1b)
se_1b = np.std(payoff_1b) / np.sqrt(npaths)
print(f"Q1(b) price (Feynman-Kac):       {price_1b:.4f} ± {1.96*se_1b:.4f}")
```

### Question 2(a) — Proof of (E1) and (E2)

Denote $E := \mathbb{E}^{\mathbb{Q}}_{i-1}[Y_{t_i}]$, $U := u^D(t_{i-1}, X_{t_{i-1}})$, $Y := Y_{t_{i-1}}$, $\Delta t := \Delta t_i$.

The implicit scheme reads:

$$Y = E - \alpha\Delta t + \Lambda^D(U - Y)(U - Y)\Delta t.$$

**Case 1: $U \geq Y$**, so $\Lambda^D(U-Y) = \bar\lambda^D$. Then

$$Y = E - \alpha\Delta t + \bar\lambda^D(U - Y)\Delta t \implies Y(1 + \bar\lambda^D\Delta t) = E - \alpha\Delta t + \bar\lambda^D U\Delta t.$$

Hence $Y = \frac{E - \alpha\Delta t + \bar\lambda^D U\Delta t}{1 + \bar\lambda^D\Delta t}$. The condition $U \geq Y$ becomes

$$U \geq \frac{E - \alpha\Delta t + \bar\lambda^D U\Delta t}{1 + \bar\lambda^D\Delta t} \iff U(1 + \bar\lambda^D\Delta t) \geq E - \alpha\Delta t + \bar\lambda^D U\Delta t \iff U \geq E - \alpha\Delta t.$$

**Case 2: $U < Y$**, so $\Lambda^D(U-Y) = \underline\lambda^D$. Similarly,

$$Y = \frac{E - \alpha\Delta t + \underline\lambda^D U\Delta t}{1 + \underline\lambda^D\Delta t}, \quad \text{and } U < Y \iff U < E - \alpha\Delta t.$$

In both cases the condition $U \geq Y$ is equivalent to $U \geq E - \alpha\Delta t$, which proves **(E1)**.

Substituting back into the two cases yields **(E2)** directly:

$$Y_{t_{i-1}} = \frac{E - \alpha\Delta t + \bar\lambda^D U\Delta t}{1 + \bar\lambda^D\Delta t}\,\mathbb{1}_{U \geq E - \alpha\Delta t} + \frac{E - \alpha\Delta t + \underline\lambda^D U\Delta t}{1 + \underline\lambda^D\Delta t}\,\mathbb{1}_{U < E - \alpha\Delta t}.\quad\square$$

### Question 2(b) — Implicit Euler BSDE scheme

We regress $Y_{t_i}$ on $X_{t_{i-1}}$ using piecewise-linear fit (10 knots between percentiles 1 and 99 of $X_{t_{i-1}}$) to estimate $\mathbb{E}^{\mathbb{Q}}_{i-1}[Y_{t_i}]$. Then we apply formula (E2).

```python
np.random.seed(0)

S0 = 100; vol = 0.3; T = 10; alpha = 3.0
K_mat = 90.0; K_D = 100.0
lambda_min = 0.005; lambda_max = 0.04
nsteps = int(np.round(T * 12))
ts = np.linspace(0, T, nsteps + 1)
dt = ts[1] - ts[0]

def u_mat(x):
    return np.maximum(K_mat - x, 0)

def u_D(x):
    return np.maximum(K_D - x, 0)

# --- BSDE regression step ---
npaths_reg = 100000
paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths_reg)

Y = u_mat(paths[-1])
regression_params = [None] * nsteps  # index j: regression at time t_j

for i in range(nsteps, 0, -1):
    j = i - 1
    X_prev = paths[j]
    xmin, xmax = np.percentile(X_prev, 1), np.percentile(X_prev, 99)
    xknots = np.linspace(xmin, xmax, 10)

    ps, fs = pwlinear_fit(X_prev, Y, xknots)
    regression_params[j] = (ps, xknots)

    EY = np.vstack([f(X_prev) for f in fs]).T @ ps
    U = u_D(paths[j])
    cond = U >= EY - alpha * dt

    Y = np.where(cond,
                 (EY - alpha*dt + lambda_max * U * dt) / (1 + lambda_max * dt),
                 (EY - alpha*dt + lambda_min * U * dt) / (1 + lambda_min * dt))

price_2b = np.mean(Y)
print(f"Q2(b) BSDE implicit (regression paths): {price_2b:.4f}")
```

### Question 2(c) — Independent simulation (lower bound)

We simulate fresh paths and use the regression from (b) to determine the optimal $\lambda$ at each step. The backward recursion uses the explicit formula with these fixed controls, giving an unbiased lower bound estimate.

```python
np.random.seed(99)
npaths_indep = 200000
paths2 = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths_indep)

V = u_mat(paths2[-1])
lambd_arr = np.where(u_D(paths2[-1]) >= V, lambda_max, lambda_min)

for i in range(nsteps - 1, -1, -1):
    p = 1 - np.exp(-lambd_arr * dt)
    V = p * u_D(paths2[i+1]) + (1 - p) * V - alpha * dt

    # determine lambda at time t_i using stored regression
    ps_i, xknots_i = regression_params[i]
    fs_i = pwlinear_basis(xknots_i)
    EY_hat = np.vstack([f(paths2[i]) for f in fs_i]).T @ ps_i

    lambd_arr = np.where(u_D(paths2[i]) >= EY_hat - alpha * dt, lambda_max, lambda_min)

price_2c = np.mean(V)
se_2c = np.std(V) / np.sqrt(npaths_indep)
print(f"Q2(c) BSDE lower bound (independent): {price_2c:.4f} ± {1.96*se_2c:.4f}")
```

### Question 2(d) — Comparison with deterministic mortality rates

For a deterministic death rate $\lambda^D$, the Feynman-Kac formula from Q1(b) gives

$$u(0, X_0) = \mathbb{E}\left[ u^{\mathrm{mat}}(X_T)\, e^{-\lambda^D T} + \int_0^T \big(\lambda^D\, u^D(s, X_s) - \alpha\big)\, e^{-\lambda^D s}\, ds \right].$$

The uncertain mortality price should satisfy $u_{\underline\lambda} \leq u_{\text{uncertain}} \leq u_{\bar\lambda}$ is NOT necessarily true in general. The uncertain mortality model picks the worst-case $\lambda$ at each time, so its price should be at least as large as the price under any fixed $\lambda$, i.e., $u_{\text{uncertain}} \geq \max(u_{\underline\lambda}, u_{\bar\lambda})$.

```python
np.random.seed(77)
npaths_det = 200000
paths_det = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths_det)

def feynman_kac_price(paths, lambd_det):
    discount = np.exp(-lambd_det * ts[:-1])[:, np.newaxis]
    integrand = (lambd_det * u_D(paths[:-1]) - alpha) * discount * dt
    integral = np.sum(integrand, axis=0)
    terminal = u_mat(paths[-1]) * np.exp(-lambd_det * T)
    return np.mean(terminal + integral), np.std(terminal + integral) / np.sqrt(paths.shape[1])

price_lmin, se_lmin = feynman_kac_price(paths_det, lambda_min)
price_lmax, se_lmax = feynman_kac_price(paths_det, lambda_max)

print(f"Deterministic lambda_min={lambda_min}: {price_lmin:.4f} ± {1.96*se_lmin:.4f}")
print(f"Deterministic lambda_max={lambda_max}: {price_lmax:.4f} ± {1.96*se_lmax:.4f}")
print(f"Uncertain mortality (BSDE):      {price_2c:.4f} ± {1.96*se_2c:.4f}")
print(f"\nAs expected, the uncertain mortality price >= max of the two deterministic prices.")
```

### Question 3(a) — Longstaff-Schwartz with piecewise-linear fit

### Question 3(b) — Variance comparison: BSDE vs Longstaff-Schwartz

We run the two-step procedure (regression + independent pricing) 20 times for each method, using 5,000 regression paths and 50,000 pricing paths.

```python
nruns = 20
npaths_reg = 5000
npaths_price = 50000

def bsde_two_step(seed):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=npaths_reg)
    Y = u_mat(paths[-1])
    reg = [None] * nsteps
    for i in range(nsteps, 0, -1):
        j = i - 1
        X_prev = paths[j]
        xmin, xmax = np.percentile(X_prev, 1), np.percentile(X_prev, 99)
        xknots = np.linspace(xmin, xmax, 10)
        ps, fs = pwlinear_fit(X_prev, Y, xknots)
        reg[j] = (ps, xknots)
        EY = np.vstack([f(X_prev) for f in fs]).T @ ps
        U = u_D(paths[j])
        cond = U >= EY - alpha * dt
        Y = np.where(cond,
                     (EY - alpha*dt + lambda_max * U * dt) / (1 + lambda_max * dt),
                     (EY - alpha*dt + lambda_min * U * dt) / (1 + lambda_min * dt))

    paths2 = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=npaths_price)
    V = u_mat(paths2[-1])
    lam = np.where(u_D(paths2[-1]) >= V, lambda_max, lambda_min)
    for i in range(nsteps - 1, -1, -1):
        p = 1 - np.exp(-lam * dt)
        V = p * u_D(paths2[i+1]) + (1-p) * V - alpha * dt
        ps_i, xk_i = reg[i]
        fs_i = pwlinear_basis(xk_i)
        EY_hat = np.vstack([f(paths2[i]) for f in fs_i]).T @ ps_i
        lam = np.where(u_D(paths2[i]) >= EY_hat - alpha * dt, lambda_max, lambda_min)
    return np.mean(V)

def ls_two_step(seed):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=npaths_reg)
    ls_reg = [None] * len(ts)
    V = u_mat(paths[-1])
    lam = np.where(u_D(paths[-1]) >= V, lambda_max, lambda_min)
    for i in range(len(ts)-2, 0, -1):
        dt_i = ts[i+1] - ts[i]
        p = 1 - np.exp(-lam * dt_i)
        V = p * u_D(paths[i+1]) + (1-p) * V - alpha * dt_i
        xmin, xmax = np.percentile(paths[i], 1), np.percentile(paths[i], 99)
        xknots = np.linspace(xmin, xmax, 10)
        ps, fs = pwlinear_fit(paths[i], V, xknots)
        ls_reg[i] = (ps, xknots)
        Vhat = np.vstack([f(paths[i]) for f in fs]).T @ ps
        lam = np.where(u_D(paths[i]) >= Vhat, lambda_max, lambda_min)

    paths2 = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=npaths_price)
    V = u_mat(paths2[-1])
    lam = np.where(u_D(paths2[-1]) >= V, lambda_max, lambda_min)
    for i in range(len(ts)-2, -1, -1):
        dt_i = ts[i+1] - ts[i]
        p = 1 - np.exp(-lam * dt_i)
        V = p * u_D(paths2[i+1]) + (1-p) * V - alpha * dt_i
        if ls_reg[i] is not None:
            ps_i, xk_i = ls_reg[i]
            fs_i = pwlinear_basis(xk_i)
            Vhat = np.vstack([f(paths2[i]) for f in fs_i]).T @ ps_i
            lam = np.where(u_D(paths2[i]) >= Vhat, lambda_max, lambda_min)
        else:
            lam = np.where(u_D(paths2[i]) >= V, lambda_max, lambda_min)
    return np.mean(V)

bsde_prices = [bsde_two_step(1000 + k) for k in range(nruns)]
ls_prices = [ls_two_step(2000 + k) for k in range(nruns)]

print(f"BSDE:  mean = {np.mean(bsde_prices):.4f}, std = {np.std(bsde_prices):.4f}, var = {np.var(bsde_prices):.6f}")
print(f"LS:    mean = {np.mean(ls_prices):.4f}, std = {np.std(ls_prices):.4f}, var = {np.var(ls_prices):.6f}")
print(f"Variance ratio (LS/BSDE): {np.var(ls_prices)/np.var(bsde_prices):.2f}")
```

### Question 3(c) — Fee $\alpha$ for a costless deal

We seek $\alpha$ such that $u(0, X_0) = 0$. Since the price is roughly affine and decreasing in $\alpha$ (the fee reduces the contract value), we can use bisection. We use the LS method for speed.

```python
def ls_price_alpha(alpha_val, seed=42):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=10000)
    ls_reg = [None] * len(ts)
    V = u_mat(paths[-1])
    lam = np.where(u_D(paths[-1]) >= V, lambda_max, lambda_min)
    for i in range(len(ts)-2, 0, -1):
        dt_i = ts[i+1] - ts[i]
        p = 1 - np.exp(-lam * dt_i)
        V = p * u_D(paths[i+1]) + (1-p) * V - alpha_val * dt_i
        xmin, xmax = np.percentile(paths[i], 1), np.percentile(paths[i], 99)
        xknots = np.linspace(xmin, xmax, 10)
        ps, fs = pwlinear_fit(paths[i], V, xknots)
        ls_reg[i] = (ps, xknots)
        Vhat = np.vstack([f(paths[i]) for f in fs]).T @ ps
        lam = np.where(u_D(paths[i]) >= Vhat, lambda_max, lambda_min)

    paths2 = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=100000)
    V = u_mat(paths2[-1])
    lam = np.where(u_D(paths2[-1]) >= V, lambda_max, lambda_min)
    for i in range(len(ts)-2, -1, -1):
        dt_i = ts[i+1] - ts[i]
        p = 1 - np.exp(-lam * dt_i)
        V = p * u_D(paths2[i+1]) + (1-p) * V - alpha_val * dt_i
        if ls_reg[i] is not None:
            ps_i, xk_i = ls_reg[i]
            fs_i = pwlinear_basis(xk_i)
            Vhat = np.vstack([f(paths2[i]) for f in fs_i]).T @ ps_i
            lam = np.where(u_D(paths2[i]) >= Vhat, lambda_max, lambda_min)
        else:
            lam = np.where(u_D(paths2[i]) >= V, lambda_max, lambda_min)
    return np.mean(V)

# bisection
a_lo, a_hi = 0.0, 10.0
for _ in range(20):
    a_mid = 0.5 * (a_lo + a_hi)
    p_mid = ls_price_alpha(a_mid)
    if p_mid > 0:
        a_lo = a_mid
    else:
        a_hi = a_mid

alpha_star = 0.5 * (a_lo + a_hi)
print(f"Q3(c) alpha for costless deal: {alpha_star:.4f}")
print(f"       price at alpha*:         {ls_price_alpha(alpha_star):.4f}")
```

### Question 4(a) — Optimal lapse as an American option

With $K_D = 0$, we have $u^D(t, x) = (0 - x)_+ = 0$, so the death/lapse payoff is zero. The subscriber can lapse (exercise) at any time $\tau \leq T$, receiving $u^D = 0$ and stopping fee payments. If the subscriber lapses optimally, they choose the stopping time $\tau$ that minimizes the cost to the issuer, or equivalently maximizes their own benefit.

The contract value from the issuer's perspective is

$$u(0, X_0) = \sup_{\tau \leq T} \mathbb{E}\left[ u^{\mathrm{mat}}(X_T)\,\mathbb{1}_{\tau \geq T} - \int_0^{\tau \wedge T} \alpha\, ds \right] = \sup_{\tau \leq T} \mathbb{E}\left[ u^{\mathrm{mat}}(X_T)\,\mathbb{1}_{\tau \geq T} - \alpha(\tau \wedge T) \right].$$

Since $u^D = 0$, the subscriber will choose to lapse (stop paying fees) when continuing is not beneficial. At each time $t$, the subscriber compares:
- **Continue:** pay $\alpha\,dt$ and keep the contract (value $u(t, X_t)$)
- **Lapse:** receive $u^D = 0$

The optimal stopping condition is: lapse when $u(t, X_t) \leq 0$, i.e., when the expected future payout is less than the remaining fees.

This is exactly an **American option** problem: the issuer prices the worst-case early termination, which amounts to pricing an American-style contract where at each time $t$ the holder can exercise (receive 0) or continue (keep paying $\alpha$ and receive $u^{\mathrm{mat}}(X_T)$ at maturity).

### Question 4(b) — American price from the uncertain lapse model

When $K_D = 0$ (so $u^D = 0$), the HJB equation becomes

$$\partial_t u + \tfrac{1}{2}\sigma^2 x^2 \partial_x^2 u - \alpha + \Lambda^D(t, -u)(-u) = 0.$$

The worst-case control picks $\Lambda^D = \bar\lambda^D$ when $u^D - u = -u \geq 0$ (i.e., $u \leq 0$) and $\Lambda^D = \underline\lambda^D$ when $u > 0$.

In the limit $\bar\lambda^D \to +\infty$ and $\underline\lambda^D \to 0$:

- When $u \leq 0$: the term $\bar\lambda^D(-u) \to +\infty$ unless $u = 0$, forcing $u \geq 0$.
- When $u > 0$: $\underline\lambda^D = 0$, so the lapse term vanishes and the PDE reduces to $\partial_t u + \tfrac{1}{2}\sigma^2 x^2 \partial_x^2 u - \alpha = 0$.

Combined, this gives the variational inequality of an American option:
$$\min\left(-\partial_t u - \tfrac{1}{2}\sigma^2 x^2 \partial_x^2 u + \alpha, \; u \right) = 0,$$
i.e., $u \geq 0$ and $\partial_t u + \tfrac{1}{2}\sigma^2 x^2 \partial_x^2 u - \alpha \leq 0$, with equality in each region.

So we can recover the American option price by running the uncertain lapse model code with $K_D = 0$, $\underline\lambda^D \approx 0$, and $\bar\lambda^D$ very large.

```python
np.random.seed(42)

K_D_lapse = 0.0
lambda_min_lapse = 1e-6
lambda_max_lapse = 1000.0

def u_D_lapse(x):
    return np.maximum(K_D_lapse - x, 0)  # = 0

# regression step
npaths = 5000
paths = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=npaths)
ls_reg_lapse = [None] * len(ts)

V = u_mat(paths[-1])
lam = np.where(u_D_lapse(paths[-1]) >= V, lambda_max_lapse, lambda_min_lapse)

for i in range(len(ts)-2, 0, -1):
    dt_i = ts[i+1] - ts[i]
    p = 1 - np.exp(-lam * dt_i)
    V = p * u_D_lapse(paths[i+1]) + (1-p) * V - alpha * dt_i
    xmin, xmax = np.percentile(paths[i], 1), np.percentile(paths[i], 99)
    xknots = np.linspace(xmin, xmax, 10)
    ps, fs = pwlinear_fit(paths[i], V, xknots)
    ls_reg_lapse[i] = (ps, xknots)
    Vhat = np.vstack([f(paths[i]) for f in fs]).T @ ps
    lam = np.where(u_D_lapse(paths[i]) >= Vhat, lambda_max_lapse, lambda_min_lapse)

# independent pricing
npaths2 = 100000
paths2 = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=npaths2)

V = u_mat(paths2[-1])
lam = np.where(u_D_lapse(paths2[-1]) >= V, lambda_max_lapse, lambda_min_lapse)

for i in range(len(ts)-2, -1, -1):
    dt_i = ts[i+1] - ts[i]
    p = 1 - np.exp(-lam * dt_i)
    V = p * u_D_lapse(paths2[i+1]) + (1-p) * V - alpha * dt_i
    if ls_reg_lapse[i] is not None:
        ps_i, xk_i = ls_reg_lapse[i]
        fs_i = pwlinear_basis(xk_i)
        Vhat = np.vstack([f(paths2[i]) for f in fs_i]).T @ ps_i
        lam = np.where(u_D_lapse(paths2[i]) >= Vhat, lambda_max_lapse, lambda_min_lapse)
    else:
        lam = np.where(u_D_lapse(paths2[i]) >= V, lambda_max_lapse, lambda_min_lapse)

price_4b = np.mean(V)
se_4b = np.std(V) / np.sqrt(npaths2)
print(f"Q4(b) Uncertain lapse price (lambda_max={lambda_max_lapse}): {price_4b:.4f} ± {1.96*se_4b:.4f}")
```

### Question 4(c) — Longstaff-Schwartz for the American option

With $u^D = 0$, the optimal lapse problem becomes: the subscriber pays fees $\alpha$ per unit time and receives $u^{\mathrm{mat}}(X_T)$ at maturity, with the option to stop (lapse) and receive $0$ at any time. The issuer's worst-case price is the American option

$$u(0, X_0) = \sup_{\tau \leq T} \mathbb{E}\left[-\alpha\,\tau + u^{\mathrm{mat}}(X_T)\,\mathbb{1}_{\tau = T}\right].$$

We implement Longstaff-Schwartz: at each step, compare the continuation value (keep paying fees) with the exercise value (lapse, receive 0).

```python
np.random.seed(55)

# --- Binomial tree (exact benchmark) ---
nsteps_tree = 120
dt_tree = T / nsteps_tree
u_tree = np.exp(vol * np.sqrt(dt_tree))
d_tree = 1.0 / u_tree
p_tree = (1 - d_tree) / (u_tree - d_tree)

S_mat_tree = S0 * u_tree ** np.arange(nsteps_tree, -nsteps_tree-1, -2)
V_tree = np.maximum(K_mat - S_mat_tree, 0)
for ii in range(nsteps_tree-1, -1, -1):
    V_tree = p_tree * V_tree[:-1] + (1-p_tree) * V_tree[1:]
    V_tree = np.maximum(V_tree - alpha * dt_tree, 0)

price_tree = V_tree[0]
print(f"Q4(c) American option (tree):    {price_tree:.4f}")

# --- American option via LS ---
npaths = 5000
paths = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=npaths)
am_reg = [None] * len(ts)

V = u_mat(paths[-1])

for i in range(len(ts)-2, 0, -1):
    dt_i = ts[i+1] - ts[i]
    V_cont = V - alpha * dt_i

    xmin, xmax = np.percentile(paths[i], 1), np.percentile(paths[i], 99)
    xknots = np.linspace(xmin, xmax, 10)
    ps, fs = pwlinear_fit(paths[i], V_cont, xknots)
    am_reg[i] = (ps, xknots)

    Vhat = np.vstack([f(paths[i]) for f in fs]).T @ ps
    V = np.where(Vhat >= 0, V_cont, 0.0)

# Independent pricing
npaths2 = 100000
paths2 = blackscholes_mc(S=S0, vol=vol, ts=ts, npaths=npaths2)
V = u_mat(paths2[-1])

for i in range(len(ts)-2, -1, -1):
    dt_i = ts[i+1] - ts[i]
    V_cont = V - alpha * dt_i

    if am_reg[i] is not None:
        ps_i, xk_i = am_reg[i]
        fs_i = pwlinear_basis(xk_i)
        Vhat = np.vstack([f(paths2[i]) for f in fs_i]).T @ ps_i
        V = np.where(Vhat >= 0, V_cont, 0.0)
    else:
        # at i=0, X_0 is deterministic: decision is uniform across paths
        V = V_cont if np.mean(V_cont) >= 0 else np.zeros_like(V_cont)

price_4c = np.mean(V)
se_4c = np.std(V) / np.sqrt(npaths2)
print(f"Q4(c) American option (LS):      {price_4c:.4f} \u00b1 {1.96*se_4c:.4f}")
print(f"Q4(b) Uncertain lapse model:     {price_4b:.4f} \u00b1 {1.96*se_4b:.4f}")
print(f"\nThe LS and tree prices are close. The uncertain lapse model converges to the American price")
print(f"as lambda_max -> infinity and lambda_min -> 0.")
```
