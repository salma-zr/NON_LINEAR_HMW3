# Uncertain Mortality Model vs American Options

#### Re-insurance Deals

In this assignment, we study the *Uncertain Mortality Model* for the pricing of reinsurance deals. For the sake of simplicity, in this assignment we only consider one type of default: the risk of death.

We consider the following reinsurance product:

- At maturity, if the insurance subscriber is alive, the issuer delivers a put on the underlying $X$

$$
u^\textrm{mat}( x) = (K_\textrm{mat} - x)_+.
$$

- At the time of death, if it is before the maturity, the issuer delivers an exit payoff, typically another put on the underlying $X$.

$$
u^D(t, x) = (K_D - x)_+.
$$

- The subscriber pays a constant fee $α\Delta t$ at every time step until death or the maturity or the product

- The insurance sells a large number of these contracts to subscribers. We assume that the times of death of the subscribers $\tau^D$ are independent, and identically distributed, and also independent of the underlying's stock price.

- We assume that the underlying's risk neutral price dynamics is the Black-Scholes model with zero interest rate/repo/dividend yield

$$
dX_t = \sigma X_t d W_t.
$$

#### The Insurers' Approach and Risk-Neutral Pricing

This contract shows two types of risk: the times of death of the subscribers and the changes in the price of the underlying. 

In this case, the issuer can apply the insurer's approach to the risk of death times, i.e., the law of large numbers. The more people buy the contract, the less risk.

Choosing a risk-neutral measure under which the death times $\tau^D$ have the same distribution as under the historical probability measure is equivalent to applying the arbitrage-pricing approach to the financial risk insurer's rule on the risk of death. The price of the contract is then

$$
u(t, x) = \mathbb{E}^\mathbb{Q}_{t, x} \left[u^\textrm{mat}(X_T) \mathbb{1}_{\tau^D \geq T} -\int_t^T\alpha\mathbb{1}_{\tau^D>s}ds+ u^D(\tau^D, X_{\tau^D}) \mathbb{1}_{\tau^D < T} \right].
$$

#### Deterministic Death Rate

If the death intensity is a deterministic function $\lambda_t^D$ (i.e. $\tau^D$ has an exponential distribution with time-dependent intensity $\lambda_t^D$), then we have seen that $u$ is the solution to the linear PDE

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial_x^2 u -\alpha + \lambda_t^D \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat}.
\end{array}\right.
$$

#### Uncertain Mortality Model

If the death rate is uncertain, we assume that it is adapted (i.e., it does not look into the future) and belongs to a moving corridor
$\left[\underline{\lambda}_t, \overline{\lambda}_t\right]$. The most conservative way to price the contract is to compute the (financially) worst death rate process $\lambda_t^D$ as being chosen so as to maximize the value of the contract. The resulting HJB equation is

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial^2_x u -\alpha + \Lambda^D(t, u^D - u) \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat},
\end{array}\right.
$$
where the function $\Lambda^D$ is defined by
$$
\Lambda^D (t, y) = \left\{\begin{array}{l}
    \overline{\lambda}^D_t \quad \textrm{if} \ y \geq 0, \\
    \underline{\lambda}^D_t \quad \textrm{otherwise.}
\end{array}\right.
$$

## Link with $1$-BSDE and Numerical Schemes

From the Pardoux-Peng theorem, we know that the solution $u(0, x)$ can be represented as the solution $Y_0^x$ to the $1$-BSDE

$$
dY_t = -f(t, X_t, Y_t, Z_t) \, dt + Z_t \, dW_t,
$$

with the terminal condition $Y_T = u^\textrm{mat} (X_T)$, where $X_0 = x$ and

$$
f(t, x, y, z) = -\alpha + \Lambda^D(t, u^D(t, x) - y) \cdot (u^D(t, x) - y).
$$


## BSDE Discretization

### Explicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i + \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1} , X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ]\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \right) \Delta t_i\\
=& -\alpha\Delta t_i + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \overline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]}\\
& + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \underline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) < \mathbb{E}_{i-1}[Y_{t_i}]}.
\end{align*}

### Implicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [Y_{t_i}] -\alpha\Delta t_i+ \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}}\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}} \right) \Delta t_i.
\end{align*}

The implicit scheme involves $Y_{t_{i - 1}}$ on both sides of the equation, which generally requires a root-finding routine to find $Y_{t_{i - 1}}$. However in this specific case, it can be shown that 

\begin{equation}u^D(t_{i-1}, X_{t_{i-1}}) \geq Y_{t_{i-1}}\quad\text{if and only if}\quad u^D(t_{i-1}, X_{t_{i-1}}) \geq \mathbb{E}^{\mathbb{Q}}_{i-1}\left[Y_{t_i}\right]-\alpha\Delta t_i. \qquad (E1) \end{equation}

Thus

\begin{split}
Y_{t_{i - 1}} =& \frac{1}{1 + \overline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] - \alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}\\
& + \frac{1}{1 + \underline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D(t_{i - 1}, X_{t_{i - 1}}) < \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}. \qquad (E2)
\end{split}



```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
def blackscholes_mc(S=100, vol=0.2, r=0, q=0, ts=np.linspace(0, 1, 13), npaths=10):
    """Generate Monte-Carlo paths in Black-Scholes model.

    Parameters
    ----------
    S: scalar
        The spot price of the underlying security.
    vol: scalar
        The implied Black-Scholes volatility.
    r: scalar
        The annualized risk-free interest rate, continuously compounded.
    q: scalar
        The annualized continuous dividend yield.
    ts: array_like
        The time steps of the simualtion
    npaths: int
        the number of paths to simulate

    Returns
    -------
    paths: ndarray
        The Monte-Carlo paths.
    """
    nsteps = len(ts) - 1
    ts = np.asarray(ts, dtype=float)[:, np.newaxis]
    W = np.cumsum(
        np.vstack(
            (
                np.zeros((1, npaths), dtype=float),
                np.random.randn(nsteps, npaths) * np.sqrt(np.diff(ts, axis=0)),
            )
        ),
        axis=0,
    )
    paths = np.exp(-0.5 * vol**2 * ts + vol * W) * S * np.exp((r - q) * ts)
    return paths

def blackscholes_price(K, T, S, vol, r=0, q=0, callput='call'):
    """Compute the call/put option price in the Black-Scholes model
    
    Parameters
    ----------
    K: scalar or array_like
        The strike of the option.
    T: scalar or array_like
        The maturity of the option, expressed in years (e.g. 0.25 for 3-month and 2 for 2 years)
    S: scalar or array_like
        The current price of the underlying asset.
    vol: scalar or array_like
        The implied Black-Scholes volatility.
    r: scalar or array_like
        The annualized risk-free interest rate, continuously compounded.
    q: scalar or array_like
        The annualized continuous dividend yield.
    callput: str
        Must be either 'call' or 'put'.

    Returns
    -------
    price: scalar or array_like
        The price of the option.

    Examples
    --------
    >>> blackscholes_price(95, 0.25, 100, 0.2, r=0.05, callput='put')
    1.5342604771222823
    """
    F = S*np.exp((r-q)*T)
    v = np.sqrt(vol**2*T)
    d1 = np.log(F/K)/v + 0.5*v
    d2 = d1 - v
    try:
        opttype = {'call':1, 'put':-1}[callput.lower()]
    except:
        raise ValueError('The value of callput must be either "call" or "put".')
    price = opttype*(F*norm.cdf(opttype*d1)-K*norm.cdf(opttype*d2))*np.exp(-r*T)
    return price


def pwlinear_basis(xknots):
    """Basis that represent a piecewise linear function with given knots"""
    fs = [lambda x: np.ones_like(x, dtype=float), lambda x: x - xknots[0]]
    fs.extend(
        [lambda x, a=xknots[i]: np.maximum(x - a, 0) for i in range(len(xknots))]
    )
    return fs


def pwlinear_fit(xdata, ydata, xknots):
    """Fit a piecewise linear function with xknots to xdata and ydata"""
    fs = pwlinear_basis(xknots)
    A = np.vstack([f(xdata) for f in fs]).T
    ps = np.linalg.lstsq(A, ydata, rcond=None)[0]
    return ps, fs


def pwlinear_predict(x, ps, xknots):
    fs = pwlinear_basis(xknots)
    return np.vstack([f(x) for f in fs]).T @ ps
```

- If the death occurs at the maturity, we assume that the death comes first and the death payment $u^D$ will be made.

$$\left\{\begin{array}{l}\partial_t u + \frac{1}{2}\sigma^2 x^2 \partial_x^2 u - \alpha + \lambda^D \cdot (u^D - u) = 0\\

**(a)** With $\tau^D\sim\mathrm{Exp}(\lambda)$ independent of $X$, a monthly-grid simulation is:

- simulate one Black–Scholes path $(X_{t_i})$,
- simulate $\tau^D$ and map it to the payment date $t_{\lceil \tau^D/\Delta t\rceil}$,
- add $u^{\mathrm{mat}}(X_T)$ if $\tau^D\ge T$, otherwise add $u^D(X_{t_{\lceil \tau^D/\Delta t\rceil}})$,
- subtract the fees $\alpha\Delta t$ for months whose end date is strictly before $\tau^D$.

**(b)** Conditioning on the path of $X$ and integrating out the death event yields (constant $\lambda$)
\[
\mathbb E\big[u^{\mathrm{mat}}(X_T)\mathbf 1_{\tau^D\ge T}\mid X\big]=e^{-\lambda T}u^{\mathrm{mat}}(X_T),
\]
\[
\mathbb E\big[u^D(X_{t_{i+1}})\mathbf 1_{\tau^D\in(t_i,t_{i+1}]}\mid X\big]=\big(e^{-\lambda t_i}-e^{-\lambda t_{i+1}}\big)u^D(X_{t_{i+1}}),
\]
\[
\mathbb E\big[\alpha\Delta t\,\mathbf 1_{\tau^D>t_{i+1}}\mid X\big]=\alpha\Delta t\,e^{-\lambda t_{i+1}}.
\]
So one can simulate only $X$ and weight the cashflows by these survival/death probabilities; this is the discrete-time counterpart of the Feynman–Kac representation of the PDE.

```python
# note that lambda is a reserved keyword in python, so we use lambd instead
def exponential_samples(n, lambd):
    """Generate n samples from exponential distribution with rate lambd"""
    return -np.log(np.random.rand(n))/lambd
```

```python
S0 = 100.0
sigma = 0.3
T = 10.0
K_mat = 90.0
K_D = 100.0
alpha = 3.0
lambda_det = 0.025

ts = np.linspace(0.0, T, int(np.round(T * 12)) + 1)
dt = ts[1] - ts[0]

u_mat = lambda x: np.maximum(K_mat - x, 0.0)
u_D = lambda x: np.maximum(K_D - x, 0.0)


def q1a_direct_mc(npaths=200000, seed=0):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths)
    tau = exponential_samples(npaths, lambda_det)

    alive = tau >= T
    pay_idx = np.searchsorted(ts, tau, side="left")
    pay_idx = np.minimum(pay_idx, len(ts) - 1)

    payoff = np.where(alive, u_mat(paths[-1]), u_D(paths[pay_idx, np.arange(npaths)]))

    n_fees = np.searchsorted(ts[1:], tau, side="left")
    payoff = payoff - alpha * dt * n_fees

    price = float(np.mean(payoff))
    se = float(np.std(payoff, ddof=1) / np.sqrt(npaths))
    return price, se


def q1b_average_death_events_mc(npaths=200000, seed=1):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths)

    surv_T = np.exp(-lambda_det * T)
    w_death = np.exp(-lambda_det * ts[:-1]) - np.exp(-lambda_det * ts[1:])  # length N
    death_term = np.sum(w_death[:, None] * u_D(paths[1:]), axis=0)

    fee_weight_sum = float(np.sum(np.exp(-lambda_det * ts[1:])))

    payoff = surv_T * u_mat(paths[-1]) + death_term - alpha * dt * fee_weight_sum

    price = float(np.mean(payoff))
    se = float(np.std(payoff, ddof=1) / np.sqrt(npaths))
    return price, se


q1a_direct_mc(), q1b_average_death_events_mc()
```

Let $m:=\mathbb E^{\mathbb Q}_{i-1}[Y_{t_i}]$ and $h:=u^D(t_{i-1},X_{t_{i-1}})$. The implicit step is
\[
Y=m-\alpha\Delta t+\Lambda(h-Y)(h-Y)\Delta t.
\]

**(a) Proof of (E1).**

If $h\ge Y$, then $\Lambda=\overline\lambda$ and
\[(1+\overline\lambda\Delta t)Y=m-\alpha\Delta t+\overline\lambda\Delta t\,h.
\]
Hence
\[h-Y=\frac{h-(m-\alpha\Delta t)}{1+\overline\lambda\Delta t},
\]
so $h\ge Y\iff h\ge m-\alpha\Delta t$. The case $h<Y$ is identical with $\underline\lambda$.

**(a) Proof of (E2).** Using (E1), pick $\overline\lambda$ on $\{h\ge m-\alpha\Delta t\}$ and $\underline\lambda$ otherwise, then solve the corresponding linear equation for $Y$.

```python
lambda_min = 0.005
lambda_max = 0.04


def bsde_implicit_regression_models(
    S0,
    vol,
    T,
    alpha,
    lambda_min,
    lambda_max,
    u_mat,
    u_D,
    n_reg=100000,
    nknots=10,
    pmin=1,
    pmax=99,
    seed=0,
):
    ts = np.linspace(0.0, T, int(np.round(T * 12)) + 1)
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_reg)

    ps_list = [None] * len(ts)
    knots_list = [None] * len(ts)

    Y = u_mat(paths[-1])
    for i in range(len(ts) - 1, 0, -1):
        dt = ts[i] - ts[i - 1]
        x = paths[i - 1]

        lo, hi = np.percentile(x, [pmin, pmax])
        xknots = np.linspace(lo, hi, nknots)
        ps, _ = pwlinear_fit(x, Y, xknots)
        ps_list[i - 1] = ps
        knots_list[i - 1] = xknots

        m = pwlinear_predict(x, ps, xknots)
        h = u_D(x)
        use_max = h >= m - alpha * dt

        Y = np.where(
            use_max,
            (m - alpha * dt + lambda_max * dt * h) / (1.0 + lambda_max * dt),
            (m - alpha * dt + lambda_min * dt * h) / (1.0 + lambda_min * dt),
        )

    price_in_sample = float(np.mean(Y))
    return (ts, ps_list, knots_list), price_in_sample


def bsde_implicit_lower_bound(
    S0,
    vol,
    T,
    alpha,
    lambda_min,
    lambda_max,
    u_mat,
    u_D,
    models,
    n_price=100000,
    seed=0,
):
    ts, ps_list, knots_list = models

    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_price)

    Y = u_mat(paths[-1])
    for i in range(len(ts) - 1, 0, -1):
        dt = ts[i] - ts[i - 1]
        x = paths[i - 1]
        m = pwlinear_predict(x, ps_list[i - 1], knots_list[i - 1])
        h = u_D(x)
        use_max = h >= m - alpha * dt

        Y = np.where(
            use_max,
            (m - alpha * dt + lambda_max * dt * h) / (1.0 + lambda_max * dt),
            (m - alpha * dt + lambda_min * dt * h) / (1.0 + lambda_min * dt),
        )

    price = float(np.mean(Y))
    se = float(np.std(Y, ddof=1) / np.sqrt(n_price))
    return price, se


models_bsde, price_bsde_insample = bsde_implicit_regression_models(
    S0=S0,
    vol=sigma,
    T=T,
    alpha=alpha,
    lambda_min=lambda_min,
    lambda_max=lambda_max,
    u_mat=u_mat,
    u_D=u_D,
    n_reg=100000,
    nknots=10,
    pmin=1,
    pmax=99,
    seed=2026,
)

price_bsde_lb, se_bsde_lb = bsde_implicit_lower_bound(
    S0=S0,
    vol=sigma,
    T=T,
    alpha=alpha,
    lambda_min=lambda_min,
    lambda_max=lambda_max,
    u_mat=u_mat,
    u_D=u_D,
    models=models_bsde,
    n_price=100000,
    seed=2027,
)

price_bsde_insample, (price_bsde_lb, se_bsde_lb)
```

For the regression step, we approximate the continuation value by a piecewise-linear function of the state $X_{t_i}$:
\[

\]
fitted by least squares on a spline-like basis with 10 knots chosen between the 1% and 99% percentiles of $X_{t_i}$.

```python
def ls_uncertain_intensity_price(
    S0,
    vol,
    T,
    alpha,
    lambda_min,
    lambda_max,
    u_mat,
    u_D,
    n_reg=5000,
    n_price=100000,
    nknots=10,
    pmin=1,
    pmax=99,
    seed=0,
):
    ts = np.linspace(0.0, T, int(np.round(T * 12)) + 1)

    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_reg)

    ps_list = [None] * len(ts)
    knots_list = [None] * len(ts)

    V = u_mat(paths[-1])
    lambd = np.where(u_D(paths[-1]) >= V, lambda_max, lambda_min)

    for i in range(len(ts) - 2, 0, -1):
        dt = ts[i + 1] - ts[i]
        p = 1.0 - np.exp(-lambd * dt)
        V = p * u_D(paths[i + 1]) + (1.0 - p) * V - alpha * dt

        x = paths[i]
        lo, hi = np.percentile(x, [pmin, pmax])
        xknots = np.linspace(lo, hi, nknots)
        ps, _ = pwlinear_fit(x, V, xknots)
        ps_list[i] = ps
        knots_list[i] = xknots

        cont = pwlinear_predict(x, ps, xknots)
        lambd = np.where(u_D(paths[i]) >= cont, lambda_max, lambda_min)

    np.random.seed(seed + 10_000)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_price)

    V = u_mat(paths[-1])
    lambd = np.where(u_D(paths[-1]) >= V, lambda_max, lambda_min)

    for i in range(len(ts) - 2, -1, -1):
        dt = ts[i + 1] - ts[i]
        p = 1.0 - np.exp(-lambd * dt)
        V = p * u_D(paths[i + 1]) + (1.0 - p) * V - alpha * dt

        if i > 0:
            cont = pwlinear_predict(paths[i], ps_list[i], knots_list[i])
            lambd = np.where(u_D(paths[i]) >= cont, lambda_max, lambda_min)

    price = float(np.mean(V))
    se = float(np.std(V, ddof=1) / np.sqrt(n_price))
    return price, se


S0 = 100.0
vol = 0.3
T = 10.0
alpha = 3.0
lambda_min = 0.005
lambda_max = 0.04
K_mat = 90.0
K_D = 100.0

u_mat = lambda x: np.maximum(K_mat - x, 0.0)
u_D = lambda x: np.maximum(K_D - x, 0.0)

price_ls, se_ls = ls_uncertain_intensity_price(
    S0=S0,
    vol=vol,
    T=T,
    alpha=alpha,
    lambda_min=lambda_min,
    lambda_max=lambda_max,
    u_mat=u_mat,
    u_D=u_D,
    n_reg=5000,
    n_price=100000,
    nknots=10,
    pmin=1,
    pmax=99,
    seed=42,
)
price_ls, se_ls
```

**(b)** For each method, one run consists of:

- a regression run (to learn the continuation / conditional expectations),
- an independent pricing run (to remove regression bias and measure Monte Carlo variance).

Repeating this 20 times gives an empirical variance of the price estimator.

```python
def bsde_implicit_regression_models(
    S0,
    vol,
    T,
    alpha,
    lambda_min,
    lambda_max,
    u_mat,
    u_D,
    n_reg=5000,
    nknots=10,
    pmin=1,
    pmax=99,
    seed=0,
):
    ts = np.linspace(0.0, T, int(np.round(T * 12)) + 1)
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_reg)

    ps_list = [None] * len(ts)
    knots_list = [None] * len(ts)

    Y = u_mat(paths[-1])
    for i in range(len(ts) - 1, 0, -1):
        dt = ts[i] - ts[i - 1]
        x = paths[i - 1]

        lo, hi = np.percentile(x, [pmin, pmax])
        xknots = np.linspace(lo, hi, nknots)
        ps, _ = pwlinear_fit(x, Y, xknots)
        ps_list[i - 1] = ps
        knots_list[i - 1] = xknots

        m = pwlinear_predict(x, ps, xknots)
        h = u_D(x)
        use_max = h >= m - alpha * dt

        Y = np.where(
            use_max,
            (m - alpha * dt + lambda_max * dt * h) / (1.0 + lambda_max * dt),
            (m - alpha * dt + lambda_min * dt * h) / (1.0 + lambda_min * dt),
        )

    return ts, ps_list, knots_list


def bsde_implicit_lower_bound(
    S0,
    vol,
    T,
    alpha,
    lambda_min,
    lambda_max,
    u_mat,
    u_D,
    models,
    n_price=50000,
    seed=0,
):
    ts, ps_list, knots_list = models

    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_price)

    Y = u_mat(paths[-1])
    for i in range(len(ts) - 1, 0, -1):
        dt = ts[i] - ts[i - 1]
        x = paths[i - 1]
        m = pwlinear_predict(x, ps_list[i - 1], knots_list[i - 1])
        h = u_D(x)
        use_max = h >= m - alpha * dt

        Y = np.where(
            use_max,
            (m - alpha * dt + lambda_max * dt * h) / (1.0 + lambda_max * dt),
            (m - alpha * dt + lambda_min * dt * h) / (1.0 + lambda_min * dt),
        )

    price = float(np.mean(Y))
    se = float(np.std(Y, ddof=1) / np.sqrt(n_price))
    return price, se


def variance_experiment(nrep=20):
    prices_bsde = np.zeros(nrep)
    prices_ls = np.zeros(nrep)

    for k in range(nrep):
        seed = 1000 + 37 * k

        models = bsde_implicit_regression_models(
            S0=S0,
            vol=vol,
            T=T,
            alpha=alpha,
            lambda_min=lambda_min,
            lambda_max=lambda_max,
            u_mat=u_mat,
            u_D=u_D,
            n_reg=5000,
            nknots=10,
            pmin=1,
            pmax=99,
            seed=seed,
        )
        prices_bsde[k] = bsde_implicit_lower_bound(
            S0=S0,
            vol=vol,
            T=T,
            alpha=alpha,
            lambda_min=lambda_min,
            lambda_max=lambda_max,
            u_mat=u_mat,
            u_D=u_D,
            models=models,
            n_price=50000,
            seed=seed + 10_000,
        )[0]

        prices_ls[k] = ls_uncertain_intensity_price(
            S0=S0,
            vol=vol,
            T=T,
            alpha=alpha,
            lambda_min=lambda_min,
            lambda_max=lambda_max,
            u_mat=u_mat,
            u_D=u_D,
            n_reg=5000,
            n_price=50000,
            nknots=10,
            pmin=1,
            pmax=99,
            seed=seed,
        )[0]

    out = {
        "BSDE_mean": float(np.mean(prices_bsde)),
        "BSDE_std": float(np.std(prices_bsde, ddof=1)),
        "LS_mean": float(np.mean(prices_ls)),
        "LS_std": float(np.std(prices_ls, ddof=1)),
    }
    return out, prices_bsde, prices_ls

stats_var, prices_bsde, prices_ls = variance_experiment(nrep=20)
stats_var
```

**(c)** We solve for $\alpha^*$ such that $u(0,S_0;\alpha^*)\approx 0$ by bisection (the price is decreasing in $\alpha$ since fees enter with a minus sign).

```python
def alpha_costless_bisect(alpha_lo, alpha_hi, tol=5e-2, maxit=10, seed=777):
    f_lo = ls_uncertain_intensity_price(
        S0=S0,
        vol=vol,
        T=T,
        alpha=alpha_lo,
        lambda_min=lambda_min,
        lambda_max=lambda_max,
        u_mat=u_mat,
        u_D=u_D,
        n_reg=5000,
        n_price=20000,
        nknots=10,
        pmin=1,
        pmax=99,
        seed=seed,
    )[0]
    f_hi = ls_uncertain_intensity_price(
        S0=S0,
        vol=vol,
        T=T,
        alpha=alpha_hi,
        lambda_min=lambda_min,
        lambda_max=lambda_max,
        u_mat=u_mat,
        u_D=u_D,
        n_reg=5000,
        n_price=20000,
        nknots=10,
        pmin=1,
        pmax=99,
        seed=seed,
    )[0]

    if f_lo < 0.0:
        return alpha_lo, f_lo
    if f_hi > 0.0:
        return alpha_hi, f_hi

    lo, hi = alpha_lo, alpha_hi
    for _ in range(maxit):
        mid = 0.5 * (lo + hi)
        f_mid = ls_uncertain_intensity_price(
            S0=S0,
            vol=vol,
            T=T,
            alpha=mid,
            lambda_min=lambda_min,
            lambda_max=lambda_max,
            u_mat=u_mat,
            u_D=u_D,
            n_reg=5000,
            n_price=20000,
            nknots=10,
            pmin=1,
            pmax=99,
            seed=seed,
        )[0]

        if abs(f_mid) <= tol:
            return mid, f_mid
        if f_mid > 0.0:
            lo = mid
        else:
            hi = mid

    return mid, f_mid

alpha_star, price_at_star = alpha_costless_bisect(alpha_lo=0.0, alpha_hi=10.0, tol=5e-2, maxit=8, seed=999)
alpha_star, price_at_star
```

<b style="color:darkorange">Question 4.</b> Set $K_D=0$, and all the other parameters remain the same, in particualr $\alpha=3$. Then the default event is equivalent to lapse (with no mortality payoff). It has been observed that insurance subscribers do not lapse optimally, which explains the use of the uncertain lapse model with minimum and maximum lapse rates $\underline{\lambda}$, $\overline{\lambda}$. 

<b>(a).</b> If insurance subscribers were to lapse optimally, i.e., exercise optimally their option to lapse, how would you price the contract?

<b>(b).</b> Explain how you can get this price (from optimal exercise) in the uncertain lapse model. Reuse the code provided in Question 3 to show your result (you may have to change some parameters in the code).

<b>(c).</b> Implement Longstaff-Schwartz algorithm to price the American option with payoff $u^D=0$ (Always perform a second indepedent run to get a clean lower bound price). Compare the price from uncertain lapse model in part (b) with the American price.

For $K_D=0$ we have $u^D\equiv 0$ (lapse with no payoff).

**(a)** If the subscriber can lapse optimally (i.e. choose a stopping time $\tau\le T$), the time-$t$ value solves the optimal stopping problem
\[
 u(t,x)=\sup_{\tau\in\mathcal T_{t,T}}\,\mathbb E_{t,x}\Big[u^{\mathrm{mat}}(X_T)\mathbf 1_{\tau=T}-\int_t^{\tau}\alpha\,ds\Big].
\]
In PDE form (zero rates), this is the obstacle problem
\[
\max\Big(\partial_t u+\tfrac12\sigma^2x^2\partial_{xx}u-\alpha,\,-u\Big)=0,\qquad u(T,\cdot)=u^{\mathrm{mat}}.
\]

**(b)** In the uncertain lapse (intensity control) model with $\lambda_t\in[\underline\lambda,\overline\lambda]$, the HJB becomes
\[
\partial_t u+\tfrac12\sigma^2x^2\partial_{xx}u-\alpha+\Lambda(t,-u)\,(-u)=0,
\]
so $\lambda_t=\overline\lambda$ is chosen when $u\le 0$ and $\lambda_t=\underline\lambda$ when $u>0$. Taking $\underline\lambda=0$ and $\overline\lambda\to\infty$ enforces $u\ge 0$ and recovers the obstacle problem above.

```python
# Q4(b): uncertain lapse model via intensity control (reusing Q3 code)

K_mat_4 = 90.0
K_D_4 = 0.0
alpha_4 = 3.0
lambda_min_4 = 0.0
lambda_max_4 = 50.0  # large -> close to optimal lapse

u_mat_4 = lambda x: np.maximum(K_mat_4 - x, 0.0)
u_D_4 = lambda x: np.zeros_like(x, dtype=float)

price_uncertain_lapse, se_uncertain_lapse = ls_uncertain_intensity_price(
    S0=100.0,
    vol=0.3,
    T=10.0,
    alpha=alpha_4,
    lambda_min=lambda_min_4,
    lambda_max=lambda_max_4,
    u_mat=u_mat_4,
    u_D=u_D_4,
    n_reg=5000,
    n_price=50000,
    nknots=10,
    pmin=1,
    pmax=99,
    seed=123,
)
price_uncertain_lapse, se_uncertain_lapse
```

**(c)** American (optimal lapse) pricing with Longstaff–Schwartz (Bermudan on monthly grid):
\[
V_{t_i}=\max\Big(0,\ \mathbb E[V_{t_{i+1}}\mid X_{t_i}]-\alpha\Delta t\Big),\qquad V_T=u^{\mathrm{mat}}(X_T).
\]
Second independent run gives a clean lower bound by fixing the exercise rule from the regression.

```python
def american_abandonment_ls(
    S0,
    vol,
    T,
    alpha,
    u_mat,
    n_reg=5000,
    n_price=50000,
    nknots=10,
    pmin=1,
    pmax=99,
    seed=0,
):
    ts = np.linspace(0.0, T, int(np.round(T * 12)) + 1)
    dt = ts[1] - ts[0]

    # regression run: estimate continuation x -> E[V_{i+1}|X_i=x]
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_reg)

    ps_list = [None] * len(ts)
    knots_list = [None] * len(ts)

    V = u_mat(paths[-1])
    for i in range(len(ts) - 2, -1, -1):
        x = paths[i]
        lo, hi = np.percentile(x, [pmin, pmax])
        xknots = np.linspace(lo, hi, nknots)
        ps, _ = pwlinear_fit(x, V, xknots)
        ps_list[i] = ps
        knots_list[i] = xknots

        cont_hat = pwlinear_predict(x, ps, xknots) - alpha * dt
        V = np.maximum(0.0, cont_hat)

    # pricing run: apply fixed exercise rule (lower bound)
    np.random.seed(seed + 10_000)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_price)

    prem = np.zeros(n_price)
    active = np.ones(n_price, dtype=bool)

    for i in range(len(ts) - 1):
        x = paths[i]
        cont_hat = pwlinear_predict(x, ps_list[i], knots_list[i]) - alpha * dt

        stop = active & (cont_hat <= 0.0)
        active[stop] = False

        prem[active] += alpha * dt

    payoff = np.where(active, u_mat(paths[-1]) - prem, -prem)

    price = float(np.mean(payoff))
    se = float(np.std(payoff, ddof=1) / np.sqrt(n_price))
    return price, se

price_american, se_american = american_abandonment_ls(
    S0=100.0,
    vol=0.3,
    T=10.0,
    alpha=alpha_4,
    u_mat=u_mat_4,
    n_reg=5000,
    n_price=50000,
    nknots=10,
    pmin=1,
    pmax=99,
    seed=321,
)

(price_uncertain_lapse, se_uncertain_lapse), (price_american, se_american)
```
