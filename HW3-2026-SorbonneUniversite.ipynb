# Uncertain Mortality Model vs American Options

#### Re-insurance Deals

In this assignment, we study the *Uncertain Mortality Model* for the pricing of reinsurance deals. For the sake of simplicity, in this assignment we only consider one type of default: the risk of death.

We consider the following reinsurance product:

- At maturity, if the insurance subscriber is alive, the issuer delivers a put on the underlying $X$

$$
u^\textrm{mat}( x) = (K_\textrm{mat} - x)_+.
$$

- At the time of death, if it is before the maturity, the issuer delivers an exit payoff, typically another put on the underlying $X$.

$$
u^D(t, x) = (K_D - x)_+.
$$

- The subscriber pays a constant fee $α\Delta t$ at every time step until death or the maturity or the product

- The insurance sells a large number of these contracts to subscribers. We assume that the times of death of the subscribers $\tau^D$ are independent, and identically distributed, and also independent of the underlying's stock price.

- We assume that the underlying's risk neutral price dynamics is the Black-Scholes model with zero interest rate/repo/dividend yield

$$
dX_t = \sigma X_t d W_t.
$$

#### The Insurers' Approach and Risk-Neutral Pricing

This contract shows two types of risk: the times of death of the subscribers and the changes in the price of the underlying. 

In this case, the issuer can apply the insurer's approach to the risk of death times, i.e., the law of large numbers. The more people buy the contract, the less risk.

Choosing a risk-neutral measure under which the death times $\tau^D$ have the same distribution as under the historical probability measure is equivalent to applying the arbitrage-pricing approach to the financial risk insurer's rule on the risk of death. The price of the contract is then

$$
u(t, x) = \mathbb{E}^\mathbb{Q}_{t, x} \left[u^\textrm{mat}(X_T) \mathbb{1}_{\tau^D \geq T} -\int_t^T\alpha\mathbb{1}_{\tau^D>s}ds+ u^D(\tau^D, X_{\tau^D}) \mathbb{1}_{\tau^D < T} \right].
$$

#### Deterministic Death Rate

If the death intensity is a deterministic function $\lambda_t^D$ (i.e. $\tau^D$ has an exponential distribution with time-dependent intensity $\lambda_t^D$), then we have seen that $u$ is the solution to the linear PDE

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial_x^2 u -\alpha + \lambda_t^D \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat}.
\end{array}\right.
$$

#### Uncertain Mortality Model

If the death rate is uncertain, we assume that it is adapted (i.e., it does not look into the future) and belongs to a moving corridor
$\left[\underline{\lambda}_t, \overline{\lambda}_t\right]$. The most conservative way to price the contract is to compute the (financially) worst death rate process $\lambda_t^D$ as being chosen so as to maximize the value of the contract. The resulting HJB equation is

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial^2_x u -\alpha + \Lambda^D(t, u^D - u) \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat},
\end{array}\right.
$$
where the function $\Lambda^D$ is defined by
$$
\Lambda^D (t, y) = \left\{\begin{array}{l}
    \overline{\lambda}^D_t \quad \textrm{if} \ y \geq 0, \\
    \underline{\lambda}^D_t \quad \textrm{otherwise.}
\end{array}\right.
$$

## Link with $1$-BSDE and Numerical Schemes

From the Pardoux-Peng theorem, we know that the solution $u(0, x)$ can be represented as the solution $Y_0^x$ to the $1$-BSDE

$$
dY_t = -f(t, X_t, Y_t, Z_t) \, dt + Z_t \, dW_t,
$$

with the terminal condition $Y_T = u^\textrm{mat} (X_T)$, where $X_0 = x$ and

$$
f(t, x, y, z) = -\alpha + \Lambda^D(t, u^D(t, x) - y) \cdot (u^D(t, x) - y).
$$


## BSDE Discretization

### Explicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i + \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1} , X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ]\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \right) \Delta t_i\\
=& -\alpha\Delta t_i + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \overline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]}\\
& + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \underline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) < \mathbb{E}_{i-1}[Y_{t_i}]}.
\end{align*}

### Implicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [Y_{t_i}] -\alpha\Delta t_i+ \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}}\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}} \right) \Delta t_i.
\end{align*}

The implicit scheme involves $Y_{t_{i - 1}}$ on both sides of the equation, which generally requires a root-finding routine to find $Y_{t_{i - 1}}$. However in this specific case, it can be shown that 

\begin{equation}u^D(t_{i-1}, X_{t_{i-1}}) \geq Y_{t_{i-1}}\quad\text{if and only if}\quad u^D(t_{i-1}, X_{t_{i-1}}) \geq \mathbb{E}^{\mathbb{Q}}_{i-1}\left[Y_{t_i}\right]-\alpha\Delta t_i. \qquad (E1) \end{equation}

Thus

\begin{split}
Y_{t_{i - 1}} =& \frac{1}{1 + \overline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] - \alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}\\
& + \frac{1}{1 + \underline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D(t_{i - 1}, X_{t_{i - 1}}) < \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}. \qquad (E2)
\end{split}



```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
def blackscholes_mc(S=100, vol=0.2, r=0, q=0, ts=np.linspace(0, 1, 13), npaths=10):
    """Generate Monte-Carlo paths in Black-Scholes model.

    Parameters
    ----------
    S: scalar
        The spot price of the underlying security.
    vol: scalar
        The implied Black-Scholes volatility.
    r: scalar
        The annualized risk-free interest rate, continuously compounded.
    q: scalar
        The annualized continuous dividend yield.
    ts: array_like
        The time steps of the simualtion
    npaths: int
        the number of paths to simulate

    Returns
    -------
    paths: ndarray
        The Monte-Carlo paths.
    """
    nsteps = len(ts) - 1
    ts = np.asarray(ts, dtype=float)[:, np.newaxis]
    W = np.cumsum(
        np.vstack(
            (
                np.zeros((1, npaths), dtype=float),
                np.random.randn(nsteps, npaths) * np.sqrt(np.diff(ts, axis=0)),
            )
        ),
        axis=0,
    )
    paths = np.exp(-0.5 * vol**2 * ts + vol * W) * S * np.exp((r - q) * ts)
    return paths

def blackscholes_price(K, T, S, vol, r=0, q=0, callput='call'):
    """Compute the call/put option price in the Black-Scholes model
    
    Parameters
    ----------
    K: scalar or array_like
        The strike of the option.
    T: scalar or array_like
        The maturity of the option, expressed in years (e.g. 0.25 for 3-month and 2 for 2 years)
    S: scalar or array_like
        The current price of the underlying asset.
    vol: scalar or array_like
        The implied Black-Scholes volatility.
    r: scalar or array_like
        The annualized risk-free interest rate, continuously compounded.
    q: scalar or array_like
        The annualized continuous dividend yield.
    callput: str
        Must be either 'call' or 'put'.

    Returns
    -------
    price: scalar or array_like
        The price of the option.

    Examples
    --------
    >>> blackscholes_price(95, 0.25, 100, 0.2, r=0.05, callput='put')
    1.5342604771222823
    """
    F = S*np.exp((r-q)*T)
    v = np.sqrt(vol**2*T)
    d1 = np.log(F/K)/v + 0.5*v
    d2 = d1 - v
    try:
        opttype = {'call':1, 'put':-1}[callput.lower()]
    except:
        raise ValueError('The value of callput must be either "call" or "put".')
    price = opttype*(F*norm.cdf(opttype*d1)-K*norm.cdf(opttype*d2))*np.exp(-r*T)
    return price


def pwlinear_basis(xknots):
    """Basis that represent a piecewise linear function with given knots"""
    fs = [lambda x: np.ones_like(x, dtype=float), lambda x: x - xknots[0]]
    fs.extend([lambda x, a=xknots[i]: np.maximum(x - a, 0) for i in range(len(xknots))])
    return fs


def pwlinear_fit(xdata, ydata, xknots):
    """Fit a piecewise linear function with xknots to xdata and ydata"""
    fs = pwlinear_basis(xknots)
    A = np.vstack([f(xdata) for f in fs]).T
    ps = np.linalg.lstsq(A, ydata, rcond=None)[0]
    return ps, fs
```

<b>Note.</b> To be more specific about the payouts of the product, we make the following assumptions:

- We assume that all transactions occur at the end of the month, i.e. the insurance subsriber is expected to pay a fee of $\alpha/12$ at the end of each month; if the subscriber dies in the middle of the month, the death payment will be made at the end of that month. 

- If the death occurs at the maturity, we assume that the death comes first and the death payment $u^D$ will be made.

```python
# note that lambda is a reserved keyword in python, so we use lambd instead
def exponential_samples(n, lambd):
    """Generate n samples from exponential distribution with rate lambd"""
    return -np.log(np.random.rand(n))/lambd
```

### Answer (Q1)

**(a)** On the monthly grid $t_i=i/12$, simulate one equity path $(X_{t_i})_{i=0}^n$ and one death time $\tau\sim\mathrm{Exp}(\lambda^D)$, independent. With end-of-month settlement, if $\tau\in(t_{k-1},t_k]$ the death payoff is paid at $t_k$.
The pathwise cashflow is

$$
\Pi=\begin{cases}
 u^{\mathrm{mat}}(X_T)-\alpha\Delta t\,n, & \tau>T,\\
 u^D(X_{t_k})-\alpha\Delta t\,k, & \tau\in(t_{k-1},t_k],\ k\le n.
\end{cases}
$$

**(b)** On the grid, $\mathbb P(\tau>t_i)=e^{-\lambda^D t_i}$ and $\mathbb P(\tau\in(t_i,t_{i+1}])=e^{-\lambda^D t_i}-e^{-\lambda^D t_{i+1}}$. Conditioning on the equity path and averaging over $\tau$ gives

$$
\begin{aligned}
 u(0,S_0)=\mathbb E\Big[&e^{-\lambda^D T}u^{\mathrm{mat}}(X_T)+\sum_{i=0}^{n-1}e^{-\lambda^D t_i}(1-e^{-\lambda^D\Delta t})\,u^D(X_{t_{i+1}})\\
 &-\sum_{i=0}^{n-1}e^{-\lambda^D t_i}\,\alpha\Delta t\Big],
\end{aligned}
$$

which is the Feynman–Kac representation of the linear PDE (killing term $\lambda^D$), and is obtained from (a) by integrating out the exponential law of $\tau$.


```python
# Q1(a)-(b): deterministic intensity pricing

S0 = 100.0
vol = 0.3
T = 10.0
ts = np.linspace(0.0, T, int(np.round(12 * T)) + 1)
dt = ts[1] - ts[0]

K_mat = 90.0
K_D = 100.0
alpha = 3.0
lambda_D = 0.025


def u_mat(x):
    return np.maximum(K_mat - x, 0.0)


def u_D(x):
    return np.maximum(K_D - x, 0.0)


# (a) direct simulation of tau
np.random.seed(0)
npaths = 100000
X = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths)
tau = exponential_samples(npaths, lambda_D)

n = len(ts) - 1
k = np.searchsorted(ts, tau, side="left")  # k=n+1 means tau>T
alive = k > n

payoff = np.empty(npaths, dtype=float)
payoff[alive] = u_mat(X[-1, alive]) - alpha * dt * n

j = np.flatnonzero(~alive)
kk = k[j].astype(int)
payoff[j] = u_D(X[kk, j]) - alpha * dt * kk

u0_a = float(np.mean(payoff))
se_a = float(np.std(payoff, ddof=1) / np.sqrt(npaths))
print(f"(a) direct tau: u(0,S0)≈{u0_a:.4f} ± {1.96*se_a:.4f} (95% CI)")


# (b) average over death events (no tau simulation)
q = np.exp(-lambda_D * dt)
w = q ** np.arange(n)

term_mat = (q**n) * u_mat(X[-1])
term_death = (w[:, None] * (1.0 - q) * u_D(X[1:])).sum(axis=0)
term_fee = (w * alpha * dt).sum()

u0_b = float(np.mean(term_mat + term_death - term_fee))
print(f"(b) averaged deaths: u(0,S0)≈{u0_b:.4f}")
print(f"difference (b)-(a): {u0_b-u0_a:.4f}")
```

### Answer (Q2)

**(a)** Write $m:=\mathbb E_{i-1}[Y_{t_i}]$, $\Delta t:=\Delta t_i$, and $u:=u^D(t_{i-1},X_{t_{i-1}})$. The implicit step is

$$
Y_{t_{i-1}}=m-\alpha\Delta t+\Lambda^D(t_{i-1},u-Y_{t_{i-1}})(u-Y_{t_{i-1}})\Delta t.
$$

If $u\ge Y_{t_{i-1}}$, then $\Lambda^D=\overline\lambda$ and

$$
(1+\overline\lambda\Delta t)Y_{t_{i-1}}=m-\alpha\Delta t+\overline\lambda\Delta t\,u,
\qquad
u:=u-Y_{t_{i-1}}=\frac{u-(m-\alpha\Delta t)}{1+\overline\lambda\Delta t}.
$$

Hence $u\ge Y_{t_{i-1}}\iff \nu\ge0\iff u\ge m-\alpha\Delta t$. The case $u<Y_{t_{i-1}}$ is identical with $\underline\lambda$. This proves (E1), and plugging the corresponding $\Lambda^D\in\{\underline\lambda,\overline\lambda\}$ into the linear equation gives (E2).


```python
# Q2(b)-(d): implicit BSDE scheme + independent lower bound + deterministic benchmarks

S0 = 100.0
vol = 0.3
T = 10.0
ts = np.linspace(0.0, T, int(np.round(12 * T)) + 1)

dt = ts[1] - ts[0]
alpha = 3.0
lambda_min = 0.005
lambda_max = 0.04

K_mat = 90.0
K_D = 100.0


def u_mat(x):
    return np.maximum(K_mat - x, 0.0)


def u_D(x):
    return np.maximum(K_D - x, 0.0)


def pwlinear_fit_fast(x, y, nknots=10, pmin=1.0, pmax=99.0):
    x = np.asarray(x)
    y = np.asarray(y)
    xmin, xmax = np.percentile(x, [pmin, pmax])
    knots = np.linspace(xmin, xmax, nknots)

    x0 = knots[0]
    A = np.empty((x.size, 2 + nknots), dtype=float)
    A[:, 0] = 1.0
    A[:, 1] = x - x0
    A[:, 2:] = np.maximum(x[:, None] - knots[None, :], 0.0)
    beta = np.linalg.lstsq(A, y, rcond=None)[0]
    return knots, beta


def pwlinear_pred(x, knots, beta):
    x = np.asarray(x)
    x0 = knots[0]
    out = beta[0] + beta[1] * (x - x0)
    out = out + (np.maximum(x[:, None] - knots[None, :], 0.0) @ beta[2:])
    return out


# (b) regression run with 100,000 paths
np.random.seed(0)
ntrain = 100000
X = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=ntrain)
Y = u_mat(X[-1])

models = [None] * len(ts)
for i in range(len(ts) - 2, -1, -1):
    knots, beta = pwlinear_fit_fast(X[i], Y, nknots=10)
    models[i] = (knots, beta)
    m = pwlinear_pred(X[i], knots, beta)

    switch = u_D(X[i]) >= m - alpha * dt
    Ynew = np.empty_like(m)
    Ynew[switch] = (m[switch] - alpha * dt + lambda_max * dt * u_D(X[i, switch])) / (1.0 + lambda_max * dt)
    Ynew[~switch] = (m[~switch] - alpha * dt + lambda_min * dt * u_D(X[i, ~switch])) / (1.0 + lambda_min * dt)
    Y = Ynew

u0_bsde = float(np.mean(Y))
print(f"BSDE implicit Euler (training sample): u(0,S0)≈{u0_bsde:.4f}")


# (c) independent run: plug in the learned conditional expectations (lower bound)
np.random.seed(1)
X = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=ntrain)
Y = u_mat(X[-1])

for i in range(len(ts) - 2, -1, -1):
    knots, beta = models[i]
    m = pwlinear_pred(X[i], knots, beta)

    switch = u_D(X[i]) >= m - alpha * dt
    Ynew = np.empty_like(m)
    Ynew[switch] = (m[switch] - alpha * dt + lambda_max * dt * u_D(X[i, switch])) / (1.0 + lambda_max * dt)
    Ynew[~switch] = (m[~switch] - alpha * dt + lambda_min * dt * u_D(X[i, ~switch])) / (1.0 + lambda_min * dt)
    Y = Ynew

u0_lb = float(np.mean(Y))
print(f"BSDE independent lower bound: u(0,S0)≥{u0_lb:.4f}")


# (d) deterministic intensity prices (monthly discretization; average over death events)

def deterministic_price(lambd, npaths=100000, seed=2):
    np.random.seed(seed)
    X = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths)

    q = np.exp(-lambd * dt)
    n = len(ts) - 1
    w_surv = q ** n
    w = q ** np.arange(n)

    term_mat = w_surv * u_mat(X[-1])
    term_death = (w[:, None] * (1.0 - q) * u_D(X[1:])).sum(axis=0)
    term_fee = (w * alpha * dt).sum()

    return float(np.mean(term_mat + term_death - term_fee))

u_min = deterministic_price(lambda_min)
u_max = deterministic_price(lambda_max)
print(f"Deterministic intensity lambda_min={lambda_min}: u(0,S0)≈{u_min:.4f}")
print(f"Deterministic intensity lambda_max={lambda_max}: u(0,S0)≈{u_max:.4f}")
```

**(b)** For each method, use a first run to learn the regression functions $(\widehat c_i)_i$, then apply the induced control on an independent run to get a clean lower bound. Repeating this 20 times gives an empirical variance.

**(c)** The fair fee $\alpha^*$ solves $u(0,S_0;\alpha^*)\approx 0$. Since the price is decreasing in $\alpha$, a bisection on $\alpha$ using the lower-bound estimator works.


```python
# Longstaff-Schwartz-like algorithm for Uncertain Mortality Rate Model (piecewise-linear regression)
S = 100.0
vol = 0.3
T = 10.0
ts = np.linspace(0.0, T, int(np.round(T * 12)) + 1)
alpha = 3.0
lambda_min = 0.005
lambda_max = 0.04

K_mat = 90.0
K_D = 100.0


def u_mat(x):
    return np.maximum(K_mat - x, 0.0)


def u_D(x):
    return np.maximum(K_D - x, 0.0)


def pwlinear_fit_fast(x, y, nknots=10, pmin=1.0, pmax=99.0):
    x = np.asarray(x)
    y = np.asarray(y)
    xmin, xmax = np.percentile(x, [pmin, pmax])
    knots = np.linspace(xmin, xmax, nknots)

    x0 = knots[0]
    A = np.empty((x.size, 2 + nknots), dtype=float)
    A[:, 0] = 1.0
    A[:, 1] = x - x0
    A[:, 2:] = np.maximum(x[:, None] - knots[None, :], 0.0)
    beta = np.linalg.lstsq(A, y, rcond=None)[0]
    return knots, beta


def pwlinear_pred(x, knots, beta):
    x = np.asarray(x)
    x0 = knots[0]
    out = beta[0] + beta[1] * (x - x0)
    out = out + (np.maximum(x[:, None] - knots[None, :], 0.0) @ beta[2:])
    return out


# first Monte Carlo run (regression)
nreg = 5000
X = blackscholes_mc(S=S, vol=vol, r=0, q=0, ts=ts, npaths=nreg)
V = u_mat(X[-1])
lambd = np.where(u_D(X[-1]) >= V, lambda_max, lambda_min)

models = [None] * len(ts)
for i in range(len(ts) - 2, -1, -1):
    dt = ts[i + 1] - ts[i]
    p = 1.0 - np.exp(-lambd * dt)
    V = p * u_D(X[i + 1]) + (1.0 - p) * V - alpha * dt

    knots, beta = pwlinear_fit_fast(X[i], V, nknots=10)
    models[i] = (knots, beta)
    Vhat = pwlinear_pred(X[i], knots, beta)
    lambd = np.where(u_D(X[i]) >= Vhat, lambda_max, lambda_min)


# independent simulation (lower bound)
np.random.seed(1)
npr = 100000
X = blackscholes_mc(S=S, vol=vol, r=0, q=0, ts=ts, npaths=npr)
V = u_mat(X[-1])
lambd = np.where(u_D(X[-1]) >= V, lambda_max, lambda_min)

for i in range(len(ts) - 2, -1, -1):
    dt = ts[i + 1] - ts[i]
    p = 1.0 - np.exp(-lambd * dt)
    V = p * u_D(X[i + 1]) + (1.0 - p) * V - alpha * dt

    knots, beta = models[i]
    Vhat = pwlinear_pred(X[i], knots, beta)
    lambd = np.where(u_D(X[i]) >= Vhat, lambda_max, lambda_min)

np.mean(V)
```

```python
# Q3(b)-(c): variance comparison and fair fee alpha*

def ls_lower_bound(alpha, nreg=5000, npr=50000, seed=0):
    np.random.seed(seed)

    X = blackscholes_mc(S=S, vol=vol, r=0, q=0, ts=ts, npaths=nreg)
    V = u_mat(X[-1])
    lambd = np.where(u_D(X[-1]) >= V, lambda_max, lambda_min)

    models = [None] * len(ts)
    for i in range(len(ts) - 2, -1, -1):
        dt = ts[i + 1] - ts[i]
        p = 1.0 - np.exp(-lambd * dt)
        V = p * u_D(X[i + 1]) + (1.0 - p) * V - alpha * dt

        knots, beta = pwlinear_fit_fast(X[i], V, nknots=10)
        models[i] = (knots, beta)
        Vhat = pwlinear_pred(X[i], knots, beta)
        lambd = np.where(u_D(X[i]) >= Vhat, lambda_max, lambda_min)

    np.random.seed(seed + 1)
    X = blackscholes_mc(S=S, vol=vol, r=0, q=0, ts=ts, npaths=npr)
    V = u_mat(X[-1])
    lambd = np.where(u_D(X[-1]) >= V, lambda_max, lambda_min)

    for i in range(len(ts) - 2, -1, -1):
        dt = ts[i + 1] - ts[i]
        p = 1.0 - np.exp(-lambd * dt)
        V = p * u_D(X[i + 1]) + (1.0 - p) * V - alpha * dt

        knots, beta = models[i]
        Vhat = pwlinear_pred(X[i], knots, beta)
        lambd = np.where(u_D(X[i]) >= Vhat, lambda_max, lambda_min)

    return float(np.mean(V))


def bsde_lower_bound(alpha, nreg=5000, npr=50000, seed=0):
    np.random.seed(seed)

    X = blackscholes_mc(S=S, vol=vol, r=0, q=0, ts=ts, npaths=nreg)
    Y = u_mat(X[-1])

    models = [None] * len(ts)
    for i in range(len(ts) - 2, -1, -1):
        dt = ts[i + 1] - ts[i]
        knots, beta = pwlinear_fit_fast(X[i], Y, nknots=10)
        models[i] = (knots, beta)
        m = pwlinear_pred(X[i], knots, beta)

        switch = u_D(X[i]) >= m - alpha * dt
        Y = np.empty_like(m)
        Y[switch] = (m[switch] - alpha * dt + lambda_max * dt * u_D(X[i, switch])) / (1.0 + lambda_max * dt)
        Y[~switch] = (m[~switch] - alpha * dt + lambda_min * dt * u_D(X[i, ~switch])) / (1.0 + lambda_min * dt)

    np.random.seed(seed + 1)
    X = blackscholes_mc(S=S, vol=vol, r=0, q=0, ts=ts, npaths=npr)
    Y = u_mat(X[-1])

    for i in range(len(ts) - 2, -1, -1):
        dt = ts[i + 1] - ts[i]
        knots, beta = models[i]
        m = pwlinear_pred(X[i], knots, beta)

        switch = u_D(X[i]) >= m - alpha * dt
        Ynew = np.empty_like(m)
        Ynew[switch] = (m[switch] - alpha * dt + lambda_max * dt * u_D(X[i, switch])) / (1.0 + lambda_max * dt)
        Ynew[~switch] = (m[~switch] - alpha * dt + lambda_min * dt * u_D(X[i, ~switch])) / (1.0 + lambda_min * dt)
        Y = Ynew

    return float(np.mean(Y))


alpha0 = 3.0
nrep = 20
vals_bsde = np.array([bsde_lower_bound(alpha0, nreg=5000, npr=50000, seed=1000 + k) for k in range(nrep)])
vals_ls = np.array([ls_lower_bound(alpha0, nreg=5000, npr=50000, seed=2000 + k) for k in range(nrep)])

print(f"BSDE lower bound: mean={vals_bsde.mean():.4f}, var={vals_bsde.var(ddof=1):.6f}")
print(f"LS-like lower bound: mean={vals_ls.mean():.4f}, var={vals_ls.var(ddof=1):.6f}")

# approximate alpha* such that u(0,S0)=0
seed = 777
p2 = ls_lower_bound(2.0, nreg=5000, npr=50000, seed=seed)
p4 = ls_lower_bound(4.0, nreg=5000, npr=50000, seed=seed)
alpha_star = 2.0 + (0.0 - p2) * (4.0 - 2.0) / (p4 - p2)
print(f"alpha*=~{alpha_star:.4f} (secant from alpha=2,4; same random numbers)")
```

<b style="color:darkorange">Question 4.</b> Set $K_D=0$, and all the other parameters remain the same, in particualr $\alpha=3$. Then the default event is equivalent to lapse (with no mortality payoff). It has been observed that insurance subscribers do not lapse optimally, which explains the use of the uncertain lapse model with minimum and maximum lapse rates $\underline{\lambda}$, $\overline{\lambda}$. 

<b>(a).</b> If insurance subscribers were to lapse optimally, i.e., exercise optimally their option to lapse, how would you price the contract?

<b>(b).</b> Explain how you can get this price (from optimal exercise) in the uncertain lapse model. Reuse the code provided in Question 3 to show your result (you may have to change some parameters in the code).

<b>(c).</b> Implement Longstaff-Schwartz algorithm to price the American option with payoff $u^D=0$ (Always perform a second indepedent run to get a clean lower bound price). Compare the price from uncertain lapse model in part (b) with the American price.

### Answer (Q4)

**(a)** With $K_D=0$, lapsing at a stopping time $\tau\le T$ gives payoff $0$ and stops the fee stream. The price at $(t,x)$ is the optimal stopping value

$$
V(t,x)=\sup_{\tau\in\mathcal T_{t,T}}\,\mathbb E_{t,x}\Big[u^{\mathrm{mat}}(X_T)\,\mathbf 1_{\{\tau=T\}}-\int_t^{\tau}\alpha\,ds\Big].
$$

On a monthly grid $(t_i)$ with $\Delta t=t_{i+1}-t_i$, this is the Bermudan recursion

$$
V_n(x)=u^{\mathrm{mat}}(x),\qquad V_i(x)=\max\Big(0,\,\mathbb E\big[V_{i+1}(X_{t_{i+1}})\mid X_{t_i}=x\big]-\alpha\Delta t\Big).
$$

**(b)** In the uncertain lapse model, one keeps the same HJB as before with $u^D\equiv 0$.
The optimal control for the intensity is

$$
\lambda_t=\overline\lambda\,\mathbf 1_{\{0\ge V(t,X_t)\}}+\underline\lambda\,\mathbf 1_{\{0< V(t,X_t)\}},
$$

i.e. use the fastest lapse rate when continuing has nonpositive value. When $\underline\lambda=0$ and $\overline\lambda\to+\infty$, the controlled intensity reproduces the optimal stopping rule above (immediate lapse in the region $V\le0$).


```python
# Q4(b)-(c): uncertain lapse vs optimal lapse (American/Bermudan)

def pwlinear_fit_fast(x, y, nknots=10, pmin=1.0, pmax=99.0):
    x = np.asarray(x)
    y = np.asarray(y)
    xmin, xmax = np.percentile(x, [pmin, pmax])
    knots = np.linspace(xmin, xmax, nknots)

    x0 = knots[0]
    A = np.empty((x.size, 2 + nknots), dtype=float)
    A[:, 0] = 1.0
    A[:, 1] = x - x0
    A[:, 2:] = np.maximum(x[:, None] - knots[None, :], 0.0)
    beta = np.linalg.lstsq(A, y, rcond=None)[0]
    return knots, beta


def pwlinear_pred(x, knots, beta):
    x = np.asarray(x)
    x0 = knots[0]
    out = beta[0] + beta[1] * (x - x0)
    out = out + (np.maximum(x[:, None] - knots[None, :], 0.0) @ beta[2:])
    return out


def price_uncertain_lapse(alpha, nreg=5000, npr=50000, seed=0):
    np.random.seed(seed)
    S0, vol, T = 100.0, 0.3, 10.0
    ts = np.linspace(0.0, T, int(np.round(12 * T)) + 1)
    dt = ts[1] - ts[0]
    lambda_min, lambda_max = 0.005, 0.04
    K_mat = 90.0

    def u_mat(x):
        return np.maximum(K_mat - x, 0.0)

    # regression run
    X = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=nreg)
    V = u_mat(X[-1])
    lambd = np.where(0.0 >= V, lambda_max, lambda_min)

    models = [None] * len(ts)
    for i in range(len(ts) - 2, -1, -1):
        p = 1.0 - np.exp(-lambd * dt)
        V = (1.0 - p) * V - alpha * dt

        knots, beta = pwlinear_fit_fast(X[i], V, nknots=10)
        models[i] = (knots, beta)
        Vhat = pwlinear_pred(X[i], knots, beta)
        lambd = np.where(0.0 >= Vhat, lambda_max, lambda_min)

    # independent pricing run (lower bound)
    np.random.seed(seed + 1)
    X = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npr)
    V = u_mat(X[-1])
    lambd = np.where(0.0 >= V, lambda_max, lambda_min)

    for i in range(len(ts) - 2, -1, -1):
        p = 1.0 - np.exp(-lambd * dt)
        V = (1.0 - p) * V - alpha * dt

        knots, beta = models[i]
        Vhat = pwlinear_pred(X[i], knots, beta)
        lambd = np.where(0.0 >= Vhat, lambda_max, lambda_min)

    return float(np.mean(V))


def price_optimal_lapse_american(alpha, nreg=5000, npr=50000, seed=0):
    np.random.seed(seed)
    S0, vol, T = 100.0, 0.3, 10.0
    ts = np.linspace(0.0, T, int(np.round(12 * T)) + 1)
    dt = ts[1] - ts[0]
    K_mat = 90.0

    def u_mat(x):
        return np.maximum(K_mat - x, 0.0)

    # regression run
    X = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=nreg)
    V = u_mat(X[-1])

    models = [None] * len(ts)
    for i in range(len(ts) - 2, -1, -1):
        knots, beta = pwlinear_fit_fast(X[i], V, nknots=10)
        models[i] = (knots, beta)
        cont_hat = pwlinear_pred(X[i], knots, beta) - alpha * dt

        exercise = cont_hat <= 0.0
        V = np.where(exercise, 0.0, V - alpha * dt)

    # independent pricing run (lower bound)
    np.random.seed(seed + 1)
    X = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npr)

    alive = np.ones(npr, dtype=bool)
    val = np.zeros(npr, dtype=float)

    for i in range(len(ts) - 1):
        knots, beta = models[i]
        cont_hat = pwlinear_pred(X[i], knots, beta) - alpha * dt

        stop = alive & (cont_hat <= 0.0)
        alive[stop] = False

        cont = alive
        val[cont] -= alpha * dt

    val[alive] += u_mat(X[-1, alive])
    return float(np.mean(val))


alpha = 3.0
p_uncertain = price_uncertain_lapse(alpha, nreg=5000, npr=50000, seed=123)
p_american = price_optimal_lapse_american(alpha, nreg=5000, npr=50000, seed=123)
print(f"Uncertain lapse (K_D=0), lower bound: {p_uncertain:.4f}")
print(f"Optimal lapse (American/Bermudan), lower bound: {p_american:.4f}")
```
