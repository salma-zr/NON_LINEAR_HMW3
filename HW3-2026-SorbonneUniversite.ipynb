# Uncertain Mortality Model vs American Options

#### Re-insurance Deals

In this assignment, we study the *Uncertain Mortality Model* for the pricing of reinsurance deals. For the sake of simplicity, in this assignment we only consider one type of default: the risk of death.

We consider the following reinsurance product:

- At maturity, if the insurance subscriber is alive, the issuer delivers a put on the underlying $X$

$$
u^\textrm{mat}( x) = (K_\textrm{mat} - x)_+.
$$

- At the time of death, if it is before the maturity, the issuer delivers an exit payoff, typically another put on the underlying $X$.

$$
u^D(t, x) = (K_D - x)_+.
$$

- The subscriber pays a constant fee $α\Delta t$ at every time step until death or the maturity or the product

- The insurance sells a large number of these contracts to subscribers. We assume that the times of death of the subscribers $\tau^D$ are independent, and identically distributed, and also independent of the underlying's stock price.

- We assume that the underlying's risk neutral price dynamics is the Black-Scholes model with zero interest rate/repo/dividend yield

$$
dX_t = \sigma X_t d W_t.
$$

#### The Insurers' Approach and Risk-Neutral Pricing

This contract shows two types of risk: the times of death of the subscribers and the changes in the price of the underlying. 

In this case, the issuer can apply the insurer's approach to the risk of death times, i.e., the law of large numbers. The more people buy the contract, the less risk.

Choosing a risk-neutral measure under which the death times $\tau^D$ have the same distribution as under the historical probability measure is equivalent to applying the arbitrage-pricing approach to the financial risk insurer's rule on the risk of death. The price of the contract is then

$$
u(t, x) = \mathbb{E}^\mathbb{Q}_{t, x} \left[u^\textrm{mat}(X_T) \mathbb{1}_{\tau^D \geq T} -\int_t^T\alpha\mathbb{1}_{\tau^D>s}ds+ u^D(\tau^D, X_{\tau^D}) \mathbb{1}_{\tau^D < T} \right].
$$

#### Deterministic Death Rate

If the death intensity is a deterministic function $\lambda_t^D$ (i.e. $\tau^D$ has an exponential distribution with time-dependent intensity $\lambda_t^D$), then we have seen that $u$ is the solution to the linear PDE

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial_x^2 u -\alpha + \lambda_t^D \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat}.
\end{array}\right.
$$

#### Uncertain Mortality Model

If the death rate is uncertain, we assume that it is adapted (i.e., it does not look into the future) and belongs to a moving corridor
$\left[\underline{\lambda}_t, \overline{\lambda}_t\right]$. The most conservative way to price the contract is to compute the (financially) worst death rate process $\lambda_t^D$ as being chosen so as to maximize the value of the contract. The resulting HJB equation is

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial^2_x u -\alpha + \Lambda^D(t, u^D - u) \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat},
\end{array}\right.
$$
where the function $\Lambda^D$ is defined by
$$
\Lambda^D (t, y) = \left\{\begin{array}{l}
    \overline{\lambda}^D_t \quad \textrm{if} \ y \geq 0, \\
    \underline{\lambda}^D_t \quad \textrm{otherwise.}
\end{array}\right.
$$

## Link with $1$-BSDE and Numerical Schemes

From the Pardoux-Peng theorem, we know that the solution $u(0, x)$ can be represented as the solution $Y_0^x$ to the $1$-BSDE

$$
dY_t = -f(t, X_t, Y_t, Z_t) \, dt + Z_t \, dW_t,
$$

with the terminal condition $Y_T = u^\textrm{mat} (X_T)$, where $X_0 = x$ and

$$
f(t, x, y, z) = -\alpha + \Lambda^D(t, u^D(t, x) - y) \cdot (u^D(t, x) - y).
$$


## BSDE Discretization

### Explicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i + \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1} , X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ]\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \right) \Delta t_i\\
=& -\alpha\Delta t_i + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \overline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]}\\
& + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \underline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) < \mathbb{E}_{i-1}[Y_{t_i}]}.
\end{align*}

### Implicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [Y_{t_i}] -\alpha\Delta t_i+ \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}}\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}} \right) \Delta t_i.
\end{align*}

The implicit scheme involves $Y_{t_{i - 1}}$ on both sides of the equation, which generally requires a root-finding routine to find $Y_{t_{i - 1}}$. However in this specific case, it can be shown that 

\begin{equation}u^D(t_{i-1}, X_{t_{i-1}}) \geq Y_{t_{i-1}}\quad\text{if and only if}\quad u^D(t_{i-1}, X_{t_{i-1}}) \geq \mathbb{E}^{\mathbb{Q}}_{i-1}\left[Y_{t_i}\right]-\alpha\Delta t_i. \qquad (E1) \end{equation}

Thus

\begin{split}
Y_{t_{i - 1}} =& \frac{1}{1 + \overline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] - \alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}\\
& + \frac{1}{1 + \underline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D(t_{i - 1}, X_{t_{i - 1}}) < \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}. \qquad (E2)
\end{split}



```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
def blackscholes_mc(S=100, vol=0.2, r=0, q=0, ts=np.linspace(0, 1, 13), npaths=10):
    """Generate Monte-Carlo paths in Black-Scholes model.

    Parameters
    ----------
    S: scalar
        The spot price of the underlying security.
    vol: scalar
        The implied Black-Scholes volatility.
    r: scalar
        The annualized risk-free interest rate, continuously compounded.
    q: scalar
        The annualized continuous dividend yield.
    ts: array_like
        The time steps of the simulation.
    npaths: int
        The number of paths to simulate.

    Returns
    -------
    paths: ndarray
        The Monte-Carlo paths.
    """
    nsteps = len(ts) - 1
    ts = np.asarray(ts, dtype=float)[:, np.newaxis]
    dW = np.random.randn(nsteps, npaths) * np.sqrt(np.diff(ts, axis=0))
    W = np.vstack([np.zeros((1, npaths), dtype=float), np.cumsum(dW, axis=0)])
    paths = np.exp(-0.5 * vol**2 * ts + vol * W) * S * np.exp((r - q) * ts)
    return paths


def blackscholes_price(K, T, S, vol, r=0, q=0, callput='call'):
    """Compute the call/put option price in the Black-Scholes model.

    Parameters
    ----------
    K: scalar or array_like
        The strike of the option.
    T: scalar or array_like
        The maturity of the option, expressed in years.
    S: scalar or array_like
        The current price of the underlying asset.
    vol: scalar or array_like
        The implied Black-Scholes volatility.
    r: scalar or array_like
        The annualized risk-free interest rate, continuously compounded.
    q: scalar or array_like
        The annualized continuous dividend yield.
    callput: str
        Must be either 'call' or 'put'.

    Returns
    -------
    price: scalar or array_like
        The price of the option.
    """
    F = S * np.exp((r - q) * T)
    v = np.sqrt(vol**2 * T)
    d1 = np.log(F / K) / v + 0.5 * v
    d2 = d1 - v
    try:
        opttype = {'call': 1, 'put': -1}[callput.lower()]
    except KeyError:
        raise ValueError('The value of callput must be either "call" or "put".')
    price = opttype * (F * norm.cdf(opttype * d1) - K * norm.cdf(opttype * d2)) * np.exp(-r * T)
    return price


def pwlinear_basis(xknots):
    """Basis that represents a piecewise linear function with given knots."""
    fs = [lambda x: np.ones_like(x, dtype=float), lambda x: x - xknots[0]]
    fs.extend([lambda x, a=xknots[i]: np.maximum(x - a, 0.0) for i in range(len(xknots))])
    return fs


def pwlinear_fit(xdata, ydata, xknots):
    """Fit a piecewise linear function with xknots to xdata and ydata."""
    fs = pwlinear_basis(xknots)
    A = np.vstack([f(xdata) for f in fs]).T
    ps = np.linalg.lstsq(A, ydata, rcond=None)[0]
    return ps, fs


def pwlinear_eval(x, ps, fs):
    """Evaluate a piecewise linear fit given coefficients and basis functions."""
    A = np.vstack([f(x) for f in fs]).T
    return A @ ps
```

- If the death occurs at the maturity, we assume that the death comes first and the death payment $u^D$ will be made.

For constant $\lambda^D$ and independence of $\tau^D$ and $X$,
\[
\mathbb{P}(\tau^D>t+s\mid \tau^D>t)=e^{-\lambda^D s},\qquad
\mathbb{P}(\tau^D\in(t_{i-1},t_i]\mid\tau^D>t)=e^{-\lambda^D(t_{i-1}-t)}-e^{-\lambda^D(t_i-t)}.
\]
Hence, conditioning first on the death event and then on $X$,
\[
u(t,x)=\mathbb{E}_{t,x}\!\left[e^{-\lambda^D(T-t)}u^{\mathrm{mat}}(X_T)
+\int_t^T e^{-\lambda^D(s-t)}\big(\lambda^D u^D(s,X_s)-\alpha\big)ds\right].
\]
This is exactly the Feynman-Kac representation of
\[
\partial_t u+\tfrac12\sigma^2x^2\partial_{xx}u-\alpha+\lambda^D(u^D-u)=0,
\quad u(T,\cdot)=u^{\mathrm{mat}}.
\]
With monthly dates $t_i=i\Delta t$,
\[
u(0,x)=\mathbb{E}\!\left[e^{-\lambda^D T}u^{\mathrm{mat}}(X_T)
+\sum_{i=1}^{N}\big(e^{-\lambda^D t_{i-1}}-e^{-\lambda^D t_i}\big)u^D(X_{t_i})

```python
# note that lambda is a reserved keyword in python, so we use lambd instead
def exponential_samples(n, lambd):
    """Generate n samples from exponential distribution with rate lambd."""
    # inverse transform sampling
    return -np.log(np.random.rand(n)) / lambd


# Question 1: Monte Carlo pricing (deterministic mortality)
S0 = 100.0
sigma = 0.3
T = 10.0
K_mat = 90.0
K_D = 100.0
alpha = 3.0
lambd = 0.025

ts = np.linspace(0.0, T, int(np.round(12 * T)) + 1)
dt = ts[1] - ts[0]


def u_mat(x):
    return np.maximum(K_mat - x, 0.0)


def u_D(x):
    return np.maximum(K_D - x, 0.0)


def price_direct_death_mc(npaths=200000, seed=123):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, ts=ts, npaths=npaths)
    tau = exponential_samples(npaths, lambd)

    nsteps = len(ts) - 1
    alive_to_T = tau >= T

    death_idx = np.ceil(tau / dt).astype(int)
    death_idx = np.clip(death_idx, 1, nsteps)

    fee_months = np.minimum(np.floor(tau / dt).astype(int), nsteps)
    fees = alpha * dt * fee_months

    death_payoff = u_D(paths[death_idx, np.arange(npaths)])
    mat_payoff = u_mat(paths[-1])

    values = np.where(alive_to_T, mat_payoff, death_payoff) - fees
    return values.mean(), values.std(ddof=1) / np.sqrt(npaths)


def price_feynman_kac_mc(npaths=200000, seed=456):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, ts=ts, npaths=npaths)

    surv = np.exp(-lambd * ts)
    death_prob = surv[:-1] - surv[1:]

    values = surv[-1] * u_mat(paths[-1])
    values += np.sum(death_prob[:, None] * u_D(paths[1:]), axis=0)
    values -= alpha * dt * np.sum(surv[1:])

    return values.mean(), values.std(ddof=1) / np.sqrt(npaths)


q1_direct, q1_direct_se = price_direct_death_mc()
q1_fk, q1_fk_se = price_feynman_kac_mc()

print(f"Q1(a) direct death-time MC: {q1_direct:.4f}  (s.e. {q1_direct_se:.4f})")
print(f"Q1(b) FK conditional MC:   {q1_fk:.4f}  (s.e. {q1_fk_se:.4f})")
```

The conditional expectation $\mathbb E_{i-1}[Y_{t_i}]$ is approximated by a piecewise-linear regression in $X_{t_{i-1}}$.
Knots are selected from empirical quantiles of $X_{t_{i-1}}$ (1% and 99%), then spaced uniformly between these two quantiles.

From the implicit step,
\[
Y_{t_{i-1}}=\mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t
+\Lambda^D\big(t_{i-1},u^D_{i-1}-Y_{t_{i-1}}\big)(u^D_{i-1}-Y_{t_{i-1}})\Delta t,
\]
with $u^D_{i-1}=u^D(t_{i-1},X_{t_{i-1}})$.

If $u^D_{i-1}\ge Y_{t_{i-1}}$, then $\Lambda^D=\overline\lambda^D$ and
\[
(1+\overline\lambda^D\Delta t)Y_{t_{i-1}}
=\mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t+\overline\lambda^D\Delta t\,u^D_{i-1},
\]
so
\[
u^D_{i-1}-Y_{t_{i-1}}
=\frac{u^D_{i-1}-\big(\mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t\big)}{1+\overline\lambda^D\Delta t}.
\]
Hence
\[
u^D_{i-1}\ge Y_{t_{i-1}}\iff u^D_{i-1}\ge \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t.
\]
The same computation with $\underline\lambda^D$ gives the converse branch. This proves (E1).

Using (E1), the correct branch is selected by the sign of
$u^D_{i-1}-\big(\mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t\big)$,
therefore
\[
Y_{t_{i-1}}=
\frac{\mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t+\overline\lambda^D\Delta t\,u^D_{i-1}}
{1+\overline\lambda^D\Delta t}
\mathbf 1_{\{u^D_{i-1}\ge \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t\}}
+\frac{\mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t+\underline\lambda^D\Delta t\,u^D_{i-1}}
{1+\underline\lambda^D\Delta t}
\mathbf 1_{\{u^D_{i-1}< \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t\}},
\]
which is exactly (E2).

```python
# Question 2: implicit Euler BSDE with piecewise-linear regressions
lambda_min = 0.005
lambda_max = 0.04


def bsde_implicit_regression(alpha=3.0, npaths=100000, n_knots=10, seed=77):
    # backward BSDE step
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths)

    Y = u_mat(paths[-1])
    models = [None] * (len(ts) - 1)

    for i in range(len(ts) - 2, -1, -1):
        Xi = paths[i]
        x_min, x_max = np.percentile(Xi, [1, 99])
        knots = np.linspace(x_min, x_max, n_knots)

        beta, fs = pwlinear_fit(Xi, Y, knots)
        Y_next = pwlinear_eval(Xi, beta, fs)  # proxy for E_{i-1}[Y_i]

        lambda_val = np.where(u_D(Xi) >= Y_next - alpha * dt, lambda_max, lambda_min)
        f = -alpha + lambda_val * (u_D(Xi) - Y_next)  # BSDE driver made explicit

        # implicit Euler update (closed form from E1-E2)
        Y = (Y_next - alpha * dt + lambda_val * dt * u_D(Xi)) / (1.0 + lambda_val * dt)
        models[i] = (beta, fs)

    beta0, fs0 = models[0]
    Y_next0 = pwlinear_eval(np.array([S0]), beta0, fs0)[0]
    lambda0 = lambda_max if u_D(np.array([S0]))[0] >= Y_next0 - alpha * dt else lambda_min
    f0 = -alpha + lambda0 * (u_D(np.array([S0]))[0] - Y_next0)
    Y0 = (Y_next0 - alpha * dt + lambda0 * dt * u_D(np.array([S0]))[0]) / (1.0 + lambda0 * dt)

    return Y0, models


def bsde_policy_price(models, alpha=3.0, npaths=100000, seed=78):
    # independent pricing step under frozen policy
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths)

    V = u_mat(paths[-1])
    for i in range(len(ts) - 2, -1, -1):
        Xi = paths[i]
        Xi1 = paths[i + 1]

        beta, fs = models[i]
        Y_next = pwlinear_eval(Xi, beta, fs)
        lambda_val = np.where(u_D(Xi) >= Y_next - alpha * dt, lambda_max, lambda_min)
        f = -alpha + lambda_val * (u_D(Xi) - Y_next)

        p = 1.0 - np.exp(-lambda_val * dt)
        V = p * u_D(Xi1) + (1.0 - p) * V - alpha * dt

    return V.mean(), V.std(ddof=1) / np.sqrt(npaths)


def deterministic_price_constant_lambda(lambd, alpha=3.0, npaths=120000, seed=222):
    """Monthly FK representation for deterministic mortality rate lambd."""
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=npaths)

    surv = np.exp(-lambd * ts)
    death_prob = surv[:-1] - surv[1:]

    V = surv[-1] * u_mat(paths[-1])
    V += np.sum(death_prob[:, None] * u_D(paths[1:]), axis=0)
    V -= alpha * dt * np.sum(surv[1:])

    return V.mean(), V.std(ddof=1) / np.sqrt(npaths)


# (b) implicit Euler BSDE with 100,000 regression paths
q2_regression_price, q2_models = bsde_implicit_regression(alpha=alpha, npaths=100000, n_knots=10, seed=77)

# (c) independent lower bound estimate
q2_lower_bound, q2_lower_bound_se = bsde_policy_price(q2_models, alpha=alpha, npaths=100000, seed=78)

# (d) deterministic benchmarks
q2_det_min, q2_det_min_se = deterministic_price_constant_lambda(lambda_min, alpha=alpha, npaths=120000, seed=222)
q2_det_max, q2_det_max_se = deterministic_price_constant_lambda(lambda_max, alpha=alpha, npaths=120000, seed=223)

print(f"Q2(b) BSDE implicit Euler estimate:     {q2_regression_price:.4f}")
print(f"Q2(c) independent lower bound estimate: {q2_lower_bound:.4f}  (s.e. {q2_lower_bound_se:.4f})")
print(f"Q2(d) deterministic lambda_min price:   {q2_det_min:.4f}  (s.e. {q2_det_min_se:.4f})")
print(f"Q2(d) deterministic lambda_max price:   {q2_det_max:.4f}  (s.e. {q2_det_max_se:.4f})")
```

The uncertain mortality model can be interpreted as a control problem where the mortality intensity is the control variable. This is analogous to the exercise decision of an American option, which motivates a Longstaff-Schwartz-type regression scheme.

The Longstaff-Schwartz-like recursion is kept as in the provided code; only the regression basis is changed to piecewise-linear splines (10 knots, 1%-99% quantiles). The lower-bound estimate is computed with an independent second Monte Carlo run under the policy obtained from the first run.

For (b), the same two-step protocol is repeated 20 times for BSDE and for Longstaff-Schwartz-like pricing, and the empirical variance of the 20 lower-bound estimators is reported.

For (c), $\alpha^\star$ is obtained by bisection on $\alpha \mapsto u_0(\alpha)$ with the Longstaff-Schwartz-like lower-bound estimator.

```python
# Question 3: Longstaff-Schwartz-like scheme with piecewise-linear regressions

def ls_uncertain_pwlinear(alpha=3.0, n_reg=5000, n_price=50000, n_knots=10,
                         lam_lo=0.005, lam_hi=0.04, K_D_local=100.0, seed=11):
    # regression + independent pricing
    def u_D_local(x):
        return np.maximum(K_D_local - x, 0.0)

    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=n_reg)
    V = u_mat(paths[-1])
    lambd = np.where(u_D_local(paths[-1]) >= V, lam_hi, lam_lo)

    # Store one regression model per date i=1,...,N-1
    models = [None] * len(ts)
    for i in range(len(ts) - 2, 0, -1):
        p = 1.0 - np.exp(-lambd * dt)
        V = p * u_D_local(paths[i + 1]) + (1.0 - p) * V - alpha * dt

        Xi = paths[i]
        x_min, x_max = np.percentile(Xi, [1, 99])
        knots = np.linspace(x_min, x_max, n_knots)
        beta, fs = pwlinear_fit(Xi, V, knots)
        models[i] = (beta, fs)

        cont = pwlinear_eval(Xi, beta, fs)
        lambd = np.where(u_D_local(Xi) >= cont, lam_hi, lam_lo)

    np.random.seed(seed + 1)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=n_price)
    V = u_mat(paths[-1])
    lambd = np.where(u_D_local(paths[-1]) >= V, lam_hi, lam_lo)

    for i in range(len(ts) - 2, 0, -1):
        p = 1.0 - np.exp(-lambd * dt)
        V = p * u_D_local(paths[i + 1]) + (1.0 - p) * V - alpha * dt

        beta, fs = models[i]
        cont = pwlinear_eval(paths[i], beta, fs)
        lambd = np.where(u_D_local(paths[i]) >= cont, lam_hi, lam_lo)

    p0 = 1.0 - np.exp(-lambd * dt)
    V0 = p0 * u_D_local(paths[1]) + (1.0 - p0) * V - alpha * dt
    return V0.mean(), V0.std(ddof=1) / np.sqrt(n_price)


# (a) Price with piecewise-linear basis
q3_price, q3_se = ls_uncertain_pwlinear(alpha=alpha, n_reg=5000, n_price=50000,
                                        n_knots=10, lam_lo=lambda_min, lam_hi=lambda_max,
                                        K_D_local=K_D, seed=90)
print(f"Q3(a) Longstaff-Schwartz-like price: {q3_price:.4f}  (s.e. {q3_se:.4f})")


# (b) Variance comparison over 20 independent two-step runs
n_rep = 20
bsde_vals = []
ls_vals = []

for k in range(n_rep):
    _, models_bsde = bsde_implicit_regression(alpha=alpha, npaths=5000, n_knots=10, seed=1000 + 2 * k)
    y_bsde, _ = bsde_policy_price(models_bsde, alpha=alpha, npaths=50000, seed=1001 + 2 * k)
    bsde_vals.append(y_bsde)

    y_ls, _ = ls_uncertain_pwlinear(alpha=alpha, n_reg=5000, n_price=50000, n_knots=10,
                                    lam_lo=lambda_min, lam_hi=lambda_max, K_D_local=K_D,
                                    seed=2000 + 2 * k)
    ls_vals.append(y_ls)

bsde_vals = np.array(bsde_vals)
ls_vals = np.array(ls_vals)

print(f"Q3(b) empirical variance BSDE: {np.var(bsde_vals, ddof=1):.6f}")
print(f"Q3(b) empirical variance LS-like: {np.var(ls_vals, ddof=1):.6f}")


# (c) Fee alpha such that u(0,S0) ~ 0 (bisection on LS-like estimator)
def costless_fee_bisection(alpha_lo=0.0, alpha_hi=8.0, tol=5e-2, max_iter=9):
    lo, hi = alpha_lo, alpha_hi
    f_lo = ls_uncertain_pwlinear(alpha=lo, n_reg=6000, n_price=30000,
                                 n_knots=10, lam_lo=lambda_min, lam_hi=lambda_max,
                                 K_D_local=K_D, seed=300)[0]
    f_hi = ls_uncertain_pwlinear(alpha=hi, n_reg=6000, n_price=30000,
                                 n_knots=10, lam_lo=lambda_min, lam_hi=lambda_max,
                                 K_D_local=K_D, seed=301)[0]

    if f_lo * f_hi > 0:
        return np.nan

    for it in range(max_iter):
        mid = 0.5 * (lo + hi)
        f_mid = ls_uncertain_pwlinear(alpha=mid, n_reg=6000, n_price=30000,
                                      n_knots=10, lam_lo=lambda_min, lam_hi=lambda_max,
                                      K_D_local=K_D, seed=400 + it)[0]
        if abs(f_mid) < tol:
            return mid
        if f_mid > 0:
            lo = mid
        else:
            hi = mid
    return 0.5 * (lo + hi)


alpha_star = costless_fee_bisection()
print(f"Q3(c) costless fee alpha*: {alpha_star:.4f}")

print("Price (deterministic mortality) =", q1_fk)
print("Price (uncertain mortality BSDE) =", q2_lower_bound)
print("Price (Longstaff-Schwartz) =", q3_price)
```

<b style="color:darkorange">Question 4.</b> Set $K_D=0$, and all the other parameters remain the same, in particualr $\alpha=3$. Then the default event is equivalent to lapse (with no mortality payoff). It has been observed that insurance subscribers do not lapse optimally, which explains the use of the uncertain lapse model with minimum and maximum lapse rates $\underline{\lambda}$, $\overline{\lambda}$. 

<b>(a).</b> If insurance subscribers were to lapse optimally, i.e., exercise optimally their option to lapse, how would you price the contract?

<b>(b).</b> Explain how you can get this price (from optimal exercise) in the uncertain lapse model. Reuse the code provided in Question 3 to show your result (you may have to change some parameters in the code).

<b>(c).</b> Implement Longstaff-Schwartz algorithm to price the American option with payoff $u^D=0$ (Always perform a second indepedent run to get a clean lower bound price). Compare the price from uncertain lapse model in part (b) with the American price.

With $u^D\equiv 0$, lapse corresponds to stopping with immediate payoff $0$ and continuation value
\[
\mathcal C_i(x)=\mathbb E_i[V_{i+1}] - \alpha\Delta t.
\]
Optimal lapse gives the Snell envelope recursion
\[
V_i(x)=\max\{0,\mathcal C_i(x)\},\qquad V_N(x)=u^{\mathrm{mat}}(x),
\]
which is an American-style optimal stopping problem. Optimal lapse corresponds exactly to an American option with payoff $0$ at exercise and terminal payoff $u^{\mathrm{mat}}$.

In the uncertain lapse model,
\[
\partial_t u+\tfrac12\sigma^2x^2\partial_{xx}u-\alpha+\Lambda(t,-u)(-u)=0,
\]
and taking $\underline\lambda=0$, $\overline\lambda\gg 1$ enforces nearly instantaneous lapse when continuation is negative and almost no lapse when continuation is positive. This reproduces the optimal stopping value. The numerical comparison below checks this against a direct Longstaff-Schwartz American implementation.

```python
# Question 4: uncertain lapse (K_D=0) vs American optimal lapse

def american_lapse_lsm(alpha=3.0, n_reg=5000, n_price=50000, n_knots=10, seed=1234):
    # Longstaff-Schwartz for optimal lapse with immediate payoff 0
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=n_reg)
    V = u_mat(paths[-1])

    models = [None] * (len(ts) - 1)
    for i in range(len(ts) - 2, -1, -1):
        Xi = paths[i]
        x_min, x_max = np.percentile(Xi, [1, 99])
        knots = np.linspace(x_min, x_max, n_knots)
        beta, fs = pwlinear_fit(Xi, V, knots)
        cont = pwlinear_eval(Xi, beta, fs) - alpha * dt

        # If continuation is negative, optimal decision is immediate lapse.
        V = np.where(cont > 0.0, V - alpha * dt, 0.0)
        models[i] = (beta, fs)

    np.random.seed(seed + 1)
    paths = blackscholes_mc(S=S0, vol=sigma, r=0, q=0, ts=ts, npaths=n_price)
    V = u_mat(paths[-1])

    for i in range(len(ts) - 2, -1, -1):
        beta, fs = models[i]
        cont = pwlinear_eval(paths[i], beta, fs) - alpha * dt
        V = np.where(cont > 0.0, V - alpha * dt, 0.0)

    return V.mean(), V.std(ddof=1) / np.sqrt(n_price)


# (b) Uncertain lapse approximation of optimal lapse: lambda_min=0, lambda_max large
q4_uncertain, q4_uncertain_se = ls_uncertain_pwlinear(alpha=alpha, n_reg=5000, n_price=50000,
                                                       n_knots=10, lam_lo=0.0, lam_hi=20.0,
                                                       K_D_local=0.0, seed=1234)

# (c) Direct American optimal lapse with Longstaff-Schwartz
q4_american, q4_american_se = american_lapse_lsm(alpha=alpha, n_reg=5000, n_price=50000,
                                                 n_knots=10, seed=1234)

print(f"Q4(b) uncertain lapse price (lambda_min=0, lambda_max=20): {q4_uncertain:.4f}  (s.e. {q4_uncertain_se:.4f})")
print(f"Q4(c) American optimal lapse price:                    {q4_american:.4f}  (s.e. {q4_american_se:.4f})")
```
