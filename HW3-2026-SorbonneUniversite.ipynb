# Uncertain Mortality Model vs American Options

#### Re-insurance Deals

In this assignment, we study the *Uncertain Mortality Model* for the pricing of reinsurance deals. For the sake of simplicity, in this assignment we only consider one type of default: the risk of death.

We consider the following reinsurance product:

- At maturity, if the insurance subscriber is alive, the issuer delivers a put on the underlying $X$

$$
u^\textrm{mat}( x) = (K_\textrm{mat} - x)_+.
$$

- At the time of death, if it is before the maturity, the issuer delivers an exit payoff, typically another put on the underlying $X$.

$$
u^D(t, x) = (K_D - x)_+.
$$

- The subscriber pays a constant fee $α\Delta t$ at every time step until death or the maturity or the product

- The insurance sells a large number of these contracts to subscribers. We assume that the times of death of the subscribers $\tau^D$ are independent, and identically distributed, and also independent of the underlying's stock price.

- We assume that the underlying's risk neutral price dynamics is the Black-Scholes model with zero interest rate/repo/dividend yield

$$
dX_t = \sigma X_t d W_t.
$$

#### The Insurers' Approach and Risk-Neutral Pricing

This contract shows two types of risk: the times of death of the subscribers and the changes in the price of the underlying. 

In this case, the issuer can apply the insurer's approach to the risk of death times, i.e., the law of large numbers. The more people buy the contract, the less risk.

Choosing a risk-neutral measure under which the death times $\tau^D$ have the same distribution as under the historical probability measure is equivalent to applying the arbitrage-pricing approach to the financial risk insurer's rule on the risk of death. The price of the contract is then

$$
u(t, x) = \mathbb{E}^\mathbb{Q}_{t, x} \left[u^\textrm{mat}(X_T) \mathbb{1}_{\tau^D \geq T} -\int_t^T\alpha\mathbb{1}_{\tau^D>s}ds+ u^D(\tau^D, X_{\tau^D}) \mathbb{1}_{\tau^D < T} \right].
$$

#### Deterministic Death Rate

If the death intensity is a deterministic function $\lambda_t^D$ (i.e. $\tau^D$ has an exponential distribution with time-dependent intensity $\lambda_t^D$), then we have seen that $u$ is the solution to the linear PDE

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial_x^2 u -\alpha + \lambda_t^D \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat}.
\end{array}\right.
$$

#### Uncertain Mortality Model

If the death rate is uncertain, we assume that it is adapted (i.e., it does not look into the future) and belongs to a moving corridor
$\left[\underline{\lambda}_t, \overline{\lambda}_t\right]$. The most conservative way to price the contract is to compute the (financially) worst death rate process $\lambda_t^D$ as being chosen so as to maximize the value of the contract. The resulting HJB equation is

$$
\left\{\begin{array}{l}
\partial_t u + \frac{1}{2} \sigma^2 x^2 \partial^2_x u -\alpha + \Lambda^D(t, u^D - u) \cdot (u^D - u) = 0,\\
u(T, \cdot) \equiv u^\textrm{mat},
\end{array}\right.
$$
where the function $\Lambda^D$ is defined by
$$
\Lambda^D (t, y) = \left\{\begin{array}{l}
    \overline{\lambda}^D_t \quad \textrm{if} \ y \geq 0, \\
    \underline{\lambda}^D_t \quad \textrm{otherwise.}
\end{array}\right.
$$

## Link with $1$-BSDE and Numerical Schemes

From the Pardoux-Peng theorem, we know that the solution $u(0, x)$ can be represented as the solution $Y_0^x$ to the $1$-BSDE

$$
dY_t = -f(t, X_t, Y_t, Z_t) \, dt + Z_t \, dW_t,
$$

with the terminal condition $Y_T = u^\textrm{mat} (X_T)$, where $X_0 = x$ and

$$
f(t, x, y, z) = -\alpha + \Lambda^D(t, u^D(t, x) - y) \cdot (u^D(t, x) - y).
$$


## BSDE Discretization

### Explicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i + \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1} , X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ]\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \right) \Delta t_i\\
=& -\alpha\Delta t_i + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \overline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]}\\
& + \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] \left( 1 - \underline{\lambda}^D \Delta t_i \right) +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) < \mathbb{E}_{i-1}[Y_{t_i}]}.
\end{align*}

### Implicit Euler Scheme

\begin{align*}
Y_{t_n} =& \ u^{\text{mat}}\left(X_{t_n}\right)\\
Y_{t_{i - 1}} =& \ \mathbb{E}^\mathbb{Q}_{i - 1} [Y_{t_i}] -\alpha\Delta t_i+ \Lambda^D \left(t_{i - 1}, u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}}\right) \cdot \left( u^D(t_{i - 1}, X_{t_{i - 1}}) - Y_{t_{i - 1}} \right) \Delta t_i.
\end{align*}

The implicit scheme involves $Y_{t_{i - 1}}$ on both sides of the equation, which generally requires a root-finding routine to find $Y_{t_{i - 1}}$. However in this specific case, it can be shown that 

\begin{equation}u^D(t_{i-1}, X_{t_{i-1}}) \geq Y_{t_{i-1}}\quad\text{if and only if}\quad u^D(t_{i-1}, X_{t_{i-1}}) \geq \mathbb{E}^{\mathbb{Q}}_{i-1}\left[Y_{t_i}\right]-\alpha\Delta t_i. \qquad (E1) \end{equation}

Thus

\begin{split}
Y_{t_{i - 1}} =& \frac{1}{1 + \overline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] - \alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \overline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D\left(t_{i - 1}, X_{t_{i - 1}}\right) \geq \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}\\
& + \frac{1}{1 + \underline{\lambda}^D \Delta t_i} \left( \mathbb{E}^\mathbb{Q}_{i - 1} [ Y_{t_i} ] -\alpha\Delta t_i +
u^D(t_{i - 1}, X_{t_{i - 1}}) \underline{\lambda}^D \Delta t_i \right) \mathbb{1}_{u^D(t_{i - 1}, X_{t_{i - 1}}) < \mathbb{E}_{i-1}[Y_{t_i}]-\alpha\Delta t_i}. \qquad (E2)
\end{split}



```python
import numpy as np


def norm_cdf(x):
    x = np.asarray(x, dtype=float)
    k = 1.0/(1.0 + 0.2316419*np.abs(x))
    poly = (((((1.330274429*k - 1.821255978)*k) + 1.781477937)*k - 0.356563782)*k + 0.319381530)*k
    phi = np.exp(-0.5*x*x)/np.sqrt(2.0*np.pi)
    cdf = 1.0 - phi*poly
    return np.where(x >= 0.0, cdf, 1.0 - cdf)
```

```python
def blackscholes_mc(S=100, vol=0.2, r=0, q=0, ts=np.linspace(0, 1, 13), npaths=10):
    """Generate Monte-Carlo paths in Black-Scholes model.

    Parameters
    ----------
    S: scalar
        The spot price of the underlying security.
    vol: scalar
        The implied Black-Scholes volatility.
    r: scalar
        The annualized risk-free interest rate, continuously compounded.
    q: scalar
        The annualized continuous dividend yield.
    ts: array_like
        The time steps of the simualtion
    npaths: int
        the number of paths to simulate

    Returns
    -------
    paths: ndarray
        The Monte-Carlo paths.
    """
    nsteps = len(ts) - 1
    ts = np.asarray(ts, dtype=float)[:, np.newaxis]
    W = np.cumsum(np.vstack((np.zeros((1, npaths), dtype=float),
                             np.random.randn(nsteps, npaths) * np.sqrt(np.diff(ts, axis=0)))),
                  axis=0)
    paths = np.exp(-0.5*vol**2*ts + vol*W)*S*np.exp((r-q)*ts)
    return paths

def blackscholes_price(K, T, S, vol, r=0, q=0, callput='call'):
    """Compute the call/put option price in the Black-Scholes model
    
    Parameters
    ----------
    K: scalar or array_like
        The strike of the option.
    T: scalar or array_like
        The maturity of the option, expressed in years (e.g. 0.25 for 3-month and 2 for 2 years)
    S: scalar or array_like
        The current price of the underlying asset.
    vol: scalar or array_like
        The implied Black-Scholes volatility.
    r: scalar or array_like
        The annualized risk-free interest rate, continuously compounded.
    q: scalar or array_like
        The annualized continuous dividend yield.
    callput: str
        Must be either 'call' or 'put'.

    Returns
    -------
    price: scalar or array_like
        The price of the option.

    Examples
    --------
    >>> blackscholes_price(95, 0.25, 100, 0.2, r=0.05, callput='put')
    1.5342604771222823
    """
    F = S*np.exp((r-q)*T)
    v = np.sqrt(vol**2*T)
    d1 = np.log(F/K)/v + 0.5*v
    d2 = d1 - v
    try:
        opttype = {'call':1, 'put':-1}[callput.lower()]
    except:
        raise ValueError('The value of callput must be either "call" or "put".')
    price = opttype*(F*norm_cdf(opttype*d1)-K*norm_cdf(opttype*d2))*np.exp(-r*T)
    return price


def pwlinear_basis(xknots):
    """Basis that represent a piecewise linear function with given knots"""
    fs = [lambda x: np.ones_like(x, dtype=float), lambda x: x-xknots[0]]
    fs.extend([lambda x, a=xknots[i]: np.maximum(x-a, 0) for i in range(len(xknots))])
    return fs


def pwlinear_fit(xdata, ydata, xknots):
    """Fit a piecewise linear function with xknots to xdata and ydata"""
    fs = pwlinear_basis(xknots)
    A = np.vstack([f(xdata) for f in fs]).T
    ps = np.linalg.lstsq(A, ydata, rcond=None)[0]
    return ps, fs
```

<b>Note.</b> To be more specific about the payouts of the product, we make the following assumptions:

- We assume that all transactions occur at the end of the month, i.e. the insurance subsriber is expected to pay a fee of $\alpha/12$ at the end of each month; if the subscriber dies in the middle of the month, the death payment will be made at the end of that month. 

- If the death occurs at the maturity, we assume that the death comes first and the death payment $u^D$ will be made.

\widehat u_0=\frac1M\sum_{m=1}^M\Big[(K_{\mathrm{mat}}-X_T^{(m)})_+\mathbf 1_{\{\tau_m^D>T\}}+(K_D-X_{t_{N_m^D}}^{(m)})_+\mathbf 1_{\{\tau_m^D\le T\}}-\alpha\Delta t\sum_{i=1}^N\mathbf 1_{\{\tau_m^D>t_i\}}\Big].

```python
# note that lambda is a reserved keyword in python, so we use lambd instead
def exponential_samples(n, lambd):
    """Generate n samples from exponential distribution with rate lambd"""
    return -np.log(np.random.rand(n))/lambd


S0 = 100.0
vol = 0.3
T = 10.0
alpha = 3.0
lambda_D = 0.025

K_mat = 90.0
K_D = 100.0

ts = np.linspace(0.0, T, int(np.round(12*T)) + 1)
dt = ts[1] - ts[0]


def u_mat(x):
    return np.maximum(K_mat - x, 0.0)


def u_D(x):
    return np.maximum(K_D - x, 0.0)


def q1_direct_mc(npaths=200000, seed=0):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths)
    tau = exponential_samples(npaths, lambda_D)

    nsteps = len(ts) - 1
    idx_death = np.minimum(np.ceil(tau/dt).astype(int), nsteps)
    dead = tau <= T

    payoff = np.empty(npaths)
    payoff[~dead] = u_mat(paths[-1, ~dead])
    j = np.where(dead)[0]
    payoff[j] = u_D(paths[idx_death[j], j])

    n_fees = np.minimum(np.floor(tau/dt).astype(int), nsteps)
    payoff -= alpha*dt*n_fees

    return np.mean(payoff), np.std(payoff, ddof=1)/np.sqrt(npaths)


def q1_conditional_mc(npaths=200000, seed=1):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths)

    surv = np.exp(-lambda_D*ts)
    death_prob = surv[:-1] - surv[1:]

    death_leg = np.tensordot(death_prob, u_D(paths[1:]), axes=(0, 0))
    fee_leg = alpha*dt*np.sum(surv[1:])

    payoff = surv[-1]*u_mat(paths[-1]) + death_leg - fee_leg
    return np.mean(payoff), np.std(payoff, ddof=1)/np.sqrt(npaths)


price_q1a, se_q1a = q1_direct_mc()
price_q1b, se_q1b = q1_conditional_mc()

print(f"Q1(a) direct death-time simulation: {price_q1a:.4f}  (SE {se_q1a:.4f})")
print(f"Q1(b) conditional/Feynman-Kac MC:   {price_q1b:.4f}  (SE {se_q1b:.4f})")
```

Let $m_{i-1}=\mathbb E^{\mathbb Q}_{i-1}[Y_{t_i}]$, $u_{i-1}=u^D(t_{i-1},X_{t_{i-1}})$ and $\Delta t_i=\Delta t$. The implicit step is
$$
Y_{t_{i-1}}=m_{i-1}-\alpha\Delta t+\Lambda^D\!\left(t_{i-1},u_{i-1}-Y_{t_{i-1}}\right)\left(u_{i-1}-Y_{t_{i-1}}\right)\Delta t.
$$
Hence
$$
\big(1+\Lambda^D\Delta t\big)\big(u_{i-1}-Y_{t_{i-1}}\big)=u_{i-1}-m_{i-1}+\alpha\Delta t.
$$
Since $1+\Lambda^D\Delta t>0$, both sides have the same sign, so
$$
u_{i-1}\ge Y_{t_{i-1}}\iff u_{i-1}\ge m_{i-1}-\alpha\Delta t,
$$
which is (E1).

Using (E1), if $u_{i-1}\ge m_{i-1}-\alpha\Delta t$, then $\Lambda^D=\overline\lambda^D$ and
$$
Y_{t_{i-1}}=\frac{m_{i-1}-\alpha\Delta t+\overline\lambda^D\Delta t\,u_{i-1}}{1+\overline\lambda^D\Delta t}.
$$
Otherwise $\Lambda^D=\underline\lambda^D$ and
$$
Y_{t_{i-1}}=\frac{m_{i-1}-\alpha\Delta t+\underline\lambda^D\Delta t\,u_{i-1}}{1+\underline\lambda^D\Delta t}.
$$
This is exactly (E2).

```python
lambda_min = 0.005
lambda_max = 0.04


def fit_pwlinear_model(x, y, nknots=10):
    lo, hi = np.percentile(x, [1, 99])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or hi <= lo + 1e-10:
        return {"kind": "const", "c": float(np.mean(y))}
    knots = np.linspace(lo, hi, nknots)
    ps, _ = pwlinear_fit(x, y, knots)
    return {"kind": "pwlin", "ps": ps, "knots": knots}


def predict_model(model, x):
    if model["kind"] == "const":
        return np.full_like(x, model["c"], dtype=float)
    knots = model["knots"]
    cols = [np.ones_like(x, dtype=float), x - knots[0]]
    cols.extend([np.maximum(x - a, 0.0) for a in knots])
    A = np.vstack(cols)
    return model["ps"] @ A


def bsde_implicit_regression(npaths=100000, seed=123, nknots=10):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths)
    nsteps = len(ts) - 1

    Y = u_mat(paths[-1])
    models = [None] * nsteps

    for i in range(nsteps - 1, 0, -1):
        x = paths[i]
        model = fit_pwlinear_model(x, Y, nknots=nknots)
        m = predict_model(model, x)
        ud = u_D(x)

        exercise_max = ud >= m - alpha*dt
        Y = np.where(
            exercise_max,
            (m - alpha*dt + lambda_max*dt*ud) / (1.0 + lambda_max*dt),
            (m - alpha*dt + lambda_min*dt*ud) / (1.0 + lambda_min*dt),
        )
        models[i] = model

    m0 = float(np.mean(Y))
    models[0] = {"kind": "const", "c": m0}
    ud0 = u_D(paths[0])
    exercise_max0 = ud0 >= m0 - alpha*dt
    Y0 = np.where(
        exercise_max0,
        (m0 - alpha*dt + lambda_max*dt*ud0) / (1.0 + lambda_max*dt),
        (m0 - alpha*dt + lambda_min*dt*ud0) / (1.0 + lambda_min*dt),
    )
    return float(np.mean(Y0)), models


def bsde_lower_bound(models, npaths=100000, seed=456):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths)
    nsteps = len(ts) - 1

    V = u_mat(paths[-1])
    for i in range(nsteps - 1, -1, -1):
        x = paths[i]
        m = predict_model(models[i], x)
        lambd = np.where(u_D(x) >= m - alpha*dt, lambda_max, lambda_min)
        p = 1.0 - np.exp(-lambd*dt)
        V = p*u_D(paths[i + 1]) + (1.0 - p)*(V - alpha*dt)

    return float(np.mean(V)), float(np.std(V, ddof=1) / np.sqrt(npaths))


def deterministic_conditional_price(lambd, npaths=100000, seed=777):
    np.random.seed(seed)
    paths = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=npaths)

    surv = np.exp(-lambd*ts)
    death_prob = surv[:-1] - surv[1:]

    death_leg = np.tensordot(death_prob, u_D(paths[1:]), axes=(0, 0))
    fee_leg = alpha*dt*np.sum(surv[1:])

    payoff = surv[-1]*u_mat(paths[-1]) + death_leg - fee_leg
    return float(np.mean(payoff)), float(np.std(payoff, ddof=1) / np.sqrt(npaths))


price_bsde_reg, models_bsde = bsde_implicit_regression(npaths=100000, seed=123, nknots=10)
price_bsde_lb, se_bsde_lb = bsde_lower_bound(models_bsde, npaths=100000, seed=456)
price_det_min, se_det_min = deterministic_conditional_price(lambda_min, npaths=100000, seed=333)
price_det_max, se_det_max = deterministic_conditional_price(lambda_max, npaths=100000, seed=444)

print(f"Q2(b) implicit BSDE (regression run): {price_bsde_reg:.4f}")
print(f"Q2(c) lower bound (independent run): {price_bsde_lb:.4f}  (SE {se_bsde_lb:.4f})")
print(f"Q2(d) deterministic lambda_min={lambda_min:.3f}: {price_det_min:.4f}  (SE {se_det_min:.4f})")
print(f"Q2(d) deterministic lambda_max={lambda_max:.3f}: {price_det_max:.4f}  (SE {se_det_max:.4f})")
print("As expected: deterministic lambda_min <= uncertain model lower bound <= deterministic lambda_max.")
```

<b>(c).</b> What is approximately the value of the fee $\alpha$ that makes the deal costless at inception?

At time $t_i$, denote by $C_i(x)$ the estimated continuation value. The adverse intensity is selected by
$$
\hat\lambda_i(x)=\begin{cases}
\overline\lambda^D,& u^D(x)\ge C_i(x),\\
\underline\lambda^D,& u^D(x)<C_i(x),
\end{cases}
$$
and with $p_i(x)=1-e^{-\hat\lambda_i(x)\Delta t}$ the one-step update is
$$
V_i=p_i(X_i)u^D(X_{i+1})+\big(1-p_i(X_i)\big)(V_{i+1}-\alpha\Delta t).
$$
I use piecewise-linear regression (10 knots, percentiles 1 and 99) both in the BSDE and Longstaff-Schwartz-like procedures, then compare empirical variances over 20 independent two-step runs.

```python
def lsm_two_step(alpha_value=3.0, K_D_value=100.0, lambda_lo=0.005, lambda_hi=0.04,
                 n_reg=5000, n_prc=50000, seed_reg=0, seed_prc=1, nknots=10):
    def u_D_loc(x):
        return np.maximum(K_D_value - x, 0.0)

    nsteps = len(ts) - 1

    np.random.seed(seed_reg)
    paths_reg = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_reg)
    V = u_mat(paths_reg[-1])
    models = [None] * nsteps

    for i in range(nsteps - 1, -1, -1):
        x = paths_reg[i]
        model = fit_pwlinear_model(x, V, nknots=nknots)
        m = predict_model(model, x)
        lambd = np.where(u_D_loc(x) >= m, lambda_hi, lambda_lo)
        p = 1.0 - np.exp(-lambd*dt)
        V = p*u_D_loc(paths_reg[i + 1]) + (1.0 - p)*(V - alpha_value*dt)
        models[i] = model

    np.random.seed(seed_prc)
    paths_prc = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_prc)
    V = u_mat(paths_prc[-1])

    for i in range(nsteps - 1, -1, -1):
        x = paths_prc[i]
        m = predict_model(models[i], x)
        lambd = np.where(u_D_loc(x) >= m, lambda_hi, lambda_lo)
        p = 1.0 - np.exp(-lambd*dt)
        V = p*u_D_loc(paths_prc[i + 1]) + (1.0 - p)*(V - alpha_value*dt)

    return float(np.mean(V)), float(np.std(V, ddof=1)/np.sqrt(n_prc))


def bsde_two_step_once(seed_reg=0, seed_prc=1):
    _, models = bsde_implicit_regression(npaths=5000, seed=seed_reg, nknots=10)
    price, _ = bsde_lower_bound(models, npaths=50000, seed=seed_prc)
    return price


price_ls, se_ls = lsm_two_step(alpha_value=alpha, K_D_value=K_D, lambda_lo=lambda_min, lambda_hi=lambda_max,
                               n_reg=5000, n_prc=50000, seed_reg=2026, seed_prc=2027)
print(f"Q3(a) piecewise-linear Longstaff-Schwartz lower bound: {price_ls:.4f}  (SE {se_ls:.4f})")


nrep = 20
bsde_vals = np.empty(nrep)
lsm_vals = np.empty(nrep)
for k in range(nrep):
    bsde_vals[k] = bsde_two_step_once(seed_reg=1000 + 2*k, seed_prc=2000 + 2*k)
    lsm_vals[k], _ = lsm_two_step(alpha_value=alpha, K_D_value=K_D, lambda_lo=lambda_min, lambda_hi=lambda_max,
                                  n_reg=5000, n_prc=50000, seed_reg=3000 + 2*k, seed_prc=4000 + 2*k)

print(f"Q3(b) BSDE mean={np.mean(bsde_vals):.4f}, variance={np.var(bsde_vals, ddof=1):.6f}")
print(f"Q3(b) LSM  mean={np.mean(lsm_vals):.4f}, variance={np.var(lsm_vals, ddof=1):.6f}")


def estimate_value_alpha(alpha_value):
    val, _ = lsm_two_step(alpha_value=alpha_value, K_D_value=K_D, lambda_lo=lambda_min, lambda_hi=lambda_max,
                          n_reg=7000, n_prc=70000, seed_reg=111, seed_prc=222)
    return val


alpha_grid = np.linspace(2.0, 6.0, 9)
value_grid = np.array([estimate_value_alpha(a) for a in alpha_grid])

idx = np.where(np.sign(value_grid[:-1]) != np.sign(value_grid[1:]))[0]
if len(idx) > 0:
    j = idx[0]
    a0, a1 = alpha_grid[j], alpha_grid[j + 1]
    v0, v1 = value_grid[j], value_grid[j + 1]
    alpha_star = a0 - v0*(a1 - a0)/(v1 - v0)
else:
    alpha_star = np.nan

print("Q3(c) values on alpha grid:")
for a, v in zip(alpha_grid, value_grid):
    print(f"  alpha={a:.2f} -> value={v:.4f}")
print(f"Approximate costless fee alpha*: {alpha_star:.4f}")
```

<b style="color:darkorange">Question 4.</b> Set $K_D=0$, and all the other parameters remain the same, in particualr $\alpha=3$. Then the default event is equivalent to lapse (with no mortality payoff). It has been observed that insurance subscribers do not lapse optimally, which explains the use of the uncertain lapse model with minimum and maximum lapse rates $\underline{\lambda}$, $\overline{\lambda}$. 

<b>(a).</b> If insurance subscribers were to lapse optimally, i.e., exercise optimally their option to lapse, how would you price the contract?

<b>(b).</b> Explain how you can get this price (from optimal exercise) in the uncertain lapse model. Reuse the code provided in Question 3 to show your result (you may have to change some parameters in the code).

<b>(c).</b> Implement Longstaff-Schwartz algorithm to price the American option with payoff $u^D=0$ (Always perform a second indepedent run to get a clean lower bound price). Compare the price from uncertain lapse model in part (b) with the American price.

With $u^D\equiv 0$, optimal lapse is an optimal stopping problem with immediate exercise payoff $0$ and continuation value
$$
\mathcal C_i=\mathbb E_i[V_{i+1}]-\alpha\Delta t.
$$
So the dynamic programming step is
$$
V_i=\max\{0,\mathcal C_i\},\qquad V_T=u^{\mathrm{mat}}(X_T).
$$
This is an American-style contract (obstacle $0$).

In the uncertain-lapse model,
$$
\partial_t u+\tfrac12\sigma^2x^2\partial_{xx}u-\alpha+\Lambda(t,-u)(-u)=0,
$$
with $\Lambda\in[\underline\lambda,\overline\lambda]$. Taking $\underline\lambda=0$ and $\overline\lambda\to\infty$ enforces immediate lapse when continuation is negative, hence it converges to the optimal stopping value above.

```python
def american_lapse_two_step(alpha_value=3.0, n_reg=5000, n_prc=50000,
                            seed_reg=0, seed_prc=1, nknots=10):
    nsteps = len(ts) - 1

    np.random.seed(seed_reg)
    paths_reg = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_reg)
    V = u_mat(paths_reg[-1])
    models = [None] * nsteps

    for i in range(nsteps - 1, -1, -1):
        x = paths_reg[i]
        target = V - alpha_value*dt
        model = fit_pwlinear_model(x, target, nknots=nknots)
        cont = predict_model(model, x)
        V = np.where(cont > 0.0, target, 0.0)
        models[i] = model

    np.random.seed(seed_prc)
    paths_prc = blackscholes_mc(S=S0, vol=vol, r=0, q=0, ts=ts, npaths=n_prc)
    V = u_mat(paths_prc[-1])

    for i in range(nsteps - 1, -1, -1):
        x = paths_prc[i]
        target = V - alpha_value*dt
        cont = predict_model(models[i], x)
        V = np.where(cont > 0.0, target, 0.0)

    return float(np.mean(V)), float(np.std(V, ddof=1)/np.sqrt(n_prc))


lambda_hi_grid = [20.0, 50.0, 100.0, 200.0]
values_uncertain_lapse = []
for lam_hi in lambda_hi_grid:
    v, _ = lsm_two_step(alpha_value=alpha, K_D_value=0.0, lambda_lo=0.0, lambda_hi=lam_hi,
                        n_reg=5000, n_prc=50000, seed_reg=501, seed_prc=502)
    values_uncertain_lapse.append(v)

price_uncertain_lapse, se_uncertain_lapse = lsm_two_step(alpha_value=alpha, K_D_value=0.0,
                                                         lambda_lo=0.0, lambda_hi=200.0,
                                                         n_reg=5000, n_prc=50000,
                                                         seed_reg=501, seed_prc=502)

price_american, se_american = american_lapse_two_step(alpha_value=alpha, n_reg=5000, n_prc=50000,
                                                       seed_reg=801, seed_prc=802, nknots=10)

print("Q4(b) uncertain lapse values (lambda_min=0):")
for lam_hi, val in zip(lambda_hi_grid, values_uncertain_lapse):
    print(f"  lambda_max={lam_hi:6.1f} -> value={val:.4f}")
print(f"Q4(b) proxy for lambda_max -> +infty (lambda_max=200): {price_uncertain_lapse:.4f} (SE {se_uncertain_lapse:.4f})")
print(f"Q4(c) American (optimal lapse) LSM lower bound:           {price_american:.4f} (SE {se_american:.4f})")
print(f"Difference (uncertain-lapse proxy - American): {price_uncertain_lapse - price_american:.4f}")
```
